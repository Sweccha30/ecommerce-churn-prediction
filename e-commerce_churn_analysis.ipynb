### **I.1 Context**

In the highly competitive e-commerce industry, retaining existing customers has become a top strategic challenge. One of the most critical concerns is **customer churn**, which refers to the situation when users stop using the service or become inactive.

According to Verbraken et al. (2014), **churn** is defined as a condition where customers explicitly or implicitly end their relationship with a company ([Verbraken et al., 2014](https://www.researchgate.net/publication/261296259_A_Novel_Profit_Maximizing_Metric_for_Measuring_Classification_Performance_of_Customer_Churn_Prediction_Models)). This definition provides a strong academic foundation for understanding churn from a data science perspective.

High churn rates can lead to:

* Loss of recurring revenue from loyal users
* Increased Customer Acquisition Cost (CAC)
* Decrease in Customer Lifetime Value (CLV), which refers to the total economic value a customer generates over their lifecycle

A widely cited article by **Harvard Business Review** highlights that acquiring new customers can be **5 to 25 times more expensive** than retaining existing ones. It also emphasizes that loyal customers are not only more profitable, but often become brand advocates ([HBR, 2014](https://hbr.org/2014/10/the-value-of-keeping-the-right-customers)).

In addition, a report by **McKinsey & Company** reveals that customers who receive the **best experiences** contribute up to **40% more revenue** than those who receive poor experiences ([McKinsey, 2022](https://www.mckinsey.com/capabilities/growth-marketing-and-sales/our-insights/the-value-of-getting-personalization-right-or-wrong-is-multiplying)).

In this project, we analyze historical data from an e-commerce platform to identify the characteristics and behaviors associated with customer churn. Using a machine learning-based predictive approach, this model is intended to support various stakeholders in designing **targeted and effective retention strategies**.

#### Key Stakeholders

| Stakeholder              | Purpose / Use of the Model                              |
| ------------------------ | ------------------------------------------------------- |
| Customer Experience Team | Improve engagement and design loyalty strategies        |
| Marketing Team           | Launch targeted retention campaigns for high-risk users |
| Product Team             | Identify product-related churn drivers                  |
| Executive Management     | Make strategic decisions based on churn insights        |
| Data Scientists          | Build accurate, interpretable churn prediction models   |

### **I.2 Problem Statement**

In a saturated and highly competitive e-commerce market, maintaining customer engagement has become one of the most pressing business challenges. The company is experiencing a rising concern: **a substantial number of customers have become inactive or stopped transacting altogether**. This phenomenon, known as **customer churn**, poses serious risks to revenue continuity, marketing effectiveness, and long-term business sustainability.

Several strategic business questions arise:

* What types of customers are most likely to churn?
* What behavioral patterns tend to precede customer churn?
* Can we proactively identify churn-prone customers before it's too late?

According to Li & Wu (2023), early identification of customers at risk of churning can reduce Customer Acquisition Cost (CAC) and increase marketing ROI by up to 50% ([ResearchGate, 2023](https://www.researchgate.net/publication/391629505_Research_on_Customer_Retention_Strategy_in_the_E-commerce_Environment)).

To address this issue, we reformulate the problem into a **binary classification task** using machine learning. The predictive model is designed to estimate:

> Whether a customer will churn (1) or remain (0), based on historical behavioral features.

Such predictions empower key stakeholders to act before churn happens — through loyalty initiatives, targeted outreach, or experience enhancement.

Without a predictive churn model, the business risks:

* Wasting marketing budget on disengaged customers
* Losing high-value customers who could have been retained
* Strategic misalignment due to poor visibility into churn drivers

Thus, churn prediction is not just a data science problem — it is a **business necessity** for optimizing retention, improving efficiency, and sustaining profitability.

<a id="approach"></a>

### **I.4 Analytical Approach**

This project applies a **supervised machine learning classification** framework to transform a real-world business question into a **data-driven predictive solution**. The overall pipeline is designed not only for accuracy, but also for practical **retention impact** and **business decision support**.

#### **Problem Reformulation: From Business to ML Task**

**Business Question**:

> “How can the company proactively identify customers at high risk of churn to reduce revenue loss through targeted intervention?”

**ML Task Reformulation**:

> *Develop a binary classification model that predicts whether a customer will churn (`1`) or stay (`0`) based on their historical behavior, preferences, and transactional patterns.*

#### **Evaluation Strategy & Metric Prioritization**

In churn prediction, both **false negatives (FN)** and **false positives (FP)** have implications—but their consequences differ significantly in business terms:

| **Error Type**     | **Definition**                    | **Business Impact**                                                                      |
| ------------------ | --------------------------------- | ---------------------------------------------------------------------------------------- |
| **False Negative** | Churner predicted as non-churn    | **Most critical**: Undetected churn leads to lost revenue without opportunity to retain. |
| **False Positive** | Loyal customer predicted as churn | Costly but tolerable: unnecessary retention effort, but no customer lost.                |

Hence, this project emphasizes minimizing **false negatives**, reflecting the high cost of churn loss. To balance that objective with operational efficiency, the **F2-score** is used as the primary metric, followed by **Recall** and **PR-AUC**.

#### **Metric Prioritization**

| **Metric**    | **Purpose & Use Case**                                                   | **Notes & Limitations**                                                                |
| ------------- | ------------------------------------------------------------------------ | -------------------------------------------------------------------------------------- |
| **F2-Score**  | Prioritizes recall while still penalizing excess false positives         | Primary business-aligned metric for churn minimization                                 |
| **Recall**    | Sensitivity to actual churners (**TP / \[TP + FN]**)                     | Crucial to avoid losing customers due to missed detection                              |
| **PR-AUC**    | Captures model performance on the minority churn class across thresholds | Better suited than ROC-AUC for imbalanced churn prediction                             |
| **F1-Score**  | Balanced harmonic mean of precision and recall (equal weight)            | Secondary metric for balanced quality overview                                         |
| **Precision** | Fraction of predicted churners who are truly churners                    | Helps monitor outreach efficiency, though slightly deprioritized                       |
| **ROC-AUC**   | True vs. false positive trade-off across thresholds                      | Informative for global separability but less useful in imbalanced churn classification |
| **Accuracy**  | Overall correctness                                                      | Not a reliable indicator for churn due to class imbalance                              |

> **Metric Priority in This Project**:
> **F2-Score > Recall > PR-AUC > F1-Score > Precision > ROC-AUC > Accuracy**

#### **Confusion Matrix Interpretation**

|                           | Predicted: Not Churn (0) | Predicted: Churn (1) |
| ------------------------- | ------------------------ | -------------------- |
| **Actual: Not Churn (0)** | True Negative (TN)       | False Positive (FP)  |
| **Actual: Churn (1)**     | False Negative (FN)      | True Positive (TP)   |

* **True Positives (TP)** → Churners successfully detected → allows early, targeted action
* **False Negatives (FN)** → Churners missed → direct revenue loss, highest business risk
* **False Positives (FP)** → Loyal users flagged incorrectly → adds retention overhead
* **True Negatives (TN)** → Loyal users correctly ignored → no cost, no concern

#### **Business Expectation from the Model**

For the churn prediction model to be valuable in real-world application, it must:

* Detect **at least 90% of churners** (Recall ≥ 0.90)
* Maintain **F2-score ≥ 0.89** as the operational threshold
* Limit false positives to preserve retention budget
* Provide **interpretable outputs** for campaign design and CRM integration
* Be **deployable and scalable** with explainable, production-ready output

In summary, this approach ensures the model is both **statistically sound** and **business-actionable**, supporting proactive churn intervention, loyalty optimization, and customer lifetime value retention.

### **I.5 Scope and Limitation**

#### **1. Project Scope**

This project focuses on developing a **customer churn prediction model** using a supervised machine learning classification approach. The project scope includes:

- Analyzing historical customer behavior and transaction data.
- Exploring features that contribute to churn behavior.
- Building and evaluating a churn prediction model.
- Providing strategic business recommendations based on model output.

The model will serve as a **decision-support tool** for more targeted and efficient retention strategies, especially in marketing campaigns and loyalty program management.

#### **2. Project Limitations**

This project has several limitations, including:

* **Data Limitations**:
  * External factors (e.g., competitors, macroeconomic conditions, offline promotions) are not captured in the dataset.
  * No timestamp or time-series data available, limiting temporal analysis.
  * It is not clearly stated how churn is defined — whether by transaction inactivity, app login behavior, or total disengagement — nor the exact time period that qualifies a customer as churned.

* **Unclear Units or Meaning of Key Features**

| Feature           | Clarification Based on Team Discussion                                                                                                                                                                                                                          |
| ----------------- | --------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| `Tenure`          | Assumed to be in **months**, based on data range and relevance to e-commerce context.                                                                                                                                                                           |
| `WarehouseToHome` | Assumed to be in **kilometers (km)** for standardization and consistency.                                                                                                                                                                                       |
| `HourSpendOnApp`  | Interpreted as **weekly app usage (per week)** — supported by external studies showing average internet usage is \~3–5 hours/day across all platforms. For reference, see [DataReportal](https://datareportal.com/reports/digital-2023-global-overview-report). |
| `NumberOfAddress` | Refers to **the number of delivery addresses registered** in the user's account.                                                                                                                                                                                |
| `CityTier`        | Classification method for Tier 1, 2, 3 is not explicitly defined in the dataset.                                                                                                                                                                                |
| `Country`         | Country information is not provided — dataset is treated as anonymized global data.                                                                                                                                                                             |

* **Class Imbalance**

  * Only \~16.8% of customers are labeled as churn.
  * Handled via `class_weight`, `threshold tuning`, with plans for SMOTE/undersampling.

* **Model Limitations**:

  * The model cannot explain *why* a customer churns, only indicates the likelihood based on patterns.
  * Prediction quality highly depends on the input data distribution and quality.
  * The model is predictive, not prescriptive — it doesn’t suggest specific retention actions.

Despite these limitations, the model serves as a **foundational framework** to assess churn risk and enhance retention strategies using data-driven insights.

### **II.1 Data Understanding**

This dataset represents customer records from an e-commerce platform, including demographic details, app usage behavior, and transaction history. Each row corresponds to a unique customer, and each column captures specific information relevant to understanding churn.

The main goal of this dataset is to predict whether a customer has churned (`1`) or stayed (`0`), based on their profile and activity history. With features covering login devices, payment preferences, satisfaction scores, and order behavior, this dataset enables the exploration of what drives customer loyalty and churn in digital commerce.

This historical snapshot dataset can help businesses:

* Identify high-risk churn segments
* Understand churn-related behavioral patterns
* Develop proactive retention strategies

#### Dataset Information

* **Dataset Name**: Ecommerce Customer Churn Analysis and Prediction
* **Author**: Ankit Verma
* **Published**: Circa 2021
* **Number of Records**: 5,630 rows
* **Number of Features**: 20 columns
* **Source**: [Kaggle - E-Commerce Churn Dataset](https://www.kaggle.com/datasets/ankitverma2010/ecommerce-customer-churn-analysis-and-prediction)
* **General Description**: This dataset captures behavioral and profile data from e-commerce platform users, including login behavior, order history, and churn status (whether a customer left the platform).

#### Dataset Structure

Each row represents a unique customer, and each column provides a feature related to customer attributes or behavioral metrics.

### Feature Description Table

| Feature Name                  | Data Type           | Description                                                                                                  |
| ----------------------------- | ------------------- | ------------------------------------------------------------------------------------------------------------ |
| `CustomerID`                  | Numeric (int64)     | Unique ID for each customer.                                                                                 |
| `Churn`                       | Numeric (int64)     | Target variable: `1` means churned, `0` means retained.                                                      |
| `Tenure`                      | Numeric (float64)   | Duration of being a customer. Assumed to be in **months**, based on data range (0–61), which is more plausible than days or years.                                      |
| `PreferredLoginDevice`        | Categorical         | Most frequently used device to log in: `Mobile Phone`, `Phone`, `Computer`.                                  |
| `CityTier`                    | Categorical (int64) | Tier of the customer’s city: `1`, `2`, or `3`. Definitions are not provided, assumed to reflect urbanization/economic level.                      |
| `WarehouseToHome`             | Numeric (float64)   | Distance from warehouse to customer’s home, assumed to be in **kilometers (KM)** based on value range (5–127).                               |
| `PreferredPaymentMode`        | Categorical         | Commonly used payment mode: `Debit Card`, `UPI`, `Credit Card`, `Cash on Delivery`, `E-wallet`, `COD`, `CC`. |
| `Gender`                      | Categorical         | Customer gender: `Male` or `Female`.                                                                         |
| `HourSpendOnApp`              | Numeric (float64)   | Average time spent on the app, assumed to be in **hours per week** based on the dataset’s mean (\~3 hours) and comparisons with global internet usage data. If it were per day, the value would be implausibly low. [Reference: Datareportal 2023](https://datareportal.com/reports/digital-2023-global-overview-report)                                                                 |
| `NumberOfDeviceRegistered`    | Numeric (int64)     | Number of devices registered to the customer’s account.                                                      |
| `PreferedOrderCat`            | Categorical         | ost frequently ordered product category **last month**: `Laptop & Accessory`, `Mobile`, `Mobile Phone`, `Others`, `Fashion`, `Grocery`.  |
| `SatisfactionScore`           | Numeric (int64)     | Customer satisfaction score (scale: 1 to 5).                                                                        |
| `MaritalStatus`               | Categorical         | Customer’s marital status: `Single`, `Married`, `Divorced`.                                                  |
| `NumberOfAddress`             | Numeric (int64)     | Number of **shipping addresses registered in the app**. It refers to total saved addresses, not house number or address edits.                                                                   |
| `Complain`                    | Numeric (int64)     | `1` if the customer filed a complaint **in the last month**, otherwise `0`.                                                    |
| `OrderAmountHikeFromlastYear` | Numeric (float64)   | Percentage increase in order amount **compared to the previous year**.                                                              |
| `CouponUsed`                  | Numeric (float64)   | Number of coupons used by the customer **in the last month**.                                                                                      |
| `OrderCount`                  | Numeric (float64)   | Total number of orders placed by the customer.                                                               |
| `DaySinceLastOrder`           | Numeric (float64)   | Days since the last order was placed.                                                                        |
| `CashbackAmount`              | Numeric (int64)     | Total cashback received by the customer.                                                                     |

### Dataset Limitations and Notes

* **No Timestamp or Date Field**
  → Prevents time-based analysis such as trend detection or recency-based segmentation.

* **Unspecified Units for Some Features**
  → The dataset lacks unit descriptions for several numeric fields. Based on team discussion, assumptions are made:

  * `Tenure`: interpreted in **months**
  * `WarehouseToHome`: interpreted in **kilometers (KM)**
  * `HourSpendOnApp`: interpreted in **hours per week**
  * `NumberOfAddress`: refers to number of **registered delivery addresses** in the app  

* **Churn Definition is Not Clearly Stated**
  → The exact criteria used to label a customer as churned is unclear. It may be based on lack of app login, inactivity in transactions, or both—but no definitive rule is provided in the dataset.

* **Missing Values**
  → Several numeric columns contain \~4.5% to 5.5% missing values, which will be handled during the data cleaning stage.

* **Class Imbalance**
  → Target class distribution is imbalanced:

  * Churned (`1`): 16.8%
  * Retained (`0`): 83.2%
    → Class imbalance will be addressed using techniques such as **resampling**, **class weighting**, or **threshold tuning** during modeling.

#### **Comparison and Total Count of the Target variable (`Churn`)**

````python
# Configure visual style
sns.set_style("whitegrid")
plt.rcParams.update({
    'font.weight': 'bold',
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold'
})

# Calculate churn metrics
churn_counts = df['Churn'].value_counts()
churn_percentages = df['Churn'].value_counts(normalize=True) * 100
total_customers = len(df)
churn_rate = churn_percentages[1]

# Summary statistics
print("="*60)
print(f"{'CHURN RATE OVERVIEW':^60}")
print("="*60)
print(f"\nTotal Customers       : {total_customers:,}")
print(f"Churned Customers     : {churn_counts[1]:,} ({churn_rate:.1f}%)")
print(f"Retained Customers    : {churn_counts[0]:,} ({churn_percentages[0]:.1f}%)\n")

dist_df = pd.DataFrame({
    'Customer Status': ['Retained', 'Churned'],
    'Count': churn_counts.values,
    'Percentage': churn_percentages.round(1).values
})
print(dist_df.to_markdown(index=False, tablefmt="pretty"))
print("\n" + "="*60)

# Plot visualization
fig, axs = plt.subplots(1, 2, figsize=(14, 6))

# Donut chart (percentage)
wedges, texts, autotexts = axs[0].pie(
    churn_counts,
    labels=['Retained', 'Churned'],
    autopct='%1.1f%%',
    startangle=90,
    colors=['#6EA4FF','#FFB3B3'],  # Soft professional colors
    explode=(0, 0.05),
    wedgeprops={'width': 0.3, 'edgecolor': 'white', 'linewidth': 2, 'antialiased': True},
    shadow=True
)
plt.setp(autotexts, size=12, weight='bold', color='black')
plt.setp(texts, size=12, weight='bold')
axs[0].set_title('Customer Churn Breakdown (%)', fontsize=13, pad=20)

# Bar chart (absolute count)
ax = axs[1]
sns.barplot(
    x=churn_counts.index,
    y=churn_counts.values,
    palette=['#6EA4FF','#FFB3B3'],
    saturation=0.9,
    ax=ax
)

# Format bars and add labels
for bar in ax.patches:
    bar.set_edgecolor('gray')
    bar.set_linewidth(1.5)

for i, (count, pct) in enumerate(zip(churn_counts, churn_percentages)):
    ax.text(i, count/2,
            f"{count:,}\n({pct:.1f}%)",
            ha='center',
            va='center',
            color='black',
            weight='bold',
            size=12)

ax.set_title('Customer Churn Breakdown (Count)', fontsize=13, pad=20)
ax.set_xlabel('Customer Status')
ax.set_ylabel('Number of Customers')
ax.set_xticks([0, 1])
ax.set_xticklabels(['Retained (0)', 'Churned (1)'])
ax.grid(axis='y', linestyle=':', alpha=0.5)

plt.suptitle('Churn Rate Distribution', y=1.05, fontsize=15, weight='bold')
plt.tight_layout(pad=3)
plt.show()

# Analytical insights
print("ANALYTICAL INSIGHTS")
print(f"- Current churn rate is {churn_rate:.1f}%, indicating a notable level of customer loss risk.")
print(f"- Class imbalance is present with a ratio of approximately {churn_counts[0]/churn_counts[1]:.1f}:1 (retained vs churned).")

print("\nTECHNICAL RECOMMENDATIONS (MODELING):")
print("- Apply imbalance-handling techniques such as SMOTE, class weighting, or stratified sampling.")
print("- Focus on evaluation metrics like F2 Score and Recall instead of Accuracy.")

print("\nBUSINESS RECOMMENDATIONS:")
print("- Conduct root cause analysis for churned customers.")
print("- Design targeted retention programs such as loyalty offers, email reminders, or personalized discounts.")
````

### <u>**Summary of Data Understanding**</u>

The following is a **key summary** obtained during the **data understanding phase**, based on the initial exploration:

#### **1. Dataset Overview**  
* The dataset consists of **5,630 rows** and **20 columns**, comprising:
  * **19 input features** reflecting customer demographics, app usage behavior, and transactional patterns.
  * **1 target variable**: `Churn`, indicating whether the customer has stopped using the service (`1`) or is still active (`0`).
* Each row represents a unique customer on the e-commerce platform.

#### **2. Missing Values**  
* There are **7 features** with missing values (`NaN`), including:
  * `Tenure`, `WarehouseToHome`, `HourSpendOnApp`, `OrderAmountHikeFromlastYear`, `CouponUsed`, `OrderCount`, and `DaySinceLastOrder`
* The percentage of missing values ranges between **4.46% – 5.45%**, thus requiring thoughtful handling strategies depending on data type and distribution.
* **Recommended action for data cleaning:**
  * Perform **distribution analysis** (`min`, `max`, `mean`, `median`) before deciding on imputation methods.
  * For numerical features, consider whether using the `median`, `mean`, or other techniques such as flagging is appropriate.
  * For time- or quantity-related features (`OrderCount`, `CouponUsed`, etc.), avoid imputing with median without analyzing outliers or 0-value anomalies.
  * Ensure imputation doesn’t **distort the original data distribution**.

#### **3. Data Types & Formatting Recommendations**

In this project, data cleaning focuses on **handling missing values** to avoid bias or information loss, **converting incorrect data types** to ensure each feature is correctly interpreted by the classification algorithm, **removing duplicate entries** to maintain data integrity, and **standardizing inconsistent categorical values** for uniform representation. Additionally, the process includes **detecting and managing outliers** that could skew the model's learning process. These steps are essential to ensure the dataset is **clean**, **consistent**, and ready for building an **accurate** and **reliable classification model**, as well as for generating insightful **exploratory analysis**.

#### **II.2.1 Handling Missing Values**

**Missing values** occur when data points are absent or not recorded in a dataset, typically represented as **`NaN`**. These gaps may result from **data collection issues**, **system errors**, or **non-responses**. In the context of a **classification model**, unhandled missing values can lead to **biased predictions**, **misleading feature importance**, or even **model failure**.

Based on the **data understanding phase**, there are **7 columns** with missing values, each with approximately **4.5% to 5.5%** missing data. The affected features include: `Tenure`, `WarehouseToHome`, `HourSpendOnApp`, `OrderAmountHikeFromlastYear`, `CouponUsed`, `OrderCount`, and `DaySinceLastOrder`. These missing values will be **handled** in the **data cleaning phase** to ensure the dataset is reliable and suitable for building an accurate and robust classification model.

The following table outline the recommended procedures for handling missing values:

| **Percentage of Missing Values** | **Category of Missing Values** | **Recommended Action**                                                                                     | **Source**                                                                                                                                                             |
|---------------------------------|---------------------------------|------------------------------------------------------------------------------------------------------------|-----------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| **Under 1%**                         | Very Low                        | Impute with mean/median/mode or drop rows; missingness is usually negligible | **"Statistical Analysis with Missing Data"** by Roderick J. A. Little & Donald B. Rubin (2019) |
| **1% - 5%**                     | Low                             | Impute with mean/median/mode or drop rows                                                                   | **"Practical Guide to Handling Missing Data in Statistics"** and **"Handling Missing Data in Machine Learning: A Review"** (2020) from *Nature Communications* |
| **5% - 15%**                    | Moderate                        | Impute with mean/median/mode or use algorithms that handle missing values                                   | **"Practical Guide to Handling Missing Data in Statistics"**, **"Handling Missing Data in Machine Learning: A Review"** (2020) from *Nature Communications* |
| **15% - 40%**                   | High                            | Impute, use predictive models, or drop columns                                                              | **"Missing Data: A Statistical Analysis and Practical Approach"** by Ruth M. Carroll and Joseph J. K. Hsieh and **"Practical Guide to Handling Missing Data in Statistics"** |
| **Above 40%**                   | Very High                       | Drop columns or use advanced imputation techniques                                                          | **"Handling Missing Data in Machine Learning: A Review"** (2020) from *Nature Communications* |
| **100%**                         | Completely Missing              | Drop column                                                                                                 | **"Practical Guide to Handling Missing Data in Statistics"** and **"Handling Missing Data in Machine Learning: A Review"** (2020) from *Nature Communications* |

<br>

2. **Removed Rows Missing `OrderCount`**  
   - **Action**: Dropped all rows where **`OrderCount`** was NaN (~4.58% of the data)  
   - **Reason**:  
     - **`OrderCount`** is a key behavioral variable indicating purchase frequency. Unlike passive attributes (e.g., tenure or time on app), order activity directly reflects engagement and intent.  
     - Missing values in **`OrderCount`** cannot be safely imputed with **0** because zero implies a known absence of orders, whereas a null value indicates **unknown** order history, which is two semantically different meanings.  
     - Similarly, imputing with the **median** would assign an arbitrary level of purchase activity to customers whose behavior is genuinely unknown. This can introduce misleading patterns into the model.  
     - These missing entries are more likely caused by **data availability or tracking limitations**, not customer inactivity. Retaining them through imputation would reduce the accuracy and reliability of churn prediction.  
     - Since the proportion of affected rows is small (~4.58%), removing them preserves data integrity with minimal loss while avoiding invalid behavioral assumptions.

3. **Filled `CouponUsed` with 0**  
   - **Action**: Replaced NaN in **`CouponUsed`** with `0`  

   - **Reason**:  
      - **Semantic Meaning of Missingness**  
      In the context of eCommerce behavioral data, missing values in **`CouponUsed`** often reflect the **absence of coupon activity**, not data collection errors. It is common for transactional systems to log coupon usage **only when a customer applies one**, leaving the field blank otherwise. This makes 0 a **semantically accurate default** that captures the “no coupon used” condition.

      - **Improved Interpretability**  
      Using 0 clarifies that the customer **actively did not use a coupon**, rather than leaving the value ambiguous. This improves downstream **model interpretability** and removes the need to infer missingness through correlations or assumptions. It also supports clearer business segmentation between coupon users and non-users.

      - **Avoids Inappropriate Median Imputation**  
      The **median of `CouponUsed`** is likely to be **greater than zero**, especially if most customers who used coupons did so multiple times. Imputing with the median would incorrectly suggest that all customers with missing values used coupons at least a few times. This could introduce significant **bias**, especially when combined with **`OrderCount`**, which is closely related. For example, imputing a nonzero coupon value for a customer with **`OrderCount` = 0** would create a **logical inconsistency**.

      - **Prevents Unnecessary Data Loss**  
      Dropping rows with missing values in **`CouponUsed`** would eliminate valid behavioral records where other key features are complete. Since this column has **only ~4.5% missing values**, imputation with zero is a **low-risk strategy** that preserves the dataset size while maintaining data integrity.  
<br>

4. **Imputed Remaining Numerical Features with Median**  
   - **Action**: Median imputation was applied to the following numerical columns with low missingness: `Tenure`, `WarehouseToHome`, `HourSpendOnApp`, `OrderAmountHikeFromlastYear`, `DaySinceLastOrder`  

   - **Reason**:  
      - **Robust to Outliers and Skewed Distributions**  
      Behavioral data in eCommerce often contains **long-tailed patterns**. For example, a minority of users may spend disproportionately high time on the app or have extreme delivery distances due to geographic constraints. Median, as a **rank-based statistic**, is inherently resistant to such extremes. Unlike the mean, it reflects the **true central tendency** of skewed or irregular distributions, resulting in more stable imputations that do not pull the feature toward inflated averages.

      - **Low Missingness Justifies Simple Strategy**  
      Each selected feature has a **low proportion of missing values (<5.5%)**, which falls within the widely accepted range for **single-value imputation** techniques. In such cases, complex imputation methods (e.g., KNN, MICE) are **not only computationally expensive** but also risk **overfitting patterns in noise**, especially when the feature has weak correlation with others. Median offers a **low-variance, low-bias solution** suitable for these conditions.

      - **Prevents Cascade of Row Loss**  
      Dropping missing rows across multiple columns with small individual null rates can lead to a **disproportionately large loss of data** due to overlapping missingness. For example, five columns with ~5% missingness each can result in **up to 25% row loss** if dropped naively. Median imputation **conserves these otherwise valid observations**, enabling more robust statistical inferences and higher training sample sizes for modeling.

      - **Distribution-Preserving and Assumption-Free**  
      Median imputation **preserves the natural spread and rank structure** of the feature, unlike mean imputation which may artificially concentrate values around a central point. Additionally, it makes **no assumptions about feature linearity, co-dependence, or distribution shape**, making it particularly suitable when the **mechanism of missingness is unclear** or not strongly predictable from other variables.

#### **II.2.3 Handling Duplicate Data**

**Handling duplicate data** involves identifying and addressing repeated records within a dataset. Duplicate entries occur when the **same observation is recorded multiple times**, causing redundancy that can skew analysis and model training. For a **classification model**, duplicates can lead to **biased predictions**, **overfitting**, and an overestimation of model performance, as certain samples are disproportionately represented. Therefore, removing duplicates is critical to ensure the model learns from **accurate, independent, and consistent data points**, which enhances the reliability and generalizability of the classification results.

````python
# CHECK DUPLICATE DATA (ALL COLUMNS)

# Check the number of duplicate rows across all columns
duplicates_all_columns = df.duplicated().sum()

# Print the number of duplicate rows across all columns
print(f"Number of duplicate rows across all columns: {duplicates_all_columns}")
````

There is **no duplicate data when checking across all columns**. However, this **does not necessarily mean the dataset is entirely free of duplicates**. To ensure data integrity, duplicate records were checked based on **critical columns**. These critical columns are essential features that uniquely characterize each customer interaction and are pivotal for accurate analysis and classification.

For this dataset, the **critical columns include all columns except `CustomerID`**, as the `CustomerID` serves as a unique identifier and should not be considered when identifying duplicates of actual data content.

Since `CustomerID` will be **dropped prior to modeling** because it does not carry predictive information, duplicate rows based on the remaining features **may exist**. Therefore, it is important to carefully check for duplicates after dropping `CustomerID`, as such duplicates could affect model performance and should be handled accordingly.

`CustomerID` was excluded from the duplicate check because it is a unique identifier that does not represent customer behavior or transactional characteristics. Therefore, duplicate rows based on behavioral features are still considered valid duplicates even if they have different CustomerIDs.

````python
# CHECK DUPLICATE DATA (CRITICAL COLUMNS)

# Check for duplicates in critical columns (all columns except 'CustomerID')
critical_columns = df.columns.drop('CustomerID').tolist()
duplicates_critical_columns = df.duplicated(subset=critical_columns)

# Print the number of duplicate rows for the critical columns
print(f"Number of duplicate rows in critical columns: {duplicates_critical_columns.sum()}\n")

# Display the duplicate rows
duplicate_rows = df[duplicates_critical_columns]
duplicate_rows
````

#### **II.2.4 Handling Inconsistent Data**

Inconsistent data refers to entries that do not conform to **expected categories, formats, or naming conventions**. Such inconsistencies often result from **human error, unclear data entry standards, or legacy system differences**. If left uncorrected, they can **mislead analysis, reduce model accuracy, and introduce unnecessary noise** into the dataset. Addressing these issues is essential for building a **robust and reliable machine learning model**.

The first step in handling inconsistencies involves **checking all unique values** in the categorical features. This helps identify **formatting issues, spelling variations, abbreviations, or semantically similar categories** that should be standardized. Detecting and correcting these discrepancies ensures that the dataset is **clean, consistent, and suitable for accurate analysis and modeling**.


````python
# CHECK INCONSISTENT DATA

# Define categorical columns to inspect for inconsistent labels
categorical_columns = [
    'PreferredLoginDevice',
    'PreferredPaymentMode',
    'Gender',
    'PreferedOrderCat',
    'MaritalStatus'
]

# Print total number of categorical columns being checked
print(f"\nTotal number of categorical columns to inspect: {len(categorical_columns)}\n")
print("Listing unique values and their frequencies for each categorical column:\n")

# Loop through each specified categorical column
for col in categorical_columns:
    if col in df.columns:
        print(f"Column: {col}")
        vc = df[col].value_counts(dropna=False)
        print(f"Number of unique values: {vc.shape[0]}")
        for i, (val, count) in enumerate(vc.items(), 1):
            print(f"{i:>2}. '{val}' — {count} rows")
        print("-" * 70)
    else:
        print(f"Column '{col}' not found in the dataset.")
````

````python
# HANDLING INCONSISTENT CATEGORICAL DATA

# List of categorical columns to be standardized
categorical_columns = [
    'PreferredLoginDevice',
    'PreferredPaymentMode',
    'Gender',
    'PreferedOrderCat',
    'MaritalStatus'
]

# Step 1: Remove leading/trailing whitespace from all relevant columns
for col in ['PreferredLoginDevice', 'PreferredPaymentMode', 'PreferedOrderCat']:
    if col in df.columns:
        df[col] = df[col].str.strip()

# Step 2: Standardize inconsistent labels using .replace()
# PreferredLoginDevice: unify 'Phone' into 'Mobile Phone'
df['PreferredLoginDevice'] = df['PreferredLoginDevice'].replace({
    'Phone': 'Mobile Phone'
})

# PreferredPaymentMode: unify abbreviations
df['PreferredPaymentMode'] = df['PreferredPaymentMode'].replace({
    'CC': 'Credit Card',
    'COD': 'Cash on Delivery'
})

# PreferedOrderCat: unify shortened label
df['PreferedOrderCat'] = df['PreferedOrderCat'].replace({
    'Mobile': 'Mobile Phone'
})

# Step 3: Display unique values for validation after standardization
print("\nValidation of cleaned categorical values:\n")
for col in categorical_columns:
    if col in df.columns:
        vc = df[col].value_counts(dropna=False)
        print(f"Column: {col}")
        print(f"Number of unique values: {vc.shape[0]}")
        for i, (val, count) in enumerate(vc.items(), 1):
            print(f"{i:>2}. '{val}' — {count} rows")
        print("-" * 70)
    else:
        print(f"Column '{col}' not found in the dataset.")
````

#### **II.2.5 Handling Outliers**

**Outliers** are data points that deviate significantly from the general pattern of the dataset. They may arise from **data entry errors**, **system issues**, or **unusual but valid behaviors**. In **classification modeling**, outliers can cause **misleading patterns**, **skewed decision boundaries**, and **model instability**. However, **not all outliers should be removed**, as some may reflect **important edge cases** in customer behavior. Effective outlier handling requires **contextual understanding** and should balance data quality with the preservation of valuable information. This ensures the model is trained on **reliable and representative data**, leading to more **generalizable** and **robust** outcomes.

````python
# Identify all numerical columns
all_numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Exclude specific columns that should not be treated as numerical for outlier detection
excluded_columns = ['CustomerID', 'Churn', 'CityTier', 'SatisfactionScore', 'Complain']

# List specific numerical columns for outlier analysis
numerical_columns_for_outlier_check = [col for col in all_numerical_columns if col not in excluded_columns]

# Display the relevant columns
print("Numerical columns to check for outliers:")
print(numerical_columns_for_outlier_check)
````

````python
# CHECK OUTLIERS USING IQR

# Create a function to check outliers
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)  # 25th percentile (lower quartile)
    Q3 = df[column].quantile(0.75)  # 75th percentile (upper quartile)
    IQR = Q3 - Q1  # Interquartile Range
    lower_bound = Q1 - 1.5 * IQR  # Lower bound for outliers
    upper_bound = Q3 + 1.5 * IQR  # Upper bound for outliers
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    return outliers, lower_bound, upper_bound

# Initialize an empty list to store columns that have outliers
columns_with_outliers = []

# Detect outliers in each numerical column
for col in numerical_columns_for_outlier_check:
    outliers, lower, upper = detect_outliers_iqr(df, col)

    print(f"Column: {col}")
    print(f"Lower Bound: {lower}, Upper Bound: {upper}")
    print(f"Number of Outliers: {outliers.shape[0]}")
    print("-" * 50)

    if outliers.shape[0] > 0:
        columns_with_outliers.append(col)

# Print all columns that contain outliers
if columns_with_outliers:
    print("\nThe columns with outliers are:", ", ".join(columns_with_outliers))
else:
    print("\nNo columns contain outliers.")
````

````python
# VISUALIZE COLUMNS WITH OUTLIERS USING BOXPLOT

# Set number of plots per row
plots_per_row = 3

# Calculate number of total plots
n_plots = len(numerical_columns_for_outlier_check)

# Calculate number of rows needed
n_rows = n_plots // plots_per_row + (1 if n_plots % plots_per_row != 0 else 0)

# Set up the figure size based on number of rows and columns
plt.figure(figsize=(plots_per_row * 5, n_rows * 4))

# Loop through the columns and plot boxplots
for i, column in enumerate(numerical_columns_for_outlier_check, 1):
    plt.subplot(n_rows, plots_per_row, i)
    sns.boxplot(x=df[column], showfliers=True, color='lightseagreen')
    plt.title(f'Boxplot of {column}', fontsize=10)

# Adjust spacing and show the boxplots
plt.tight_layout()
plt.show()
````

####    <u>**Summary of Handling Outliers**</u>  

1. **Numerical Columns Excluded from Outlier Checking:**  
    - The columns `Churn`, `CityTier`, `Complain`, and `SatisfactionScore` are excluded from outlier detection because they represent ordinal or categorical data rather than continuous numeric variables.


2. **Columns with Outliers:**  
    Outlier detection was performed on all 10 numerical features. The following columns were found to contain outliers:
    - `Tenure`: 4 outliers (customers with unusually long tenure)  
    - `WarehouseToHome`: 122 outliers (customers with extreme delivery distances)  
    - `HourSpendOnApp`: 5 outliers (very high or low in-app engagement)  
    - `NumberOfDeviceRegistered`: 336 outliers (customers registering an unusually high number of devices)  
    - `NumberOfAddress`: 234 outliers (customers associated with multiple addresses beyond typical range)  
    - `OrderAmountHikeFromlastYear`: 25 outliers (sharp increases in annual order amount)  
    - `CouponUsed`: 559 outliers (frequent coupon usage well above the norm)  
    - `OrderCount`: 673 outliers (very high transaction frequency)  
    - `DaySinceLastOrder`: 55 outliers (customers who haven’t ordered in an unusually long time)  
    - `CashbackAmount`: 496 outliers (customers receiving large cashback amounts)


3. **Reason for Keeping Outliers:**
    - Outliers represent real-world behavioral extremes, such as highly engaged customers, heavy coupon users, or individuals with many devices or addresses.  
    - Removing these data points could result in the loss of valuable customer segments and important behavioral patterns.  
    - Since there are no negative values and all data falls within logical ranges, the identified outliers are considered valid observations that reflect genuine variability in customer behavior.


4. **Modeling Impact:**

    - **Inclusion of Outliers:**  
        - Outliers in customer behavior, such as exceptionally high order counts or extreme cashback amounts, are not necessarily errors but can represent **high-value or high-risk customer segments**.  
        - Including these observations allows the model to **capture rare but meaningful patterns** that could otherwise be lost if these data points were removed.  
        - Excluding outliers would bias the model toward average behaviors and **undermine performance for segments that deviate from the norm**, such as power users or dormant accounts.

    - **Model Robustness:**  
        - A well-regularized model or one using tree-based algorithms (e.g., Random Forest, Gradient Boosting) can naturally handle outliers **without being overly sensitive** to extreme values.  
        - Retaining outliers ensures **real-world generalizability**, as customer behavior in production is rarely uniform.  
        - This approach supports the development of **segmentation-aware strategies**, such as targeting customers with unusually high spending or reactivation of long-inactive users.

    - **Potential Handling:**  
        - While outliers are preserved, it is important to apply **robust modeling techniques** that are less affected by skewed distributions.  
        - Techniques like **log transformations, feature scaling, or robust estimators** may be used selectively to manage extreme values during preprocessing.  
        - This ensures that the model benefits from the diversity of customer behavior **without being distorted** by it.

### **II.3.1 Churn by Demographics**

#### **Objective of the Analysis**

This analysis aims to evaluate whether **customers’ demographic characteristics** have a significant relationship with the likelihood of customer *churn* (discontinuing the service).

By understanding the demographic profile of customers who are more at risk of churn, the company can design more **targeted** and **personalized** retention strategies according to market segments.

#### **What is Demographics?**

**Demographic features** represent the basic characteristics of individuals or customer groups that are generally **stable** or unlikely to change in the short term, such as:

* Gender
* Marital Status
* Urbanization level of residence area (CityTier)

**Note:** Demographics are **not limited to geographical location**, but also include basic social identity of the customer.

#### **Business Questions Supported**

* Do certain demographic characteristics show higher churn rates?
* Can we identify demographic segments that require special attention to improve customer loyalty?

#### **Features Analyzed & Reasons for Selection**

| Feature         | Statistical Data Type | Reason for Categorization & Relevance to Demographics                                                          |
| --------------- | --------------------- | -------------------------------------------------------------------------------------------------------------- |
| `Gender`        | Nominal Categorical   | Represents the customer’s biological identity, which may influence preferences, shopping behavior, and loyalty |
| `MaritalStatus` | Nominal Categorical   | Marital status is often associated with stability level, life priorities, and tendency toward loyalty          |
| `CityTier`      | Ordinal Categorical   | Represents the urbanization level of residence, influencing purchasing power, service accessibility, and needs |

**Urgency of Analysis:**
These features are fundamental and **usually available from the onboarding stage**, so they can be utilized early for churn risk segmentation.

### **Analysis Methods**

#### **Stages & Approach**

| Stage           | Approach                                                                                                                      |
| --------------- | ----------------------------------------------------------------------------------------------------------------------------- |
| **Univariate**  | Observing the distribution of each category (`Gender`, `MaritalStatus`, `CityTier`) without considering churn                 |
| **Bivariate**   | Comparing the proportion of churn vs. non-churn within each category (using stacked bar plots)                                |
| **Statistical** | **Chi-Square Test** → Measures whether there is a significant association between each demographic feature and `Churn` status |

#### **Visualizations Used**

| Visualization Type | Purpose                                                                       |
| ------------------ | ----------------------------------------------------------------------------- |
| Barplot            | To observe the number of customers in each category                           |
| Stacked Barplot    | To show the proportion of churned and retained customers within each category |

#### **Chi-Square Test Explanation**

* **Purpose**: To measure whether there is a **statistically significant relationship** between two categorical variables (e.g., `Gender` and `Churn`)

* **Best used for**: **Categorical variables** → when we want to know whether the distribution of `Churn` differs significantly between categories.

* **Hypotheses**:

  * **H₀ (null)**: There is no association between the feature and churn
  * **H₁ (alternative)**: There is an association between the feature and churn

* **Test statistic**: p-value

* **Decision Rule**:

  * If **p-value < 0.05** → Reject H₀ → There is a significant relationship
  * If **p-value ≥ 0.05** → Fail to reject H₀ → No significant relationship

<br>



````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency
from tabulate import tabulate
import matplotlib.patches as mpatches

# Set clean visual theme
sns.set_style("whitegrid")

# Color palette for churn status (used in stacked bar only)
churn_palette = ['#A8DADC', '#E5989B']  # Soft teal for Retained, rose-pink for Churned

# Distinct and elegant color palettes for categorical features
gender_palette = {'Male': '#457B9D', 'Female': '#F4A261'}  # Deep muted blue & orange
marital_palette = {'Married': '#81B29A', 'Divorced': '#E07A5F', 'Single': '#A3A380'}  # Sage, terracotta, olive
city_palette = {'1': '#6D6875', '2': '#B5838D', '3': '#FFCDB2'}  # Muted plum, mauve, peach

# Grouped palettes per feature
demo_features = {
    'Gender': gender_palette,
    'MaritalStatus': marital_palette,
    'CityTier': city_palette
}

# Store Chi-Square results for summary output
chi_square_results = []

# Loop through each demographic feature
for feature, palette in demo_features.items():
    print(f"\n{'='*35}\nAnalysis by: {feature}\n{'='*35}")

    fig, axs = plt.subplots(1, 2, figsize=(13, 5))
    plt.suptitle(f'Churn Analysis by {feature}', fontsize=14, fontweight='bold')

    # ----- Left Plot: Total distribution of customers per category -----
    sns.countplot(data=df,
                  x=feature,
                  order=df[feature].value_counts().index,
                  palette=palette,
                  ax=axs[0])

    axs[0].set_title(f'Distribution of {feature}')
    axs[0].set_ylabel('Customer Count')
    axs[0].set_xlabel('')

    # Add value labels on top of bars
    for bar in axs[0].patches:
        axs[0].text(bar.get_x() + bar.get_width()/2,
                    bar.get_height(),
                    f'{int(bar.get_height())}',
                    ha='center',
                    va='bottom',
                    fontsize=10)

    # Add custom legend to explain category colors
    legend_patches = [mpatches.Patch(color=color, label=label) for label, color in palette.items()]
    axs[0].legend(handles=legend_patches, title=f'{feature} Category', loc='upper right')

    # ----- Right Plot: Churn proportion by category (normalized) -----
    churn_prop = pd.crosstab(df[feature], df['Churn'], normalize='index') * 100
    churn_prop.plot(kind='bar',
                    stacked=True,
                    ax=axs[1],
                    color=churn_palette,
                    edgecolor='white')

    axs[1].set_title(f'Churn Proportion by {feature}')
    axs[1].set_ylabel('Percentage (%)')
    axs[1].set_xlabel('')
    axs[1].legend(['Retained', 'Churned'], title='Churn Status')

    # Adjust spacing
    plt.tight_layout(rect=[0, 0, 1, 0.93])
    plt.show()

    # ----- Chi-Square Test of Independence -----
    crosstab = pd.crosstab(df[feature], df['Churn'])
    chi2, p, dof, _ = chi2_contingency(crosstab)
    significance = 'Yes' if p < 0.05 else 'No'

    print(f"Chi2 Statistic     : {chi2:.4f}")
    print(f"Degrees of Freedom : {dof}")
    print(f"P-value            : {p:.4f}")
    print(f"Statistically Significant? : {significance}")

    # Append to summary table
    chi_square_results.append([feature, round(p, 4), significance])

# ----- Chi-Square Summary Table -----
summary_df = pd.DataFrame(chi_square_results,
                          columns=['Feature', 'P-Value', 'Significant (p < 0.05)'])
print("\nChi-Square Test Summary\n")
print(tabulate(summary_df, headers='keys', tablefmt='pretty', showindex=False))
````

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set visualization style
sns.set(style='whitegrid')

# ==============================
# 1. Create Temporary Grouping
# ==============================

# Copy df for visualization purposes only
df_temp = df.copy()

# Create a combined column of MaritalStatus and CityTier for grouped analysis
df_temp['Marital_City_Group'] = df_temp['MaritalStatus'] + ' - Tier ' + df_temp['CityTier'].astype(str)

# Sort groups by customer count (for visualization order)
group_order = df_temp['Marital_City_Group'].value_counts().index

# Custom color palette
custom_palette = {
    0: '#B4D4FF',  # Retained
    1: '#FFB3C1'   # Churned
}

# ==============================
# 2. Visualisasi Barplot
# ==============================

plt.figure(figsize=(10, 6))
ax = sns.countplot(
    data=df_temp,
    x='Marital_City_Group',
    hue='Churn',
    order=group_order,
    palette=custom_palette
)

plt.title('Churn Distribution by Marital Status & City Tier', fontsize=14)
plt.xlabel('Marital Status & City Tier')
plt.ylabel('Customer Count')
plt.xticks(rotation=30)
plt.legend(title='Churn Status', labels=['Retained', 'Churned'])

# Add value labels on top of each bar
for container in ax.containers:
    ax.bar_label(container, label_type='edge', fontsize=9)

plt.tight_layout()
plt.show()

# ==============================
# 3. Churn Metrics per Group
# ==============================

# Calculate churn metrics from df_temp (not the main df)
churn_group = df_temp.groupby('Marital_City_Group')['Churn'].agg(['count', 'sum', 'mean']).reset_index()

# Rename columns for clearer interpretation
churn_group.rename(columns={
    'Marital_City_Group': 'Group',
    'count': 'Total Customers',
    'sum': 'Total Churned',
    'mean': 'Churn Rate'
}, inplace=True)

# Convert churn rate to percentage
churn_group['Churn Rate'] = churn_group['Churn Rate'] * 100

# Sort by highest churn rate
churn_group = churn_group.sort_values(by='Churn Rate', ascending=False).reset_index(drop=True)

# Display result
display(churn_group)
````

### **II.3.2 Churn by Behavioral Engagement**

#### **Objective of the Analysis**

This analysis aims to assess whether there are **significant differences** in *behavioral engagement* between customers who churn and those who do not. By identifying **user interaction patterns** with the platform, the company can detect **early churn signals** and design **more effective retention strategies**.

### **What is Behavioral Engagement?**

*Behavioral engagement* reflects the **actual activities of customers** when using the service. This includes subscription duration, app usage frequency, service satisfaction, and whether the customer has ever filed a complaint. Low engagement is often an **early indicator of churn**.

#### **Features Analyzed**

| Feature             | Statistical Data Type      | Meaning & Relevance to Engagement                                               |
| ------------------- | -------------------------- | ------------------------------------------------------------------------------- |
| `Tenure`            | Continuous Numeric         | Length of time as a customer. Represents long-term loyalty and user experience. |
| `HourSpendOnApp`    | Continuous Numeric         | Average time customers spend on the app. Indicator of daily engagement.         |
| `SatisfactionScore` | Discrete Numeric (Ordinal) | User perception of service satisfaction.                                        |
| `Complain`          | Binary Categorical         | Customer complaint history. Indicates potential negative experiences.           |

#### **Business Questions Answered**

* Do customers with low engagement tend to churn?
* Do satisfaction and negative experiences affect churn?
* Can duration and usage intensity be early warning signals?

#### **Analysis Methods**

| Analysis Stage  | Approach                                                                                       |
| --------------- | ---------------------------------------------------------------------------------------------- |
| **Univariate**  | Exploring the distribution of each feature using histograms/barplots                           |
| **Bivariate**   | Comparing feature distributions between churned and non-churned groups using boxplots/barplots |
| **Statistical** | Applying statistical tests to assess significant differences between churn vs non-churn groups |

#### **Statistical Tests Used**

| Feature             | Variable Type | Distribution     | Statistical Test    | Hypotheses                                        |
| ------------------- | ------------- | ---------------- | ------------------- | ------------------------------------------------- |
| `Tenure`            | Interval      | Non-Normal       | Mann-Whitney U Test | H₀: Same distribution; H₁: Different distribution |
| `HourSpendOnApp`    | Interval      | Non-Normal       | Mann-Whitney U Test | H₀: Same distribution; H₁: Different distribution |
| `SatisfactionScore` | Ordinal       | Discrete/Ordinal | Mann-Whitney U Test | H₀: Same distribution; H₁: Different distribution |
| `Complain`          | Categorical   | Discrete         | Chi-Square Test     | H₀: No relationship; H₁: Significant relationship |

#### **Rationale for Statistical Test Selection:**

1. **Mann-Whitney U Test** is used for numeric and ordinal features (non-normal distribution):

   * Alternative to independent t-test when data is **non-normal** or **ordinal**
   * Suitable for comparing distributions between two independent groups (`Churn = 0 vs 1`)
   * Answers: *Are the medians of the distributions significantly different?*

2. **Chi-Square Test of Independence** is used for categorical features:

   * To test the **relationship** between two categorical variables (`Complain` and `Churn`)
   * Answers: *Is the frequency of complaints significantly different between churned and non-churned groups?*

<br>


````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import mannwhitneyu, chi2_contingency

# Set plot style
sns.set(style='whitegrid')

# -------------------------------
# 1. UNIVARIATE VISUALIZATIONS
# -------------------------------

# Histogram – Tenure
plt.figure(figsize=(6, 4))
sns.histplot(df['Tenure'], kde=True, bins=30, color='#4C78A8')
plt.title('Customer Distribution by Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()

# Histogram – HourSpendOnApp
plt.figure(figsize=(6, 4))
sns.histplot(df['HourSpendOnApp'], kde=True, bins=30, color='#F58518')
plt.title('Customer Distribution by App Usage')
plt.xlabel('Average Hours Spent on App per Week')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()

# Countplot – Satisfaction Score
plt.figure(figsize=(6, 4))
sns.countplot(x='SatisfactionScore', data=df, palette='Blues')
plt.title('Customer Distribution by Satisfaction Score')
plt.xlabel('Satisfaction Score (1 to 5)')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()

# Countplot – Complaint Status
plt.figure(figsize=(6, 4))
sns.countplot(x='Complain', data=df, palette='Set2')
plt.title('Customer Distribution by Complaint Status')
plt.xlabel('Complain (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')
plt.xticks([0, 1], ['No Complaint', 'Complained'])
plt.tight_layout()
plt.show()

# -------------------------------
# 2. BIVARIATE VISUALIZATIONS
# -------------------------------

# Boxplot – Tenure vs Churn
plt.figure(figsize=(6, 4))
sns.boxplot(x='Churn', y='Tenure', data=df, palette='Set3')
plt.title('Tenure Distribution by Churn Status')
plt.xlabel('Churn (0 = No, 1 = Yes)')
plt.ylabel('Tenure (Months)')
plt.xticks([0, 1], ['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# Boxplot – HourSpendOnApp vs Churn
plt.figure(figsize=(6, 4))
sns.boxplot(x='Churn', y='HourSpendOnApp', data=df, palette='Set3')
plt.title('App Usage Distribution by Churn Status')
plt.xlabel('Churn (0 = No, 1 = Yes)')
plt.ylabel('Average Hours on App per Week')
plt.xticks([0, 1], ['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# Boxplot – Satisfaction Score vs Churn
plt.figure(figsize=(6, 4))
sns.boxplot(x='Churn', y='SatisfactionScore', data=df, palette='Set3')
plt.title('Satisfaction Score by Churn Status')
plt.xlabel('Churn (0 = No, 1 = Yes)')
plt.ylabel('Satisfaction Score')
plt.xticks([0, 1], ['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# Countplot – Complain vs Churn
plt.figure(figsize=(6, 4))
sns.countplot(x='Complain', hue='Churn', data=df, palette='Set1')
plt.title('Churn Count by Complaint Status')
plt.xlabel('Complaint Status (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# -------------------------------
# 3. STATISTICAL TESTING
# -------------------------------

print("="*60)
print("BEHAVIORAL ENGAGEMENT – STATISTICAL TEST RESULTS")
print("="*60)

# Mann-Whitney U Test for numerical/ordinal variables
numerical_features = ['Tenure', 'HourSpendOnApp', 'SatisfactionScore']
for col in numerical_features:
    group_0 = df[df['Churn'] == 0][col]
    group_1 = df[df['Churn'] == 1][col]
    stat, p = mannwhitneyu(group_0, group_1, alternative='two-sided')

    print(f"\nFeature: {col}")
    print("-" * 50)
    print("Group Statistics:")
    print(f"Not Churned - Mean: {group_0.mean():.2f}, Median: {group_0.median():.2f}")
    print(f"Churned     - Mean: {group_1.mean():.2f}, Median: {group_1.median():.2f}")
    print("\nMann-Whitney U Test:")
    print(f"U Statistic : {stat:.2f}")
    print(f"P-Value     : {p:.4f}")
    if p < 0.05:
        print("Conclusion  : Significant difference (Reject H₀)")
    else:
        print("Conclusion  : No significant difference (Fail to reject H₀)")

# Chi-Square Test – Complain vs Churn
contingency = pd.crosstab(df['Complain'], df['Churn'])
chi2, p, dof, expected = chi2_contingency(contingency)

print("\n" + "-" * 50)
print("Chi-Square Test: Complain vs Churn")
print("-" * 50)
print("Contingency Table:")
print(contingency)
print("\nChi-Square Test Results:")
print(f"Chi² Statistic : {chi2:.2f}")
print(f"P-Value        : {p:.4f}")
if p < 0.05:
    print("Conclusion     : Significant association (Reject H₀)")
else:
    print("Conclusion     : No significant association (Fail to reject H₀)")
````

After conducting the statistical test, we decided to perform an additional analysis focused on gaining insights into the churn patterns within this dataset.  

````python
# Calculate churn percentage for each tenure
tenure_churn = df.groupby('Tenure')['Churn'].mean() * 100  # Multiply by 100 to get percentage

# Create a DataFrame for visualization
tenure_churn_df = tenure_churn.reset_index()
tenure_churn_df.columns = ['Tenure', 'ChurnPercentage']

# Create the bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x='Tenure', y='ChurnPercentage', data=tenure_churn_df, palette='Blues')

# Add labels and title
plt.title('Churn Percentage by Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Churn Percentage (%)')
plt.xticks(rotation=45)  # Rotate the tenure labels to prevent overlap

# Display the plot
plt.tight_layout()
plt.show()
````

This bar chart illustrates the churn percentage in relation to customer tenure in months. It shows that:

- Churn is highest for customers with shorter tenures, particularly those with 0-2 months.

- The churn percentage gradually decreases as the tenure increases, stabilizing after about 10 months.

- The chart also highlights a notable spike in churn around the 20-month mark.

This suggests that customers tend to leave more frequently in the early stages of their subscription, with retention improving over time.

Afterward, we need to perform some insightful analysis on other features, starting with plots that compare the count and percentage for each unique value of every variable.


````python
# List of features to calculate churn percentage for
features = ['HourSpendOnApp', 'WarehouseToHome', 'NumberOfDeviceRegistered', 'NumberOfAddress']

# Create a 2x2 grid for subplots
fig, axs = plt.subplots(2, 4, figsize=(20, 10))  # We need 4 columns: 2 for each set of plots

# Loop through the features and create both churn percentage and churn count plots
for i, feature in enumerate(features):
    # Churn Percentage Plot
    churn_percentage = df.groupby(feature)['Churn'].mean() * 100  # Multiply by 100 to get percentage
    churn_percentage_df = churn_percentage.reset_index()
    churn_percentage_df.columns = [feature, 'ChurnPercentage']

    # Create the churn percentage bar plot
    sns.barplot(x=feature, y='ChurnPercentage', data=churn_percentage_df, ax=axs[i//2, 2*(i%2)], palette='Blues')
    axs[i//2, 2*(i%2)].set_title(f'Churn Percentage by {feature}')
    axs[i//2, 2*(i%2)].set_xlabel(feature)
    axs[i//2, 2*(i%2)].set_ylabel('Churn Percentage (%)')
    if feature == 'WarehouseToHome':
        axs[i//2, 2*(i%2)].tick_params(axis='x', rotation=90)
    else:
        axs[i//2, 2*(i%2)].tick_params(axis='x', rotation=0)

    # Churn Count Plot
    churn_count = df.groupby([feature, 'Churn']).size().unstack(fill_value=0)
    churn_count_df = churn_count.reset_index()
    churn_count_df.columns = [feature, 'NonChurnCount', 'ChurnCount']

    # Create the churn count stacked bar plot
    churn_count_df.set_index(feature)[['ChurnCount', 'NonChurnCount']].plot(kind='bar', stacked=True, ax=axs[i//2, 2*(i%2)+1], color=['blue', 'lightblue'])
    axs[i//2, 2*(i%2)+1].set_title(f'Churn vs Non-Churn Count by {feature}')
    axs[i//2, 2*(i%2)+1].set_xlabel(feature)
    axs[i//2, 2*(i%2)+1].set_ylabel('Count')
    if feature == 'WarehouseToHome':
        axs[i//2, 2*(i%2)+1].tick_params(axis='x', rotation=90)
    else:
        axs[i//2, 2*(i%2)+1].tick_params(axis='x', rotation=0)

    # Annotate the bars with counts (no percentage) for each part of the stack
    for p in axs[i//2, 2*(i%2)+1].patches:
        height = p.get_height()
        x = p.get_x() + p.get_width() / 2
        y = p.get_y() + height / 2
        axs[i//2, 2*(i%2)+1].annotate(f'{int(height)}', (x, y), ha='center', va='center', color='white', fontsize=10)

# Adjust the layout to avoid overlap
plt.tight_layout()
plt.show()
````

### **Churn Rate Trends & Interactions**

This section presents an exploration of churn trends based on customer service duration (tenure) as well as the influence of combined customer behavioral features. The objectives are to:

* Assess how churn rates change as customer tenure increases.
* Observe churn patterns based on interactions between features such as complaint status, satisfaction level, and app usage duration.

For visualization purposes, the categorical feature `TenureGroup` is used, along with combined features resulting from column interactions. All additional features are created in `df_temp` to avoid affecting the main data during the modeling process.

### **Churned Users – App Usage Behavior**

This analysis focuses solely on customers who have churned, to further understand their behavioral patterns—particularly in terms of app usage duration.

Visualizations are conducted using scatterplots and trend lines to capture the relationship between tenure and app usage, as well as heatmaps to illustrate the density distribution of churn based on the combination of usage time and tenure.

The findings from this section are expected to help identify **critical churn segments** that can be targeted for further intervention.

<br>

---


````python
# ===============================
# CHURN RATE TRENDS & INTERACTIONS
# ===============================

# Churn rate per Tenure (lineplot)
churn_tenure = df_temp.groupby('Tenure')['Churn'].mean().reset_index()
churn_tenure['ChurnRate'] = churn_tenure['Churn'] * 100

plt.figure(figsize=(10, 5))
sns.lineplot(data=churn_tenure, x='Tenure', y='ChurnRate', marker='o', color='crimson')
plt.title('Churn Rate Across Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Churn Rate (%)')
plt.tight_layout()
plt.show()

# Create tenure group (categorical)
def categorize_tenure(t):
    if t <= 1: return '0–1 month'
    elif t <= 3: return '2–3 months'
    elif t <= 5: return '4–5 months'
    else: return '6+ months'

df_temp['TenureGroup'] = df_temp['Tenure'].apply(categorize_tenure)

# Define interaction features
interactions = {
    'Tenure_Complain': df_temp['TenureGroup'] + ' | ' + df_temp['Complain'].map({0: 'No Complaint', 1: 'Complained'}),
    'Satisfaction_Complain': df_temp['SatisfactionScore'].astype(str) + ' Star | ' + df_temp['Complain'].map({0: 'No Complaint', 1: 'Complained'}),
    'Tenure_Satisfaction': df_temp['TenureGroup'] + ' | ' + df_temp['SatisfactionScore'].astype(str) + ' Star'
}

# Visualize interaction plots
for name, group in interactions.items():
    df_temp[name] = group
    churn_inter = df_temp.groupby(name)['Churn'].mean().reset_index()
    churn_inter['Churn'] *= 100

    plt.figure(figsize=(10, 6))
    sns.barplot(x=name, y='Churn', data=churn_inter, palette='coolwarm')
    plt.title(f'Churn Rate by {name.replace("_", " ")}', fontsize=14)
    plt.xlabel(name.replace('_', ' '))
    plt.ylabel('Churn Rate (%)')
    plt.xticks(rotation=30, ha='right')
    plt.tight_layout()
    plt.show()

# App usage x Tenure group interaction
hour_bins = pd.cut(
    df_temp['HourSpendOnApp'],
    bins=[0, 2, 4, 6, 8, 10, float('inf')],
    labels=['0–2h', '2–4h', '4–6h', '6–8h', '8–10h', '10h+']
)
df_temp['HourTenure'] = df_temp['TenureGroup'] + ' | ' + hour_bins.astype(str)

churn_ht = df_temp.groupby('HourTenure')['Churn'].mean().reset_index()
churn_ht['Churn'] *= 100

plt.figure(figsize=(12, 6))
sns.barplot(x='HourTenure', y='Churn', data=churn_ht, palette='coolwarm')
plt.title('Churn Rate by App Usage & Tenure Group')
plt.xlabel('Tenure Group & App Usage (hrs/week)')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()

# ===============================
# CHURNED USERS – APP USAGE BEHAVIOR
# ===============================

churn_temp = df_temp[df_temp['Churn'] == 1].copy()

# Scatterplot with trendline
plt.figure(figsize=(10, 6))
sns.scatterplot(data=churn_temp, x='Tenure', y='HourSpendOnApp', alpha=0.6, color='steelblue')
sns.regplot(data=churn_temp, x='Tenure', y='HourSpendOnApp', scatter=False, color='darkblue', line_kws={'linestyle': 'dashed'})
plt.title('App Usage Over Tenure (Churned Only)')
plt.xlabel('Tenure (Months)')
plt.ylabel('Hours on App per Week')
plt.tight_layout()
plt.show()

# Heatmap: binned tenure & app usage
churn_temp['TenureBin'] = pd.cut(churn_temp['Tenure'], bins=10)
churn_temp['AppUsageBin'] = pd.cut(churn_temp['HourSpendOnApp'], bins=10)

heatmap_data = churn_temp.groupby(['TenureBin', 'AppUsageBin']).size().unstack().fillna(0)

plt.figure(figsize=(10, 6))
sns.heatmap(heatmap_data, cmap='Reds', annot=True, fmt='.0f', linewidths=0.5)
plt.title('Density of Churned Users by Tenure & App Usage')
plt.xlabel('App Usage (hrs/week)')
plt.ylabel('Tenure')
plt.tight_layout()
plt.show()
````

### **Advanced Analysis of Behavioral Feature Interactions with Churn**

After conducting univariate, bivariate, and statistical analyses, this stage deepens the understanding of churn through the exploration of **interactions between behavioral features** such as *tenure*, *complaint status*, *satisfaction score*, and *app usage*. This approach aims to identify **the most vulnerable customer segments** and design **more targeted retention interventions**.

#### **1. Tenure Interaction with Churn Rate (Line Plot)**

#### Insight:

* **Highest churn rates** occur among customers with **0–1 month tenure**, reaching over **50%**.
* After the second month, churn drops drastically and tends to stabilize below 10% after the 10th month.

#### Interpretation:

* The spike in month 20 may signal **a new point of frustration**, such as **promo expiration** or **cumulative disappointment**.

#### Retention Strategy:

* Focus on onboarding, early rewards, and intensive monitoring during the first 1–3 months.
* Evaluate **feature, promo, or service changes** occurring around month 20.

#### **2. Churn Rate Based on Tenure & Complaint Status Combination**

#### Insight:

* Customers with **tenure < 2 months and who have complained** have the highest churn rate, reaching **73%**.

#### Interpretation:

* Complaints in the early stages of usage are highly risky → indicating **a poor onboarding experience or unmet expectations**.
* New customers are very sensitive to **early frustration**.

#### Retention Strategy:

* Build an **alert system for new customer complaints** and prioritize resolution.
* Implement a **“rescue protocol”** program for customers <2 months who submit complaints.

#### **3. Churn Rate Based on Satisfaction Score & Complaint Status**

#### Insight:

* **High-scoring customers (4–5)** who **never complained** actually have the highest churn rates (36–38%).
* Customers with **low scores (1–2)** who **did complain** show lower churn rates.

#### Interpretation:

* **Satisfied but passive customers** → may just be trying the service, not committed, or lack loyalty despite being satisfied.
* **Complaints can be a sign of engagement**. If handled well, these customers may stay.

#### Retention Strategy:

* Create a **loyalty scoring system** based not only on satisfaction but also on activity.
* Launch “**Stay Reward**” campaigns for satisfied but passive customers → e.g., vouchers, referrals, or loyalty badges.

#### **4. Churn Rate Based on Tenure & Satisfaction**

#### Insight:

* Customers with **tenure < 2 months** remain the group with the highest churn rate, even across all satisfaction scores.
* **New customers with a score of 5** still have a high churn rate (>60%).

#### Interpretation:

* **High satisfaction does not guarantee early retention**. Customers may be functionally satisfied but not yet emotionally attached.

#### Retention Strategy:

* It’s not enough to ensure customers are “satisfied” → you must build **early engagement & emotional connection**.
* Use brand storytelling, communities, or personalized experiences to create deeper bonds.

#### **5. Churn Rate Based on App Usage & Tenure**

#### Insight:

* New customers (0–1 month) across all app usage levels (0–2h, 2–4h, 4–6h) still show high churn.
* Longer-tenured customers show lower churn even with varying app usage durations.

#### Interpretation:

* Usage time has little impact on retention **if early connection hasn’t been formed**.
* Early churn behavior is more influenced by perceived value & initial orientation than app usage frequency.

#### Retention Strategy:

* Focus on early activation based on **core features**, not just usage frequency.
* Create a **“guided tour” or gamified tutorial** in the first week to ensure value understanding.

#### **6. Heatmap: Churned Customer Density Based on Tenure and App Usage**

#### Insight:

* Most churned customers are within **< 8 months of tenure** and app usage of **< 2 hours/week**.
* The highest density combination: tenure 0–2 months & app usage < 2 hours/week.

#### Interpretation:

* This group represents **high-risk customers**, both in terms of age and engagement level.

#### Retention Strategy:

* Perform **cluster targeting** for new customers with low app usage → encourage with educational content, notification reminders, or exclusive promos.
* Evaluate whether the onboarding process effectively encourages feature exploration from the start.

| ----------------------- | -------------------------------------------------------------------- | -------------------------------------------------------------------- |

* Complaints in the early stages of usage are highly risky → indicating **a poor onboarding experience or unmet expectations**.
* New customers are very sensitive to **early frustration**.

#### Retention Strategy:

* Build an **alert system for new customer complaints** and prioritize resolution.
* Implement a **“rescue protocol”** program for customers <2 months who submit complaints.

#### **3. Churn Rate Based on Satisfaction Score & Complaint Status**

#### Insight:

* **High-scoring customers (4–5)** who **never complained** actually have the highest churn rates (36–38%).
* Customers with **low scores (1–2)** who **did complain** show lower churn rates.

#### Interpretation:

* **Satisfied but passive customers** → may just be trying the service, not committed, or lack loyalty despite being satisfied.
* **Complaints can be a sign of engagement**. If handled well, these customers may stay.

#### Retention Strategy:

* Create a **loyalty scoring system** based not only on satisfaction but also on activity.
* Launch “**Stay Reward**” campaigns for satisfied but passive customers → e.g., vouchers, referrals, or loyalty badges.

#### **4. Churn Rate Based on Tenure & Satisfaction**

#### Insight:

* Customers with **tenure < 2 months** remain the group with the highest churn rate, even across all satisfaction scores.
* **New customers with a score of 5** still have a high churn rate (>60%).

#### Interpretation:

* **High satisfaction does not guarantee early retention**. Customers may be functionally satisfied but not yet emotionally attached.

#### Retention Strategy:

* It’s not enough to ensure customers are “satisfied” → you must build **early engagement & emotional connection**.
* Use brand storytelling, communities, or personalized experiences to create deeper bonds.

#### **5. Churn Rate Based on App Usage & Tenure**

#### Insight:

* New customers (0–1 month) across all app usage levels (0–2h, 2–4h, 4–6h) still show high churn.
* Longer-tenured customers show lower churn even with varying app usage durations.

#### Interpretation:

* Usage time has little impact on retention **if early connection hasn’t been formed**.
* Early churn behavior is more influenced by perceived value & initial orientation than app usage frequency.

#### Retention Strategy:

* Focus on early activation based on **core features**, not just usage frequency.
* Create a **“guided tour” or gamified tutorial** in the first week to ensure value understanding.

#### **6. Heatmap: Churned Customer Density Based on Tenure and App Usage**

#### Insight:

* Most churned customers are within **< 8 months of tenure** and app usage of **< 2 hours/week**.
* The highest density combination: tenure 0–2 months & app usage < 2 hours/week.

#### Interpretation:

* This group represents **high-risk customers**, both in terms of age and engagement level.

#### Retention Strategy:

* Perform **cluster targeting** for new customers with low app usage → encourage with educational content, notification reminders, or exclusive promos.
* Evaluate whether the onboarding process effectively encourages feature exploration from the start.

### **II.3.3 Churn by Transaction Behavior**

#### **Objective of the Analysis**

This analysis aims to evaluate whether customer purchasing patterns—including order frequency, coupon usage, cashback amount, and purchase inactivity—are associated with churn tendencies.

By understanding customer transaction behavior, the company can identify early signals of churn and design appropriate intervention strategies to retain high-value customers.

#### **What is Transaction Behavior?**

Transaction behavior reflects **customer activity in making purchases on the platform**, including:

* Order frequency
* Transaction value or spending trends
* Coupon and incentive usage
* Inactivity period or gap between orders

These characteristics generally represent the **economic value of customers** to the company as well as **loyalty in transactional engagement**.

#### **Supported Business Questions**

* Are customers with low transaction activity more likely to churn?
* Do spending trends and incentives such as cashback or coupons influence loyalty?
* Can inactivity duration (days since last order) be an indicator of churn?

#### **Features Analyzed & Reasons for Selection**

| Feature                       | Data Type | Reason for Selection                                             |
| ----------------------------- | --------- | ---------------------------------------------------------------- |
| `OrderCount`                  | Numeric   | Shopping frequency → indicates customer transactional engagement |
| `OrderAmountHikeFromlastYear` | Numeric   | Annual spending growth → signals loyalty trend or decline        |
| `CouponUsed`                  | Numeric   | Promotion usage → reflects responsiveness to company incentives  |
| `DaySinceLastOrder`           | Numeric   | Time since last purchase → proxy for customer inactivity level   |
| `CashbackAmount`              | Numeric   | Monetary incentives received → impact on loyalty and retention   |

**Urgency of analysis:**
These features are directly linked to **company revenue** and **customer engagement in economic activity**, making them highly relevant for detecting churn risk.

### **Analysis Method**

#### **Stages & Approach**

| Stage           | Approach                                                                               |
| --------------- | -------------------------------------------------------------------------------------- |
| **Univariate**  | Visualize distribution of each numeric feature (histograms)                            |
| **Bivariate**   | Compare churn vs. non-churn groups (boxplots) for each feature                         |
| **Statistical** | Test distribution differences between churn and non-churn with **Mann-Whitney U Test** |

#### **Visualizations Used**

| Visualization Type | Purpose                                                             |
| ------------------ | ------------------------------------------------------------------- |
| Boxplot            | Identify differences in numeric feature values between churn groups |
| Histogram          | Understand overall data distribution for each numeric feature       |

#### **Statistical Test Explanation**

**Mann-Whitney U Test** is used to compare numeric feature distributions between two groups (`Churn = 1` and `Churn = 0`).

* **Hypotheses:**

  * H₀: No distribution difference between churn and non-churn groups
  * H₁: There is a significant distribution difference

* **Decision Rule:**

  * p-value < 0.05 → Reject H₀ → Significant difference

### **Example Statistical Output Table**

| Feature                     | Data Type | p-value | Significant (p < 0.05)? | Business Insight                                          |
| --------------------------- | --------- | ------- | ----------------------- | --------------------------------------------------------- |
| OrderCount                  | Numeric   | 0.013   | Yes                     | Churned customers have fewer total orders                 |
| OrderAmountHikeFromlastYear | Numeric   | 0.042   | Yes                     | Declining spend from last year → churn signal             |
| CouponUsed                  | Numeric   | 0.118   | No                      | Not statistically significant                             |
| DaySinceLastOrder           | Numeric   | 0.005   | Yes                     | Churned customers have longer gaps since their last order |
| CashbackAmount              | Numeric   | 0.021   | Yes                     | Lower cashback is associated with higher churn            |

### **Interpretation & Business Insights**

* Churned customers generally have **fewer orders**, **longer inactivity periods**, and **declining spending trends**.
* **Lower cashback values** also appear to contribute to churn, possibly due to a lack of financial incentives.
* Although **coupon usage** is not statistically significant, it can still be further analyzed through segmentation.

### **Business Recommendations**

* Identify customers with **low order volume and long time since last purchase** as high-risk groups.

* Design **targeted promotional campaigns**, such as:

  * Special cashback for customers with declining transaction values
  * Purchase reminders for inactive customers

* Run **retargeting campaign experiments using cashback or coupons** to assess impact on retention.

<br>


````python
# ================================================
# Section II.3.3 – Churn Analysis by Transaction Behavior
# ================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import mannwhitneyu

# Visual config
sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# Define transaction-related numerical features
transaction_features = [
    'OrderCount',
    'OrderAmountHikeFromlastYear',
    'CouponUsed',
    'DaySinceLastOrder',
    'CashbackAmount'
]

# 1. Univariate Histogram per Feature
for feature in transaction_features:
    sns.histplot(data=df, x=feature, hue='Churn', kde=True, multiple='stack')
    plt.title(f'Distribution of {feature} by Churn Status')
    plt.xlabel(feature)
    plt.ylabel('Number of Customers')
    plt.tight_layout()
    plt.show()

# 2. Bivariate Boxplot per Feature
for feature in transaction_features:
    sns.boxplot(data=df, x='Churn', y=feature)
    plt.title(f'{feature} Distribution by Churn Status')
    plt.xlabel('Churn')
    plt.ylabel(feature)
    plt.tight_layout()
    plt.show()

# 3. Mann-Whitney U Test with Extended Summary
print("========== Mann-Whitney U Test – Full Results ==========")
detailed_results = []

for feature in transaction_features:
    churned = df[df['Churn'] == 1][feature]
    not_churned = df[df['Churn'] == 0][feature]

    try:
        u_stat, p_val = mannwhitneyu(churned, not_churned, alternative='two-sided')
    except ValueError:
        u_stat, p_val = float('nan'), float('nan')

    significance = 'Yes' if pd.notnull(p_val) and p_val < 0.05 else 'No'

    # Summary stats
    mean_churn = churned.mean()
    mean_nonchurn = not_churned.mean()
    median_churn = churned.median()
    median_nonchurn = not_churned.median()

    # Direction of difference
    if mean_churn > mean_nonchurn:
        direction = 'Higher in churn'
    elif mean_churn < mean_nonchurn:
        direction = 'Lower in churn'
    else:
        direction = 'No difference'

    detailed_results.append({
        'Feature': feature,
        'U-Statistic': round(u_stat, 2) if pd.notnull(u_stat) else 'Error',
        'p-value': round(p_val, 4) if pd.notnull(p_val) else 'Error',
        'Significant (p < 0.05)': significance,
        'Mean (Churn)': round(mean_churn, 2),
        'Mean (Non-Churn)': round(mean_nonchurn, 2),
        'Median (Churn)': round(median_churn, 2),
        'Median (Non-Churn)': round(median_nonchurn, 2),
        'Mean Difference Direction': direction
    })

# 4. Display table
results_df = pd.DataFrame(detailed_results)
results_df = results_df[
    ['Feature', 'U-Statistic', 'p-value', 'Significant (p < 0.05)',
     'Mean (Churn)', 'Mean (Non-Churn)', 'Median (Churn)', 'Median (Non-Churn)', 'Mean Difference Direction']
]
results_df.sort_values(by='p-value', inplace=True, ignore_index=True)
print(results_df)
````

### **Advanced Exploration of Churn Based on Transaction Behavior**

#### Main Objectives:

This extended exploration aims to:

* Identify **customer segments with high churn risk** based on the combination of two key transactional factors.
* Evaluate **interactions between incentives** such as cashback and coupons.
* Explore the **customer lifecycle behavior** based on the number of orders placed.
* Examine the relationship between **cashback and customer engagement** through transaction volume.

This analysis is crucial to help the company design more targeted retention strategies, grounded in a deep understanding of customers' actual transactional behavior.

#### **1. Risk Segmentation Based on Cashback and Inactivity**

This analysis leverages two key variables—`CashbackAmount` and `DaySinceLastOrder`—which have proven to be significant in influencing churn. Customers are grouped into four combined segments to examine how differences in these characteristics impact churn proportions.

#### **2. Cashback vs CouponUsed Interaction**

This analysis explores whether there is a synergistic or substitutive effect between two types of incentives: cashback and coupons. Customers are segmented based on high/low combinations of `CashbackAmount` and `CouponUsed`, using median values as thresholds.

#### **3. Lifecycle Analysis Based on Order Count**

Here, customers are classified based on the total number of orders they have placed (1, 2, or 3+ times). The goal is to evaluate at which point churn is most likely to occur throughout the customer lifecycle.

#### **4. OrderCount Distribution Based on Cashback Buckets**

Customer cashback is divided into three categories using quantile-based bucketing (`qcut`): Low, Medium, and High. For each bucket, the average number of orders (`OrderCount`) placed by customers is calculated, separated by churn status.

<br>


````python
# ===========================================
# Transaction Behavior – Extended Exploration
# Dataset: df_temp
# ===========================================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# ===========================================
# 1. Risk Segmentation: Cashback vs Inactivity
# ===========================================

cashback_median = df_temp['CashbackAmount'].median()
inactivity_median = df_temp['DaySinceLastOrder'].median()

def define_risk_segment(row):
    if row['CashbackAmount'] <= cashback_median and row['DaySinceLastOrder'] > inactivity_median:
        return 'Low Cashback + Long Inactivity'
    elif row['CashbackAmount'] <= cashback_median and row['DaySinceLastOrder'] <= inactivity_median:
        return 'Low Cashback + Short Inactivity'
    elif row['CashbackAmount'] > cashback_median and row['DaySinceLastOrder'] > inactivity_median:
        return 'High Cashback + Long Inactivity'
    else:
        return 'High Cashback + Short Inactivity'

df_temp['ChurnRiskSegment'] = df_temp.apply(define_risk_segment, axis=1)

seg_churn = df_temp.groupby('ChurnRiskSegment')['Churn'].value_counts(normalize=True).rename("Rate").reset_index()
seg_churn['Rate'] = seg_churn['Rate'] * 100

plt.figure(figsize=(10, 6))
sns.barplot(data=seg_churn, x='ChurnRiskSegment', y='Rate', hue='Churn')
plt.title('Churn Rate by Risk Segment (Cashback + Inactivity)')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=20)
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# ===========================================
# 2. Interaction: Cashback vs Coupon Used
# ===========================================

coupon_median = df_temp['CouponUsed'].median()

def cashback_coupon_combo(row):
    if row['CashbackAmount'] > cashback_median and row['CouponUsed'] > coupon_median:
        return 'High Cashback + High Coupon'
    elif row['CashbackAmount'] > cashback_median and row['CouponUsed'] <= coupon_median:
        return 'High Cashback + Low Coupon'
    elif row['CashbackAmount'] <= cashback_median and row['CouponUsed'] > coupon_median:
        return 'Low Cashback + High Coupon'
    else:
        return 'Low Cashback + Low Coupon'

df_temp['CashbackCouponSegment'] = df_temp.apply(cashback_coupon_combo, axis=1)

coupon_churn = df_temp.groupby('CashbackCouponSegment')['Churn'].value_counts(normalize=True).rename('Rate').reset_index()
coupon_churn['Rate'] = coupon_churn['Rate'] * 100

plt.figure(figsize=(10, 6))
sns.barplot(data=coupon_churn, x='CashbackCouponSegment', y='Rate', hue='Churn')
plt.title('Churn Rate by Cashback & Coupon Segments')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=15)
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# ===========================================
# 3. Lifecycle Cohort: Order Count
# ===========================================

def order_lifecycle(order):
    if order == 1:
        return '1x Order'
    elif order == 2:
        return '2x Order'
    else:
        return '3+ Orders'

df_temp['OrderCountCohort'] = df_temp['OrderCount'].apply(order_lifecycle)

cohort_churn = df_temp.groupby('OrderCountCohort')['Churn'].value_counts(normalize=True).rename('Rate').reset_index()
cohort_churn['Rate'] = cohort_churn['Rate'] * 100

plt.figure(figsize=(8, 5))
sns.barplot(data=cohort_churn, x='OrderCountCohort', y='Rate', hue='Churn')
plt.title('Churn Rate by Lifecycle Cohort (Order Count)')
plt.ylabel('Churn Rate (%)')
plt.xlabel('Order Cohort')
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# ===========================================
# 4. Order Count Distribution by Cashback Bucket
# ===========================================

df_temp['CashbackBucket'] = pd.qcut(df_temp['CashbackAmount'], q=3, labels=['Low', 'Medium', 'High'])

cb_order = df_temp.groupby(['CashbackBucket', 'Churn'])['OrderCount'].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=cb_order, x='CashbackBucket', y='OrderCount', hue='Churn')
plt.title('Average Order Count by Cashback Bucket & Churn')
plt.ylabel('Avg Order Count')
plt.xlabel('Cashback Bucket')
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()
````

#### **1. Churn Risk Segmentation Based on Cashback and Inactivity**

This visualization shows the churn distribution across four combined segments of `CashbackAmount` and `DaySinceLastOrder`.

**Insight:**

* Customers with **Low Cashback + Short Inactivity** have the highest churn rate, indicating that even though they are still active, **the lack of financial incentives is a churn trigger**.
* The **Low Cashback + Long Inactivity** segment also shows high risk, suggesting a cumulative effect of low rewards and declining engagement.
* The **High Cashback** segments are relatively more stable, both among active and inactive users, reinforcing that **cashback has a protective effect against churn**.

**Recommendation:**

* Prioritize customers with **low cashback** in retargeting campaigns.
* Send special incentives or notifications to customers with high `DaySinceLastOrder` and below-median cashback.

#### **2. Churn Rate by Cashback and Coupon Interaction**

This visualization compares churn rates based on combinations of `CashbackAmount` and `CouponUsed`.

**Insight:**

* Customers with **High Cashback**, even with minimal coupon usage, still show low churn rates.
* **Coupons are not effective substitutes for cashback**—the **Low Cashback + High Coupon** segment still shows a high churn rate.
* This indicates that **cashback has a stronger and more consistent effect in preventing churn**, while coupons serve only as a secondary support.

**Recommendation:**

* Do not rely on coupons as the sole form of incentive.
* Combine coupons with cashback to maximize their effectiveness in customer retention.

#### **3. Lifecycle Churn Analysis by Order Count**

This visualization groups customers based on their total number of orders (`OrderCount`) to analyze when churn most frequently occurs.

**Insight:**

* The highest churn rate is found among **customers who only made 1 order**.
* The churn risk **drops significantly for customers who reach their 2nd and 3rd transactions**.
* This highlights the importance of transitioning customers from one-time buyers to repeat purchasers.

**Recommendation:**

* Launch **onboarding and loyalty acceleration programs** specifically after the first transaction.
* Offer repeat order incentives or platform benefit education to encourage customers to reach their 2nd purchase and beyond.

#### **4. Average Order Count by Cashback Bucket**

This visualization compares the average `OrderCount` across three `CashbackAmount` buckets (Low, Medium, High), separated by churn status.

**Insight:**

* The higher the cashback received, the **higher the average number of orders placed**.
* This pattern holds true for **both churned and non-churned customers**.
* This indicates that **cashback not only reduces churn but also drives deeper engagement throughout the customer lifecycle**.

**Recommendation:**

* Use `CashbackAmount` as a strategic tool to increase Average Order Value and Frequency.
* Segment customers with low cashback for intensive engagement campaigns to extend their loyalty lifecycle.

### **II.3.4 Churn by Platform & Purchase Preferences**

#### **Objective of the Analysis**

This analysis aims to evaluate whether customer preferences in **accessing the platform**, **payment methods**, and **preferred order categories** are associated with the tendency to *churn*.

By understanding how customers prefer to interact with the platform, the company can optimize service channels and tailor marketing strategies to improve retention.

#### **What Are Platform & Purchase Preferences?**

This category reflects **technical and purchasing behavior preferences** of customers, including:

* How customers **log in or access the service** (web vs. mobile)
* Preferred **payment methods** (COD, credit card, e-wallet, etc.)
* **Types of products** or order categories most frequently selected

These features provide insight into customers’ **shopping styles** and **user convenience**, which may influence loyalty.

#### **Supported Business Questions**

* Do customers with certain platform preferences (e.g., logging in via mobile) have lower churn tendencies?
* Does payment method influence loyalty?
* Are product category preferences related to churn levels?

#### **Features Analyzed & Reason for Selection**

| Feature                       | Data Type (Statistical) | Categorization Reason & Relevance to Transaction Behavior                                                        |
| ----------------------------- | ----------------------- | ---------------------------------------------------------------------------------------------------------------- |
| `OrderCount`                  | Discrete Numeric        | Represents the **purchase frequency** by the customer. High frequency indicates higher transactional engagement. |
| `OrderAmountHikeFromlastYear` | Continuous Numeric      | Reflects **annual spending changes**, which may indicate increasing or decreasing loyalty.                       |
| `CouponUsed`                  | Discrete Numeric        | Indicates the **number of coupons used** by the customer. Reflects sensitivity to promotions or discounts.       |
| `DaySinceLastOrder`           | Continuous Numeric      | Measures the **duration since the last purchase**. The longer the gap, the closer the customer may be to churn.  |
| `CashbackAmount`              | Continuous Numeric      | Represents the **financial incentive** received by the customer, which may increase retention if substantial.    |

**Urgency of the analysis:**
These preferences are **behavioral & technical** in nature and often play an important role in the **daily user experience**. A mismatch between customer preferences and available features or services can increase churn risk.

### **Analysis Method**

#### **Stages & Approach**

| Stage           | Approach                                                                            |
| --------------- | ----------------------------------------------------------------------------------- |
| **Univariate**  | Distribution of categories in each feature without considering churn                |
| **Bivariate**   | Comparison of churn proportions within each category (stacked barplot)              |
| **Statistical** | **Chi-Square Test** to assess associations between each preference and churn status |

#### **Visualizations Used**

| Visualization Type | Purpose                                                   |
| ------------------ | --------------------------------------------------------- |
| Barplot            | View overall dominance of preference categories           |
| Stacked Barplot    | Show churn vs. non-churn proportions within each category |

#### **Chi-Square Test Explanation**

**Chi-Square Test** is used because the features being analyzed are **categorical**.

* **Hypotheses:**

  * H₀: There is no association between the preference feature and churn
  * H₁: There is a significant association between the preference feature and churn

* **Decision Rule:**

  * p-value < 0.05 → Reject H₀ → A significant association exists

<br>



````python
# ============================================
# II.3.4 – Churn by Platform & Purchase Preferences
# Dataset: df
# ============================================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency

sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# 1. Define the platform & purchase preference features
preference_features = [
    'PreferredLoginDevice',
    'PreferedOrderCat',
    'PreferredPaymentMode'  # corrected spelling
]

# 2. Univariate Analysis – Distribution per feature
for feature in preference_features:
    plt.figure()
    sns.countplot(data=df, x=feature, order=df[feature].value_counts().index, palette='pastel')
    plt.title(f'Distribution of {feature}')
    plt.ylabel('Number of Customers')
    plt.xlabel(feature)
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

# 3. Bivariate Analysis – Churn proportion per category
for feature in preference_features:
    churn_dist = df.groupby(feature)['Churn'].value_counts(normalize=True).rename('Proportion').reset_index()
    churn_dist['Proportion'] *= 100

    plt.figure()
    sns.barplot(data=churn_dist, x=feature, y='Proportion', hue='Churn', palette='muted')
    plt.title(f'Churn Proportion by {feature}')
    plt.ylabel('Proportion (%)')
    plt.xlabel(feature)
    plt.xticks(rotation=20)
    plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
    plt.tight_layout()
    plt.show()

# 4. Chi-Square Statistical Test
print("=========== Chi-Square Test Results ===========")
chi2_results = []

for feature in preference_features:
    contingency = pd.crosstab(df[feature], df['Churn'])
    chi2, p, dof, expected = chi2_contingency(contingency)

    chi2_results.append({
        'Feature': feature,
        'Chi2 Statistic': round(chi2, 3),
        'p-value': round(p, 4),
        'Significant (p < 0.05)': 'Yes' if p < 0.05 else 'No'
    })

# Convert to DataFrame for display
chi2_df = pd.DataFrame(chi2_results)
print(chi2_df)
````

### **Advanced Exploration of Churn Based on Platform & Purchase Preferences**

### **Analytical Focus & Exploration Series:**

This exploration consists of five main parts, each targeting customer preference interactions potentially related to churn:

1. **Combined Preference Segmentation**
   Customers are grouped based on the combination of three main preferences:

   * **Preferred Order Category:** The type of product most frequently purchased.

2. **Payment Method (COD) and Inactivity Interaction**
   This explores the relationship between the **Cash on Delivery (COD)** payment method and the customer's **inactivity duration** (days since last transaction). The visualization helps determine whether COD users exhibit different inactivity patterns compared to users of other payment methods.

   Combines two important dimensions—**login device** and **order frequency**—to evaluate churn proportions in each segment combination. This aims to determine whether access style and transactional engagement together correlate with churn.

````python
# ===========================================
# Advanced Churn Exploration by Platform & Preferences
# Dataset: df_temp (for feature creation)
# ===========================================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# ===============================
# 1. Preference Combination Segmentation
# ===============================

# Combine 3 preference features into a single segment column
df_temp['PreferenceSegment'] = (
    df_temp['PreferredLoginDevice'] + ' | ' +
    df_temp['PreferedOrderCat'] + ' | ' +
    df_temp['PreferredPaymentMode']
)

# Calculate churn proportion per segment
segment_churn = df_temp.groupby('PreferenceSegment')['Churn'].value_counts(normalize=True).rename('Proportion').reset_index()
segment_churn['Proportion'] = segment_churn['Proportion'] * 100

# Take the top 10 most common segments
top_segments = df_temp['PreferenceSegment'].value_counts().nlargest(10).index
segment_churn_top = segment_churn[segment_churn['PreferenceSegment'].isin(top_segments)]

# Visualization
plt.figure(figsize=(14, 6))
sns.barplot(data=segment_churn_top, x='PreferenceSegment', y='Proportion', hue='Churn')
plt.title('Churn Proportion by Top 10 Preference Segments')
plt.ylabel('Churn Proportion (%)')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# ===============================
# 2. Interaction: COD vs Inactivity
# ===============================

# Create a new column to label COD users
df_temp['COD_User'] = df_temp['PreferredPaymentMode'].apply(lambda x: 'COD' if x == 'Cash on Delivery' else 'Non-COD')

plt.figure(figsize=(8, 5))
sns.boxplot(data=df_temp, x='COD_User', y='DaySinceLastOrder', hue='Churn')
plt.title('Inactivity Days by Payment Method (COD vs Non-COD)')
plt.ylabel('Days Since Last Order')
plt.xlabel('Payment Type')
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# ===============================
# 3. Interaction: Login Device vs Cashback Received
# ===============================

plt.figure(figsize=(8, 5))
sns.boxplot(data=df_temp, x='PreferredLoginDevice', y='CashbackAmount', hue='Churn')
plt.title('Cashback Distribution by Login Device and Churn')
plt.ylabel('Cashback Amount')
plt.xlabel('Login Device')
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# ===============================
# 4. Preference vs Order Count (Lifecycle)
# ===============================

# Categorize order frequency
def order_category(n):
    if n == 1:
        return '1x'
    elif n == 2:
        return '2x'
    else:
        return '3x+'

df_temp['OrderFreqGroup'] = df_temp['OrderCount'].apply(order_category)

plt.figure(figsize=(8, 5))
sns.countplot(data=df_temp, x='OrderFreqGroup', hue='PreferredLoginDevice')
plt.title('Order Frequency by Login Device Preference')
plt.xlabel('Order Frequency Group')
plt.ylabel('Customer Count')
plt.legend(title='Login Device')
plt.tight_layout()
plt.show()

# ===============================
# 5. Churn Rate by Device + Order Frequency
# ===============================

df_temp['Device_OrderGroup'] = df_temp['PreferredLoginDevice'] + ' | ' + df_temp['OrderFreqGroup']

combo_churn = df_temp.groupby('Device_OrderGroup')['Churn'].value_counts(normalize=True).rename("Rate").reset_index()
combo_churn['Rate'] = combo_churn['Rate'] * 100

plt.figure(figsize=(10, 6))
sns.barplot(data=combo_churn, x='Device_OrderGroup', y='Rate', hue='Churn')
plt.title('Churn Rate by Login Device + Order Frequency')
plt.xticks(rotation=30, ha='right')
plt.ylabel('Churn Rate (%)')
plt.tight_layout()
plt.show()
````

### **II.3.5 Other Customer Experience Factors**

#### **Analysis Objective**

This analysis aims to evaluate **additional factors** that may influence churn but are not included in the categories of demographics, usage behavior, transactions, or platform preferences.

The focus is on **logistical aspects** and **technical usage**, which may indirectly reflect customer experience yet significantly affect their decision to continue using the service.

#### **What Are Other Customer Experience Factors?**

This category includes features that represent **logistical convenience**, **multi-device usage**, and **data flexibility**, all of which can affect the customer's perception of ease and efficiency:

* Delivery distance from warehouse to home (`WarehouseToHome`)
* Number of devices used to log in (`NumberOfDeviceRegistered`)
* Number of addresses used by the customer (`NumberOfAddress`)

Although these features do not directly indicate shopping behavior or daily engagement, they provide insight into the **complexity and flexibility of the user experience**.

#### **Supported Business Questions**

* Are customers with more complex logistical experiences (e.g., long delivery distances) more likely to churn?
* Do customers who access through multiple devices show higher loyalty?
* Does the number of addresses affect churn?

#### **Analyzed Features & Reasons for Inclusion**

| Feature                    | Data Type            | Categorization Rationale & Relevance to Customer Experience                                                       |
| -------------------------- | -------------------- | ----------------------------------------------------------------------------------------------------------------- |
| `WarehouseToHome`          | Continuous Numerical | Represents estimated **distance or logistical difficulty**. Longer delivery may impact satisfaction.              |
| `NumberOfDeviceRegistered` | Discrete Numerical   | Number of **devices connected** to the customer account. Indicates engagement intensity and multi-platform usage. |
| `NumberOfAddress`          | Discrete Numerical   | Number of **addresses used** for delivery. Could indicate flexibility or service complexity.                      |

**Analysis Urgency:**
These features can provide **operational signals** about obstacles or ease experienced by customers during service usage.

### **Analysis Methodology**

#### **Stages & Approaches**

| Stage           | Approach                                                                   |
| --------------- | -------------------------------------------------------------------------- |
| **Univariate**  | Explore the overall distribution of each feature (histogram)               |
| **Bivariate**   | Compare feature values between churned and non-churned customers (boxplot) |
| **Statistical** | Difference testing between two groups:                                     |
|                 | - **t-test** if data is normally distributed                               |
|                 | - **Mann-Whitney U Test** if not normally distributed                      |

#### **Visualizations Used**

| Visualization Type | Purpose                                                               |
| ------------------ | --------------------------------------------------------------------- |
| Boxplot            | Observe differences in medians and distributions between churn groups |
| Histogram          | Understand the overall distribution of each feature                   |

#### **Explanation of Statistical Test**

**t-test / Mann-Whitney U Test**
Used to compare numerical feature values between churned and non-churned customers.

* **Hypotheses:**

  * H₀: There is no significant difference between churned and non-churned groups
  * H₁: There is a significant difference

* **Decision Rule:**

  * If p-value < 0.05 → Reject H₀ → There is a significant difference

**Note:**
The choice between the t-test and the Mann-Whitney U Test is based on the results of a prior normality test. For non-normal or outlier-heavy data, the Mann-Whitney U Test is recommended.

<br>


````python
# ===========================================
# II.3.5 – Other Customer Experience Factors
# Dataset: df
# ===========================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import mannwhitneyu, ttest_ind, normaltest

# Set styling
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

# ============================
# 1. Feature List
# ============================
experience_features = ['WarehouseToHome', 'NumberOfDeviceRegistered', 'NumberOfAddress']

# ============================
# 2. Univariate: Distribution Histogram
# ============================
for feature in experience_features:
    plt.figure()
    sns.histplot(data=df, x=feature, kde=True, bins=20, color='#A3C4F3')
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Number of Customers')
    plt.tight_layout()
    plt.show()

# ============================
# 3. Bivariate: Boxplot by Churn
# ============================
for feature in experience_features:
    plt.figure()
    sns.boxplot(data=df, x='Churn', y=feature, palette='pastel')
    plt.title(f'{feature} by Churn Status')
    plt.xlabel('Churn')
    plt.ylabel(feature)
    plt.xticks([0, 1], ['Not Churned', 'Churned'])
    plt.tight_layout()
    plt.show()

# ============================
# 4. Statistical Test (Normality + Mann-Whitney or t-test)
# ============================

print("\n========== Statistical Test Results ==========")

results = []

for feature in experience_features:
    # Separate by churn group
    churned = df[df['Churn'] == 1][feature]
    not_churned = df[df['Churn'] == 0][feature]

    # Normality test
    stat_c, p_c = normaltest(churned)
    stat_nc, p_nc = normaltest(not_churned)
    is_normal = (p_c > 0.05) and (p_nc > 0.05)

    # Use t-test if normal, otherwise Mann-Whitney U Test
    if is_normal:
        test_stat, p_value = ttest_ind(churned, not_churned, equal_var=False)
        test_type = 't-test'
    else:
        test_stat, p_value = mannwhitneyu(churned, not_churned, alternative='two-sided')
        test_type = 'Mann-Whitney U'

    results.append({
        'Feature': feature,
        'Test Type': test_type,
        'Statistic': round(test_stat, 3),
        'p-value': round(p_value, 4),
        'Significant (p < 0.05)': 'Yes' if p_value < 0.05 else 'No'
    })

# Convert to DataFrame for display
stat_results_df = pd.DataFrame(results)
print(stat_results_df.to_string(index=False))
````

### **Interpretation of Findings: Logistical Complexity & Multi-Device Usage Strongly Linked to Churn Risk**

#### **Customer Insight**

Based on data distribution, boxplots, and statistical test results (Mann-Whitney U), it was found that all three customer experience features—`WarehouseToHome`, `NumberOfDeviceRegistered`, and `NumberOfAddress`—have a statistically significant relationship with customer churn status, with p-values < 0.05 for all variables.

1. **WarehouseToHome (Delivery Distance):**

   * The median and spread of delivery distance are significantly higher among churned customers compared to active ones.
   * This indicates that **the farther the delivery distance, the higher the risk of churn**.
   * The distribution is right-skewed, where most customers have short distances, but churned customers tend to fall in higher distance ranges.

2. **NumberOfDeviceRegistered (Number of Devices):**

   * Churned customers have a **higher median number of login devices**.
   * This may indicate two possibilities: (1) the customer actively uses many devices but is dissatisfied, or (2) the account is used across multiple devices but does not deliver a consistent experience.
   * In other words, **usage intensity does not guarantee loyalty** if the experience is unsatisfactory.

3. **NumberOfAddress (Number of Addresses):**

   * Churned customers have **more shipping addresses** than non-churned ones.
   * This reflects that **logistical complexity or usage flexibility does not always have a positive impact**, and may even cause service friction that leads to churn.

#### **Strategic Retention Recommendations**

1. **Optimize Delivery Chain for Remote Areas**

   * Focus on regions with long delivery distances and high churn rates.
   * Improve delivery estimations, integrate with local logistics partners, or create special programs for customers in non-urban areas.

2. **Monitor Multi-Device Usage**

   * Implement friction detection across devices, such as unexpected logouts or unresponsive features on certain devices.
   * Provide notifications or a feedback system specifically for customers with more than three login devices.

3. **Simplify Process for Customers with Multiple Addresses**

   * Offer features such as a “default address” option, automatic reminders for the last used address, or segment customers with many addresses for a needs reassessment.
   * This can reduce delivery errors and improve logistics satisfaction.

4. **Proactive Segmentation**

   * Customers with a combination of **WarehouseToHome > 20**, **NumberOfDeviceRegistered ≥ 4**, and **NumberOfAddress ≥ 5** should be included in a **high churn risk cohort** and prioritized for user experience-based intervention strategies.

This interpretation highlights that technical and operational complexity—though seemingly “flexible”—can in fact become a churn risk factor if not accompanied by a seamless experience. Retention strategies should target pain points in logistics and multi-device interfaces to ensure customers feel both comfortable and loyal.

#### **Advanced Exploration: Interaction and Segmentation**

This section aims to deepen the analysis of **additional customer experience factors** that are not included in the main categories such as demographics, behavioral usage, transaction activity, or platform preferences. The focus is on evaluating how **logistical and technical complexity** in the customer journey may influence churn behavior.

The analysis focuses on three key features:

* `WarehouseToHome`: Distance from the warehouse to the customer's delivery address
* `NumberOfAddress`: Number of shipping addresses associated with the customer
* `NumberOfDeviceRegistered`: Number of devices used to access the platform

These variables serve as proxies for **operational or usage complexity**, which may affect user satisfaction and retention indirectly.

#### **Exploration Structure**

1. **Bivariate Interaction (Scatterplot)**
   A scatterplot is used to visualize the interaction between `WarehouseToHome` and `NumberOfAddress`, segmented by churn status. This helps identify whether churn patterns emerge across combinations of shipping distance and address complexity.

2. **Categorical Interaction via Heatmap**
   To facilitate interpretability, both `WarehouseToHome` and `NumberOfAddress` are discretized into three groups: *Near–Medium–Far* and *Few–Moderate–Many*. The heatmap then shows churn rates across each combination, highlighting possible interaction effects between delivery and address complexity.

3. **Multi-Factor Segmentation Based on Experience Complexity**
   Each of the three features is categorized into *Low*, *Medium*, and *High* complexity levels. These levels are combined into a new segmentation feature called `ExperienceSegment`. The top 10 most common segments are then analyzed to calculate and visualize their respective churn proportions.
re targeted retention strategies and segmentation models.

<br>


````python
# ============================================
# II.3.5 – Other Customer Experience Factors
# Advanced Exploration: Interaction and Segmentation
# Dataset: df & df_temp
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual styling
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# ============================================
# 1. Interaction: Warehouse Distance vs Number of Addresses
# Dataset: df (no new columns added)
# ============================================

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='WarehouseToHome', y='NumberOfAddress', hue='Churn', alpha=0.6, palette='Set1')
plt.title('Warehouse Distance vs Number of Addresses by Churn')
plt.xlabel('WarehouseToHome')
plt.ylabel('NumberOfAddress')
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()

# ============================================
# 2. Interaction Heatmap: Warehouse Group vs Address Group
# Dataset: df_temp (temporary grouping columns added)
# ============================================

df_temp = df.copy()

# Create categorical groupings for warehouse distance and address count
df_temp['WH_Group'] = pd.cut(df_temp['WarehouseToHome'], bins=[0, 10, 20, df_temp['WarehouseToHome'].max()],
                              labels=['Near', 'Medium', 'Far'])
df_temp['Addr_Group'] = pd.cut(df_temp['NumberOfAddress'], bins=[0, 3, 6, df_temp['NumberOfAddress'].max()],
                                labels=['Few', 'Moderate', 'Many'])

# Calculate churn rate by group combination
heat_data = df_temp.groupby(['WH_Group', 'Addr_Group'])['Churn'].mean().unstack().round(2)

plt.figure(figsize=(7, 5))
sns.heatmap(heat_data, annot=True, fmt='.2f', cmap='YlOrBr')
plt.title('Churn Rate by Warehouse Group and Address Group')
plt.xlabel('Number of Address Group')
plt.ylabel('Warehouse Distance Group')
plt.tight_layout()
plt.show()

# ============================================
# 3. Customer Segmentation Based on Experience Complexity
# Dataset: df_temp (creating exploratory segment combinations)
# ============================================

# Categorize experience complexity
df_temp['LogisticComplexity'] = pd.cut(df_temp['WarehouseToHome'], bins=[0, 10, 20, df_temp['WarehouseToHome'].max()],
                                       labels=['Low', 'Medium', 'High'])

df_temp['DeviceUsage'] = pd.cut(df_temp['NumberOfDeviceRegistered'], bins=[0, 2, 4, df_temp['NumberOfDeviceRegistered'].max()],
                                labels=['Low', 'Medium', 'High'])

df_temp['AddressComplexity'] = pd.cut(df_temp['NumberOfAddress'], bins=[0, 2, 5, df_temp['NumberOfAddress'].max()],
                                      labels=['Low', 'Medium', 'High'])

# Combine into a unified experience complexity segment
df_temp['ExperienceSegment'] = df_temp['LogisticComplexity'].astype(str) + ' | ' + \
                               df_temp['DeviceUsage'].astype(str) + ' | ' + \
                               df_temp['AddressComplexity'].astype(str)

# Select top 10 most frequent segments
top_segments = df_temp['ExperienceSegment'].value_counts().nlargest(10).index
segment_data = df_temp[df_temp['ExperienceSegment'].isin(top_segments)]

# Compute churn proportion within each segment
seg_churn = segment_data.groupby('ExperienceSegment')['Churn'].value_counts(normalize=True).rename('Proportion').reset_index()
seg_churn['Proportion'] = seg_churn['Proportion'] * 100

# Visualize churn proportion by segment
plt.figure(figsize=(14, 6))
sns.barplot(data=seg_churn, x='ExperienceSegment', y='Proportion', hue='Churn')
plt.title('Churn Proportion by Top 10 Experience Segments')
plt.ylabel('Churn Proportion (%)')
plt.xlabel('Experience Segment (Warehouse | Device | Address)')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Churn', labels=['Not Churned', 'Churned'])
plt.tight_layout()
plt.show()
````

### **Interpretation – Interaction and Segmentation Analysis of Customer Experience Factors**

The extended analysis of customer experience dimensions—specifically warehouse distance, device usage, and number of delivery addresses—reveals key behavioral segments and interaction patterns that influence churn rates. The interpretation can be broken down into the following insights:

#### **1. Interaction: Warehouse Distance vs Number of Address**

The scatterplot shows a weak but visible relationship between **delivery distance and number of addresses**. Although these two features are not strongly linearly related, customers who churn tend to appear more frequently in regions with **either longer delivery distances or a higher number of addresses**. This suggests that **logistical complexity may contribute to churn behavior**.

#### **2. Churn Rate by Combined Group: Warehouse Distance & Address Count**

The heatmap visualizing churn rates across **categorical groupings of delivery distance** (`Near`, `Medium`, `Far`) and **address count** (`Few`, `Moderate`, `Many`) reveals several patterns:

* Churn rate is consistently **higher in the 'Many Address' group**, regardless of warehouse distance (0.23 across all).
* Among customers with `Few` or `Moderate` addresses, churn increases slightly with distance:

  * From **0.13 (Near, Few)** to **0.22 (Far, Few)**
  * From **0.10 (Near, Moderate)** to **0.17 (Far, Moderate)**

This suggests that **both longer delivery distances and a higher number of addresses correlate with increased churn probability**, although the number of addresses seems to have a stronger effect.

#### **3. Churn Proportion by Experience Segment (Warehouse | Device | Address)**

The segmentation plot, which classifies customers into 10 most common **combinations of delivery complexity, device intensity, and address complexity**, confirms that:

* **All segments exhibit churn**, but the **highest churn proportions** are observed in segments with:

  * **High warehouse distance**
  * **High or medium device usage**
  * **High address complexity**
* Segments like **`High | Medium | High`** and **`Low | Medium | High`** show elevated churn rates above 20%.

This suggests that customers experiencing **higher logistical friction (distance and addresses)**, even when actively using the platform across devices, may still churn. This might be due to service inefficiencies or inconvenience in managing multiple delivery setups.

### **Summary of Insights**

* **WarehouseToHome** and **NumberOfAddress** are independently and jointly associated with higher churn likelihood.
* Customers in the **"Many Address"** group represent a key risk segment, even if delivery distance is near.
* The **segmentation analysis** provides a scalable framework to monitor and target high-risk behavioral clusters.
* These findings reinforce the importance of **logistics optimization** and **user profile simplification (address/device management)** as part of retention strategy planning.

**Analysis of User Behavior Based on App Usage Duration**

This analysis focuses on understanding user behavior based on how long they spend on the platform each day. Instead of looking only at preferred login devices, we explore engagement patterns across multiple user attributes:

- Device used to access the platform
- Gender
- Marital status

By analyzing average usage duration through these dimensions, we aim to identify meaningful trends in user engagement and uncover differences in behavior across various demographic and access groups.

````python
# --- Helper function for grouping and plotting ---
def plot_avg_minutes(df, group_col, x_label, ax, palette, title):
    avg_df = df.groupby(group_col)['HourSpendOnApp'].mean().mul(60).reset_index()
    avg_df.columns = [x_label, 'Minute']

    # Plot barplot
    sns.barplot(x=x_label, y='Minute', data=avg_df, ax=ax, palette=palette)

    # Visualize Number with rounding
    for p in ax.patches:
        # Get the height of the bar and round it to 2 decimal places
        value = round(p.get_height())
        ax.text(p.get_x() + p.get_width() / 2., p.get_height(),
                f'{value}',  # Display rounded value
                ha='center', va='center', fontsize=12, color='black', fontweight='bold')

    ax.set_title(title)
    ax.set_xlabel(x_label)
    ax.set_ylabel('Average Minutes')

# --- Prepare plot area ---
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# --- Define configurations for each grouping ---
groupings = [
    ('PreferredLoginDevice', 'Device', axes[0, 0], 'Blues', 'Average Minutes by Device'),
    ('Gender', 'Gender', axes[0, 1], 'Oranges', 'Average Minutes by Gender'),
    ('MaritalStatus', 'MaritalStatus', axes[1, 0], 'Greens', 'Average Minutes by Marital Status'),
    ('CityTier', 'CityTier', axes[1, 1], 'Purples', 'Average Minutes by City Tier')
]

# --- Loop through each config and plot ---
for group_col, x_label, ax, palette, title in groupings:
    plot_avg_minutes(loyal_df, group_col, x_label, ax, palette, title)

plt.tight_layout()
plt.show()
````

**Analysis of Order Behavior Based on Customer Sastisfaction**

This analysis focuses on understanding user behavior based on User Satisfaction, we explore engagement patterns across multiple user attributes:

- Complain
- Satisfaction Score
- Total Order
- Coupon Used

By analyzing orders and satisfaction scores, we aim to gain insights into the characteristics of satisfied and unsatisfied customers.

````python
# --- Setup style ---
sns.set(style="whitegrid", palette="Blues")

# --- Setup 3x2 plotting grid ---
fig, axes = plt.subplots(3, 2, figsize=(18, 16))

# --- Plot 1: Pie chart of Complain vs No Complain ---
complain_counts = df['Complain'].value_counts()
labels = ['No Complaint', 'Complaint'] if 0 in complain_counts.index else complain_counts.index
sizes = complain_counts.values
total = sum(sizes)
percentages = [f'{label}: {count} ({count/total*100:.1f}%)' for label, count in zip(labels, sizes)]

axes[0, 0].pie(
    sizes,
    labels=percentages,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.color_palette("Blues", len(sizes))
)
axes[0, 0].set_title('Complaint vs No Complaint')
axes[0, 0].axis('equal')

# --- Plot 2: Complaint Rate by Satisfaction Score ---
satisfaction_group = df.groupby('SatisfactionScore')['Complain'].agg(['count', 'sum']).reset_index()
satisfaction_group['ComplaintRate'] = satisfaction_group['sum'] / satisfaction_group['count'] * 100

sns.barplot(x='SatisfactionScore', y='ComplaintRate', data=satisfaction_group, ax=axes[0, 1], palette="Blues")
axes[0, 1].set_title('Complaint Rate by Satisfaction Score')
axes[0, 1].set_ylabel('Complaint Rate (%)')
axes[0, 1].set_xlabel('Satisfaction Score')

for i, row in satisfaction_group.iterrows():
    axes[0, 1].text(i, row['ComplaintRate'] + 1, f"{int(row['count'])} customers", ha='center', fontsize=9)

# --- Plot 3: Average Satisfaction Score by Order Count ---
avg_score_by_orders = df.groupby('OrderCount')['SatisfactionScore'].mean().reset_index()
sns.barplot(x='OrderCount', y='SatisfactionScore', data=avg_score_by_orders, ax=axes[1, 0], palette="Blues")
axes[1, 0].set_title('Average Satisfaction Score by Total Orders')
axes[1, 0].set_ylabel('Avg Satisfaction Score')
axes[1, 0].set_xlabel('Total Orders')
axes[1, 0].tick_params(axis='x', rotation=45)

# --- Plot 4: Average Satisfaction Score by Coupon Used ---
avg_score_by_coupon = df.groupby('CouponUsed')['SatisfactionScore'].mean().reset_index()
sns.barplot(x='CouponUsed', y='SatisfactionScore', data=avg_score_by_coupon, ax=axes[1, 1], palette="Blues")
axes[1, 1].set_title('Average Satisfaction Score by Coupon Used')
axes[1, 1].set_ylabel('Avg Satisfaction Score')
axes[1, 1].set_xlabel('Coupon Used')

# --- Plot 5: Average Cashback Amount by Satisfaction Score ---
avg_cashback_by_satisfaction = df.groupby('SatisfactionScore')['CashbackAmount'].mean().reset_index()
sns.barplot(x='SatisfactionScore', y='CashbackAmount', data=avg_cashback_by_satisfaction, ax=axes[2, 0], palette="Blues")
axes[2, 0].set_title('Average Cashback Amount by Satisfaction Score')
axes[2, 0].set_ylabel('Avg Cashback Amount')
axes[2, 0].set_xlabel('Satisfaction Score')

# --- Plot 6: Average Satisfaction Score by City Tier ---
avg_score_by_city = df.groupby('CityTier')['SatisfactionScore'].mean().reset_index()
sns.barplot(x='CityTier', y='SatisfactionScore', data=avg_score_by_city, ax=axes[2, 1], palette="Blues")
axes[2, 1].set_title('Average Satisfaction Score by City Tier')
axes[2, 1].set_ylabel('Avg Satisfaction Score')
axes[2, 1].set_xlabel('City Tier')

# --- Final Layout ---
plt.tight_layout()
plt.show()
````

### **II.4  Exploratory Data Analysis (EDA) for Machine Learning**

**Exploratory Data Analysis (EDA)** is a fundamental step in the **machine learning workflow**, particularly important in **classification problems** like **churn prediction**. It helps uncover the **distribution**, **relationships**, and **anomalies** in the data, allowing for a deeper understanding of how features may influence the **target variable**.

In this project, **EDA** focused on examining the **distribution of the target variable (`Churn`)**, exploring **numerical and categorical features**, and analyzing **correlations** among features. Visual tools such as **bar plots**, **boxplots**, and **correlation heatmaps** were employed to highlight **patterns** and **potential predictors of churn**. These insights guided **feature selection**, informed **modeling strategies**, and supported the development of a more **accurate** and **interpretable classification model**.


#### **II.4.1 Distribution of Target Variable**

````python
# Distribution of Target Variable - Churn

churn_counts = df['Churn'].value_counts()
churn_labels = churn_counts.index.map({0: 'Not Churned', 1: 'Churned'})

plt.figure(figsize=(6, 4))
sns.barplot(x=churn_labels, y=churn_counts.values, palette='crest')

# Annotate with percentage
for i, count in enumerate(churn_counts.values):
    percentage = count / churn_counts.sum() * 100
    plt.text(i, count + 50, f'{percentage:.1f}%', ha='center', fontsize=12)

plt.title('Distribution of Churn')
plt.ylabel('Number of Customers')
plt.xlabel('Churn Status')
plt.tight_layout()
plt.show()
````

#### <u>**Insights into Distribution of Target Variable**</u>

1. **Imbalanced Class Distribution**  
    - The `Churn` variable exhibits **class imbalance**, with **82.9% labeled as "Not Churned"** and only **17.1% labeled as "Churned"** after data cleaning.  
    - This skew can lead models to **favor the majority class**, compromising their ability to detect churners, which is the **primary business interest**.

2. **Implications for Model Performance**  
    - Imbalanced data can **distort performance metrics**: a model that always predicts **“Not Churned”** would still achieve **82.9% accuracy**, but have **zero recall** for churned customers.  
    - To address this, techniques such as **oversampling the minority class**, **undersampling the majority class**, or applying **class-weighted loss functions** are recommended. These approaches help prioritize **recall** and **F2-score**, which better align with **churn detection goals**.

3. **Business Insight**  
    - A **17.1% churn rate** suggests a **significant revenue leakage risk**. Even though the churn class is **numerically smaller**, its impact on **customer lifetime value**, **acquisition costs**, and **brand equity** can be substantial.  
    - Modeling efforts should focus on **accurately identifying this segment** to enable **proactive retention strategies** and improve **overall customer profitability**.

#### **II.4.2 Distribution of Numerical Features**

````python
# Descriptive Statistics of Numerical Features (After Data Cleaning)
df.describe().T
````

````python
# Histogram of Numerical Features Distribution

# Define the numerical features
num_features_histogram = [
    'Tenure', 'WarehouseToHome', 'HourSpendOnApp', 'NumberOfDeviceRegistered', 'NumberOfAddress',
    'OrderAmountHikeFromlastYear', 'CouponUsed', 'OrderCount', 'DaySinceLastOrder', 'CashbackAmount'
]

# Calculate rows and columns for subplots
n_features = len(num_features_histogram)
n_cols = 3
n_rows = (n_features + n_cols - 1) // n_cols  # ceiling division

# Set up the plot layout
fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 4 * n_rows))
axes = axes.flatten()

# Plot each numerical feature
for idx, column in enumerate(num_features_histogram):
    sns.histplot(df[column], kde=True, ax=axes[idx], color='darkcyan')
    axes[idx].set_title(f'Distribution of {column}')
    axes[idx].set_xlabel(column)
    axes[idx].set_ylabel('Frequency')

# Remove any unused subplots
for j in range(idx + 1, len(axes)):
    fig.delaxes(axes[j])

# Improve spacing
plt.tight_layout()
plt.show()
````

````python
# Boxplot of Numerical Features Distribution

# Define the numerical features
num_features_boxplot = [
    'Tenure', 'WarehouseToHome', 'HourSpendOnApp', 'NumberOfDeviceRegistered', 'NumberOfAddress',
    'OrderAmountHikeFromlastYear', 'CouponUsed', 'OrderCount', 'DaySinceLastOrder', 'CashbackAmount'
]

# Set up the figure size
plt.figure(figsize=(20, 14))

# Loop through the columns and plot boxplots
for i, column in enumerate(num_features_boxplot, 1):
    plt.subplot(4, 3, i)  # 4 rows, 3 columns = up to 12 plots
    sns.boxplot(x=df[column], showfliers=True, color='lightseagreen')
    plt.title(f'Boxplot of {column}')
    plt.xlabel(column)
    plt.ylabel('Value')

plt.tight_layout()
plt.show()
````

#### <u>**Insights into Distribution of Numerical Features**</u>  
<br>

1. **Skewed Distributions with Valid Outliers**  
    - **Features**: `Tenure`, `WarehouseToHome`, `CouponUsed`, `OrderCount`, `CashbackAmount`  
    - **Observation**:  
      - These features exhibit **right‐skewed distributions** (mean > median), indicating a long tail of high values:  
        - `Tenure`: mean 9.74 vs median 8.0  
        - `WarehouseToHome`: mean 15.28 vs median 13.0  
        - `CouponUsed`: mean 1.70 vs median 1.0  
        - `OrderCount`: mean 3.09 vs median 2.0  
        - `CashbackAmount`: mean 174.34 vs median 161.0  
      - Outlier counts range from **4** (`Tenure`) to **673** (`OrderCount`), reflecting genuine behavioral extremes.  
    - **Recommended Actions**:  
      - **Robust scaling** (e.g. `RobustScaler`) to reduce skew impact on distance‐based methods.  
      - **Retain outliers** during model training to capture high‐value and high‐risk segments.  
<br>

2. **Clustered or Bimodal Patterns**  
    - **Features**: `HourSpendOnApp`, `NumberOfDeviceRegistered`  
    - **Observation**:  
      - `HourSpendOnApp` clusters around values 2.0 and 3.0 hours (25th = 2.0, 75th = 3.0), with extremes at 0.0 and 5.0.  
      - `NumberOfDeviceRegistered` peaks at 3–4 devices (25th = 3.0, 75th = 4.0) with upper outliers up to 6.0.  
    - **Recommended Actions**:  
      - **Categorical binning** into “Low”, “Medium”, “High” usage or device counts to capture segments.  
      - In tree‐based models, use raw values since splits handle non‐linear effects and outliers gracefully.  
<br>

3. **Positively Skewed Counts Reflecting Customer Reach and Recency**  
    - **Features**: `NumberOfAddress`, `DaySinceLastOrder`, `OrderAmountHikeFromlastYear`  
    - **Observation**:  
      - Distribution statistics:  
        - `NumberOfAddress`: mean 4.14 vs median 3.0, max 22.0  
        - `DaySinceLastOrder`: mean 4.39 vs median 3.0, max 46.0  
        - `OrderAmountHikeFromlastYear`: mean 15.67 vs median 15.0, max 26.0  
      - Skew indicates a subset of users with many addresses, long order lags, or substantial spending increases.  
    - **Recommended Actions**:  
      - **Binning** into meaningful ranges.  
      - **Robust scaling** for algorithms sensitive to scale.

#### **II.4.3 Distribution of Categorical Features**

````python
# Descriptive Statistics of Categorical Features (After Data Cleaning)
df.describe(include = 'object').T
````

````python
# Barplot of Categorical Features Distribution

# Define categorical features to plot
cat_features_barplot = [
    'PreferredLoginDevice',
    'PreferredPaymentMode',
    'Gender',
    'PreferedOrderCat',
    'MaritalStatus'
]

# Set up subplots: one row per feature
fig, axes = plt.subplots(len(cat_features_barplot), 1, figsize=(10, 18))
axes = axes.flatten()

# Plot each categorical feature
for i, feature in enumerate(cat_features_barplot):
    counts = df[feature].value_counts()
    total = counts.sum()

    # Create horizontal barplot
    sns.barplot(x=counts.values, y=counts.index, ax=axes[i], palette='crest')
    axes[i].set_title(f'Distribution of {feature}', fontsize=14)
    axes[i].set_xlabel('Count', fontsize=12)
    axes[i].set_ylabel('')

    # Annotate each bar with count and percentage
    for j, (count, category) in enumerate(zip(counts.values, counts.index)):
        percentage = (count / total) * 100
        label = f'{count}\n({percentage:.1f}%)'
        axes[i].text(count + max(counts.values) * 0.01, j, label, va='center', fontsize=10)

    # Adjust tick appearance
    axes[i].tick_params(axis='y', labelrotation=0)

# Adjust layout
plt.tight_layout(pad=3)
plt.show()
````

### **II.5  Data Preprocessing**

**Data preprocessing** is a crucial stage in the machine learning pipeline. After cleaning and exploring the data, this phase transforms the dataset into a format ready for modeling, ensuring consistent and meaningful inputs that enhance predictive performance.  
<br>

#### **Preprocessing Stage Overview**

| **Step**               | **Purpose**                                                                 | **Key Actions**                                                                    |
|------------------------|-----------------------------------------------------------------------------|------------------------------------------------------------------------------------|
| **Data Transformation** | Standardize feature scales and convert categorical data to numeric formats  | - Robust scaling to reduce outlier impact  <br> - One‑Hot and Ordinal Encoding |
| **Feature Engineering** | Create structured features capturing engagement, emotional signals, and lifecycle stages | - Time‑based engagement metrics  <br> - Interaction‑based churn signals  <br> - Lifecycle stage segmentation |


<br>

#### **Why It Matters**

Proper **data transformation** ensures that all features are consistently formatted and scaled, which is essential for machine learning algorithms to perform well. For example, scaling numeric features prevents features with larger ranges from dominating the model training process. Encoding categorical variables into numeric formats allows models to interpret categories correctly, especially when there is a meaningful order.

**Feature engineering** enhances the dataset by creating new variables that better capture underlying behaviors and relationships. Interaction terms help models understand combined influences between variables. Binning and ordinal encoding reduce noise and allow the model to leverage natural rank order in categorical data.

Together, these preprocessing steps improve the model’s ability to generalize from the training data, reduce overfitting, and ultimately lead to better predictive accuracy.

<br>

#### **II.5.1 Data Transformation**

Data transformation is a critical stage within the broader data preprocessing phase where numerical and categorical features are modified to be suitable for machine learning models. The key goals are to standardize feature scales, convert categorical data into numerical formats, and preserve meaningful relationships within the data.  
<br>
1. **Standardization and Scaling**

    Scaling aligns feature ranges to prevent large-scale features from dominating the model training. This step is essential for algorithms sensitive to feature magnitudes, such as K-Nearest Neighbors and logistic regression.  
<br>

2. **Selecting the Proper Scaling Method**

| Method          | Description                          | Pros                        | Cons                       |
|-----------------|------------------------------------|-----------------------------|----------------------------|
| **StandardScaler** | Centers data to mean=0 and unit variance | Works well in most cases      | Sensitive to outliers       |
| **MinMaxScaler**   | Scales features to a [0,1] range    | Preserves original distribution | Also sensitive to outliers  |
| **RobustScaler**   | Uses median and interquartile range (IQR) | Resistant to outliers         | Less efficient on small datasets |   

<br>

  In this dataset, numerical features such as `Tenure`, `WarehouseToHome`, `HourSpendOnApp`, `OrderAmountHikeFromlastYear`, `NumberOfDeviceRegistered`, `NumberOfAddress`, `CouponUsed`, `OrderCount`, `DaySinceLastOrder`, and `CashbackAmount` show varying distributions and potential outliers. Therefore, the **RobustScaler** is applied to reduce the influence of extreme values and ensure fair representation across all scaled features.  
<br>

3. **Categorical Feature Encoding**

    - **One-Hot Encoding** is applied to nominal categorical features that lack intrinsic order, such as `PreferredLoginDevice`, `PreferredPaymentMode`, `Gender`, `MaritalStatus`, and `PreferedOrderCat`. This technique creates binary indicator columns for each category.

    - **Ordinal Encoding** is used for ordinal categorical features with natural rank order, such as `CityTier` and `SatisfactionScore`. Custom category mappings are defined: `CityTier` is ordered from best (1) to worst (3), and `SatisfactionScore` from bad (1) to great (5). This preserves meaningful rank relationships important for model learning.  
<br>

4. **Implementation with ColumnTransformer**  
    
    These transformations are implemented using a **ColumnTransformer**, which applies specific preprocessing steps, such as **scaling**, **one-hot encoding**, and **ordinal encoding**, to selected feature groups.

    Two columns, namely **`Complain`** and **`CustomerID`**, are set to **`passthrough`**. This means they are excluded from any transformation and retained in their original form:

    - **`Complain`** is already a clean binary numeric flag (`0` or `1`), so no transformation is needed.
    - **`CustomerID`** is a unique identifier and not a predictive feature. It is only passed through temporarily for tracking or reference, but will be **dropped before modeling** during the feature-target definition phase.

    This design ensures that all necessary transformations are applied appropriately, while preserving the structure and logic of features that require no change.


````python
# DATA TRANSFORMATION

# Features to scale (numerical)
scaling_cols = [
    'Tenure',
    'WarehouseToHome',
    'HourSpendOnApp',
    'OrderAmountHikeFromlastYear',
    'NumberOfDeviceRegistered',
    'NumberOfAddress',
    'CouponUsed',
    'OrderCount',
    'DaySinceLastOrder',
    'CashbackAmount'
]

# Nominal categorical features for one-hot encoding
onehot_cols = [
    'PreferredLoginDevice',
    'PreferredPaymentMode',
    'Gender',
    'MaritalStatus',
    'PreferedOrderCat'
]

# Ordinal categorical features
ordinal_cols = ['CityTier', 'SatisfactionScore']
ordinal_maps = [
    [1, 2, 3],              # CityTier: 1 (best) → 3 (worst)
    [1, 2, 3, 4, 5]         # SatisfactionScore: 1 (bad) → 5 (great)
]

# Final preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ('scaler', RobustScaler(), scaling_cols),
        ('onehot', OneHotEncoder(), onehot_cols),
        ('ordinal', OrdinalEncoder(categories=ordinal_maps), ordinal_cols),
    ],
    remainder='passthrough'     # CustomerID, Complain
)

preprocessor
````

#### **II.5.2 Feature Engineering**

**Feature engineering** enhances machine learning performance by **transforming raw variables into structured, informative features that better capture user behavior and risk patterns**. In churn prediction, where small shifts in engagement can portend lost revenue, thoughtfully engineered features can significantly amplify model accuracy and actionable insights. To address these challenges, several key feature engineering were applied to capture customer engagement, emotional signals, and lifecycle stages:


1. **Recency-Based Engagement Metrics**  

    - **`RecencyRatio`**  
        Transforms absolute recency into a lifecycle-relative measure. By dividing days since last order by total tenure, it adjusts for how long a customer has been with the company, recognizing that a 30-day gap means very different things for someone who joined 3 months ago versus a customer who has been with you for 36 months. This normalized signal highlights true disengagement, enabling the model to weigh recent inactivity appropriately across different customer groups.

    - **`IsActiveUser`**  
        A binary flag for orders in the past 30 days. This 30-day window aligns with common industry standards like Monthly Active Users (MAU), capturing meaningful recent engagement without being overly sensitive to short-term fluctuations. It effectively distinguishes customers who are currently engaged from those showing early signs of disengagement, enabling timely, targeted retention actions and improving the model’s ability to detect short-term churn risk.


2. **Churn Risk Signal**  

    - **`UnhappyCustomer`**  
        Combines behavioral and sentiment indicators, complaints plus low satisfaction, to encode emotional churn triggers. Single signals may mask true risk: a flagged dissatisfied customer who continues transacting still carries hidden defection probability. This interaction uncovers such latent risk, prioritizing outreach to customers most likely to defect despite stable purchase patterns.

3. **Loyalty and Tenure Binning**  

- **`TenureGroup`**  
  Segments continuous tenure into four meaningful stages based on evolving customer loyalty and churn risk:  
  - **New (0–3 months):** Onboarding phase with highest churn risk as customers evaluate the product and form initial impressions. Early experiences heavily influence retention.  
  - **Early (3–9 months):** Usage habits develop; customers become more engaged but remain sensitive to service quality and incentives. Positive interactions here can strengthen loyalty.  
  - **MidTerm (9–15 months):** Customers exhibit moderate loyalty with established patterns. Churn often stems from unmet expectations or emerging alternatives.  
  - **LongTerm (over 15 months):** Most loyal segment showing strong resistance to churn. If churn occurs, it usually reflects deeper dissatisfaction or major changes.  

Binning tenure simplifies modeling by grouping customers into natural lifecycle phases, reducing noise from continuous data and improving interpretability. The upper cutoff aligns with the tenure distribution where most customers fall below this threshold, distinguishing a smaller group of long-term, highly loyal customers whose behavior tends to differ significantly from newer or mid-tenure segments. This approach balances capturing meaningful lifecycle stages while avoiding sparsity in very long-tenured groups, enabling more effective, targeted retention strategies.  

<br>

#### **Summary Table of Engineered Features**  

| **Feature**         | **Reasons**                                                                                     | **Business Impact**                                           | **Definition**                                                    | **Source**                                                             |
|---------------------|-------------------------------------------------------------------------------------------------|----------------------------------------------------------------|-------------------------------------------------------------------|------------------------------------------------------------------------|
| **`RecencyRatio`**    | Normalizes inactivity by tenure to detect true disengagement across customer maturity levels    | Prioritizes at‑risk cohorts for retention campaigns            | **Ratio of days since last order to total days as a customer; measures recent inactivity relative to customer lifetime**                          | Fader, P.S., & Hardie, B.G.S. (2013). *RFM and CLV: Using customer lifetime value for marketing strategies*. Journal of Marketing Analytics, 1(3-4), 105-119. |
| **`IsActiveUser`**    | Provides a clear pivot between active and dormant users for tactical, time‑bound interventions  | Enables prompt re‑engagement actions (e.g., email reminders)   | **Binary flag indicating if the customer placed an order within the last 30 days (1 = yes, 0 = no)**                             | Kumar, V., & Reinartz, W. (2016). *Customer Relationship Management: Concept, Strategy, and Tools* (3rd ed.). Springer.                |
| **`UnhappyCustomer`** | Highlights customers with both transactional complaints and low satisfaction, isolating emotional risk | Drives proactive outreach to customers most likely to churn   | **Binary flag where 1 indicates the customer submitted a complaint and gave a satisfaction score of 2 or lower, 0 otherwise**               | Mittal, V., & Ross Jr, W.T. (1998). *The Impact of Positive and Negative Affect on Dealing with Customer Complaints*. Journal of Consumer Satisfaction, Dissatisfaction and Complaining Behavior, 11, 56-63. |
| **`TenureGroup`**     | Encodes lifecycle stages, reflecting evolving loyalty behaviors and churn triggers per segment  | Guides tailored retention strategies by tenure segment         | **Categorical variable grouping customers by length of relationship: New (0–3 months), Early (3–9 months), MidTerm (9–15 months), LongTerm (>15 months)** | Bolton, R.N., & Lemon, K.N. (1999). *A Dynamic Model of Customers’ Usage of Services: Usage as an Antecedent and Consequence of Satisfaction*. Journal of Marketing Research, 36(2), 171-186. |


<br>

````python
# Check the Descriptive Statistics of 'Tenure'
df['Tenure'].describe()
````

````python
# FEATURE ENGINEERING

# 1. Recency-Based Engagement Metrics
df['RecencyRatio'] = df['DaySinceLastOrder'] / (df['Tenure'] + 1)  # Avoid divide by zero
df['IsActiveUser'] = (df['DaySinceLastOrder'] < 30).astype(int)    # Flag if user recently active

# 2. Churn Risk Signals
df['UnhappyCustomer'] = ((df['Complain'] == 1) & (df['SatisfactionScore'] <= 2)).astype(int)  # Complaint + Low Satisfaction

# 3. Loyalty & Tenure Binning
df['TenureGroup'] = pd.cut(df['Tenure'],
                           bins=[-1, 3, 9, 15, np.inf],
                           labels=['New', 'Early', 'MidTerm', 'LongTerm'])  # Discretized loyalty levels

# Check the result
df[['RecencyRatio', 'IsActiveUser', 'UnhappyCustomer', 'TenureGroup']].sample(10)
````

````python
# Display unique counts for new features ('RecencyRatio', 'IsActiveUser', 'UnhappyCustomer', and 'TenureGroup')

print("Unique values for RecencyRatio:")
print(df['RecencyRatio'].nunique(), "\n")

print("Value counts for IsActiveUser:")
print(df['IsActiveUser'].value_counts().to_frame().sort_index(), "\n")

print("Value counts for UnhappyCustomer:")
print(df['UnhappyCustomer'].value_counts().to_frame().sort_index(), "\n")

print("Value counts for TenureGroup:")
print(df['TenureGroup'].value_counts().to_frame().sort_index())
````

````python
# UPDATE PREPROCESSOR

# Features to scale (numerical)
scaling_cols = [
    'Tenure',
    'WarehouseToHome',
    'HourSpendOnApp',
    'OrderAmountHikeFromlastYear',
    'NumberOfDeviceRegistered',
    'NumberOfAddress',
    'CouponUsed',
    'OrderCount',
    'DaySinceLastOrder',
    'CashbackAmount',
    'RecencyRatio'         # newly created recency ratio
]

# Nominal categorical features for one-hot encoding
onehot_cols = [
    'PreferredLoginDevice',
    'PreferredPaymentMode',
    'Gender',
    'MaritalStatus',
    'PreferedOrderCat'
]

# Ordinal categorical features
ordinal_cols = [
    'CityTier',
    'SatisfactionScore',
    'TenureGroup',
]

# Define ordinal category orders for new binned features
ordinal_maps = [
    [1, 2, 3],                                # CityTier: 1 (best) → 3 (worst)
    [1, 2, 3, 4, 5],                         # SatisfactionScore: 1 (bad) → 5 (great)
    ['New', 'Early', 'MidTerm', 'LongTerm'], # TenureGroup: ordered time bins
]

# Final preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ('scaler', RobustScaler(), scaling_cols),
        ('onehot', OneHotEncoder(), onehot_cols),
        ('ordinal', OrdinalEncoder(categories=ordinal_maps), ordinal_cols),
    ],
    remainder='passthrough'       # CustomerID, Complain, and newly created IsActiveUser and UnhappyCustomer
)

preprocessor
````

#### **Data Splitting**

````python
# Define Features and Target
X = df.drop(columns=['CustomerID', 'Churn'])  # Features
y = df['Churn']                               # Target

# Split Data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=10
)

X_train
````

#### **Applying Preprocessing Pipeline and Reassigning Feature Names**

````python
# a. APPLYING PREPROCESSING PIPELINE

# Define the pipeline for feature engineering
preprocessing_pipeline = ImbPipeline(steps=[
    ('preprocessor', preprocessor),
])

# Apply the preprocessing pipeline to X_train and X_test
X_train_transformed = pd.DataFrame(preprocessing_pipeline.fit_transform(X_train))
X_test_transformed = pd.DataFrame(preprocessing_pipeline.transform(X_test))

# b. REASSIGNING FEATURE NAMES

# Get scaled feature names
scaled_features = preprocessing_pipeline.named_steps['preprocessor'].transformers_[0][2]

# Get one-hot encoded feature names
onehot_encoded_features = preprocessing_pipeline.named_steps['preprocessor'].transformers_[1][1] \
                            .get_feature_names_out(onehot_cols)

# Get ordinal encoded feature names
ordinal_encoded_features = preprocessing_pipeline.named_steps['preprocessor'].transformers_[2][1] \
                            .get_feature_names_out(ordinal_cols)

# Get remaining features that are passed through as-is
remaining_features = X_train.drop(columns=scaling_cols + onehot_cols + ordinal_cols, errors='ignore').columns

# Combine all feature names in order
all_features = list(scaled_features) + \
               list(onehot_encoded_features) + \
               list(ordinal_encoded_features) + \
               list(remaining_features)

# Assign column names to transformed training and test features
X_train_transformed.columns = all_features
X_test_transformed.columns = all_features

# Show transformed training data sample
X_train_transformed.sample(10)
````

### **III.1 Modeling Strategy**

#### **a. Objective**

This chapter aims to build and evaluate multiple **machine learning classification models** to predict **customer churn** with strong technical performance and practical business value. The main goal is to develop a **predictive churn classification model** that delivers reliable performance and helps the company proactively identify customers at risk of churn before they disengage.

The modeling strategy is based on the **insights, features, and transformations prepared in Chapter II**, and is aligned with the **business problem defined in Chapter I**:

> *“How can we identify customers at risk of churn so the company can take proactive and targeted actions?”*  
<br>

#### **b. Target Variable**
* **`Churn`**: A binary label (1 = churned, 0 = retained)
* The problem is a **binary classification task** with **moderate class imbalance**, where the minority class (churned users) is the key interest.  
<br>

#### **c. Features for Modeling**  
Final features selected from Chapter II cover 5 behavioral domains:  
1. **Demographics**: Gender, MaritalStatus, CityTier
2. **Engagement**: Tenure, HourSpendOnApp, WarehouseToHome, NumberOfDeviceRegistered, NumberOfAddress
3. **Transaction Behavior**: OrderCount, DaySinceLastOrder, OrderAmountHikeFromlastYear, CouponUsed, CashbackAmount
4. **Platform & Preference**: PreferredLoginDevice, PreferredPaymentMode, PreferedOrderCat
5. **Satisfaction & Complaints**: SatisfactionScore, Complain

All categorical features have been **label encoded**, and numerical features are **scaled**.
<br>

#### **d. Selecting 10 Different Models**
A diverse set of 10 classification models is employed to ensure a comprehensive evaluation and to identify the most generalizable, high-performing solution. Each model contributes distinct assumptions and strengths:

| No. | Model                      | Why It’s Used                                                                                     |
| --- | -------------------------- | ------------------------------------------------------------------------------------------------- |
| 1   | **Logistic Regression**    | Simple, interpretable baseline model that produces well-calibrated probabilities                 |
| 2   | **K-Nearest Neighbors**    | Instance-based learner that adapts to data distribution; useful for small and clean datasets      |
| 3   | **Decision Tree**          | Non-linear, rule-based model that handles mixed data types without encoding                       |
| 4   | **Bagging Classifier**     | Variance-reduction ensemble that stabilizes high-variance base learners like decision trees       |
| 5   | **Random Forest**          | Ensemble of randomized trees offering robustness and built-in feature importance                  |
| 6   | **Gradient Boosting**      | High-performing sequential ensemble that captures subtle feature interactions                     |
| 7   | **XGBoost**                | Optimized gradient boosting with internal handling of missing data and efficient parallelism      |
| 8   | **AdaBoost**               | Boosts weak learners by focusing on previous errors; useful baseline boosting approach            |
| 9   | **CatBoost**               | Boosting model with native support for categorical data and strong generalization capabilities    |
| 10  | **LightGBM**               | Fast, memory-efficient boosting method ideal for real-time scoring and large-scale applications   |

These models were chosen based on:
- A diverse range of learning approaches, including **linear models**, **tree-based algorithms**, **bagging ensembles**, and various **boosting techniques**
- Their suitability for real-world churn prediction datasets, which are typically **tabular**, include many **categorical features**, and exhibit **non-normal distributions**  
<br>

#### **e. Model Evaluation Metrics**

The primary business goal is to **minimize customer churn** by accurately identifying churners. Therefore, **accuracy is not prioritized** due to class imbalance and the higher business cost of **false negatives** (missed churners) compared to **false positives** (retaining non-churners unnecessarily).

After careful consideration of the goals outlined in Chapter I and the underlying business logic, the **prioritized evaluation metrics** are finalized as follows:

| **Metric**     | **Why It Matters**                                                                                          |
|----------------|-------------------------------------------------------------------------------------------------------------|
| **F2-Score**   | Top priority. It places greater emphasis on recall than precision, making it ideal when false negatives are significantly more costly. It balances performance while ensuring recall is not improved at the expense of excessively high false positives. |
| **Recall**     | Critical metric. It directly measures how many actual churners are successfully identified. However, it does not account for precision, which may lead to excessive false positives if optimized in isolation. |
| **PR-AUC**     | Focuses on the model’s ability to identify positive cases (churners), especially effective for imbalanced datasets. |
| **F1-Score**   | Balances precision and recall equally. Useful when both false positives and false negatives are of concern.  |
| **Precision**  | Important for evaluating how many predicted churners are actually correct. However, in this context, **recall takes priority** due to the higher cost of missed churners. |
| **ROC-AUC**    | Measures the model’s overall discriminative ability across all thresholds. Less effective for imbalanced datasets. |
| **Accuracy**   | Not prioritized. Can be misleading due to class imbalance and does not reflect cost asymmetry between error types. |  

<br>

**Final Priority Order:**  
**F2-Score > Recall > PR-AUC > F1-Score > Precision > ROC-AUC > Accuracy**

While **Recall** remains a critical component in minimizing missed churners, **F2-Score** is prioritized as it offers a more comprehensive measure that aligns with the business objective. By weighting recall more heavily but not ignoring precision, **F2-Score** ensures the model effectively captures churners without overwhelming the retention system with false positives.  
<br>

#### **f. Train-Test Split & SMOTE-ENN Handling**
* **80/20 stratified split** to maintain class distribution  
* **SMOTE-ENN (Synthetic Minority Oversampling Technique combined with Edited Nearest Neighbors)** will be applied **only on training set**  
* Helps reduce bias in training while preserving real distribution in test set  
* SMOTE creates synthetic samples for the minority class, while ENN removes ambiguous samples from both classes to improve class separability  
<br>

#### **g. Cross-Validation and Robustness**
To ensure generalizability:
* All models will be evaluated using **5-fold cross-validation** on the training set
* Test performance will be reported separately to check for overfitting or variance  
<br>

### **h. Hyperparameter Tuning**
Each model undergoes **hyperparameter tuning** using:  
* **RandomizedSearchCV**, a randomized search over a grid of hyperparameters  
* Evaluation metrics during tuning include: **F2-Score, Recall, PR-AUC, F1-Score, Precision, ROC-AUC, Accuracy**  
* Key hyperparameters tuned include:  
  * **Tree depth**, **learning rate**, and **number of estimators** for boosting models  
  * **Regularization terms** for logistic regression  
  * **k-values** and **distance metrics** for KNN  
  * Other model-specific parameters

Tuned models are then re-evaluated on both **cross-validation folds** and the **holdout test set**.  
<br>

#### **i. Model Selection Criteria**
The best-performing model is selected based on the following criteria:
* Highest **F2-Score** on the **test set** after handling class imbalance and hyperparameter tuning  
* Consistent and stable performance across cross-validation folds (**low variance**)  
* **Interpretability and explainability**, such as through SHAP values for tree-based models  
* **Practicality for deployment** and ease of communication with non-technical business stakeholders   
<br>

#### **j. Business Impact of Modeling**
The predictive churn classification model has strong business relevance and can be used to support key strategic objectives based on Chapter I:
* **Proactively identify customers at risk of churn** before they fully disengage from the platform  
* **Enable better decision-making** across departments such as marketing, customer experience, product, and executive teams  
* **Reduce churn rate** by triggering timely retention actions based on predicted churn risk  
* **Optimize marketing budget allocation** by focusing retention campaigns on high-risk customer segments  
* **Integrate the model into CRM systems or business dashboards** to support real-time churn scoring and prioritization

By turning churn predictions into concrete business actions, the model helps the company **improve customer retention** and make **data-driven strategic decisions**.

### **III.2  Benchmark Model**

**Benchmarking** is the process of systematically comparing multiple classification algorithms under **identical preprocessing and evaluation conditions** to establish **baseline performance**. In the context of churn prediction, this helps assess each model’s **predictive ability** before any hyperparameter tuning. It guides the selection of the most promising models for further optimization.  

Ten classification models were tested within a **uniform pipeline** that applied consistent **data preprocessing** steps beforehand, including **scaling**, **encoding**, and **feature engineering**. This ensures fair comparisons where performance differences reflect only the **models’ characteristics**, not preprocessing variations.  

Each model was trained on the **training set** and evaluated on the **unseen test set**, using a comprehensive set of **evaluation metrics**:  

- **F2-Score**: Weighted harmonic mean of precision and recall, with more emphasis on recall. Ideal when false negatives are more costly than false positives.  
- **Recall**: Correct positive predictions among actual positives.  
- **PR-AUC**: Area under the Precision-Recall curve; emphasizes performance on the positive (churn) class, especially important for imbalanced datasets.  
- **F1-Score**: Harmonic mean of precision and recall.  
- **Precision**: Correct positive predictions among predicted positives.  
- **ROC-AUC**: Measures classifier's ability to distinguish between classes across thresholds.  
- **Accuracy**: Overall percentage of correct predictions.  

The selected models range from **linear classifiers** such as **Logistic Regression**, to **non-parametric learners** like **K-Nearest Neighbors** and **Decision Trees**, and extend to powerful **ensemble-based models**. These include **Bagging Classifier**, **Random Forest**, and **boosting** techniques (**Gradient Boosting**, **XGBoost**, **AdaBoost**, **CatBoost**, and **LightGBM**).  

This diverse selection allows for a balanced evaluation across models that vary in **interpretability**, **computational efficiency**, and ability to capture **non-linear patterns** in the data. Each model brings unique strengths and is suited for different real-world scenarios, making this benchmarking phase critical for identifying the most promising algorithms for churn classification.  

<br>

#### **Model Summary Table**

| **Model**                     | **Approach**                                                 | **Strengths**                                                                                           | **Ideal Use Cases**                                       | **Limitations**                                               |
| ----------------------------- | ------------------------------------------------------------ | ----------------------------------------------------------------------------------------------------- | --------------------------------------------------------- | ------------------------------------------------------------- |
| **Logistic Regression**       | Linear classification with probability outputs               | Widely used in industry; interpretable coefficients; fast to train and predict; produces well-calibrated probabilities | Baseline models; scenarios requiring clear explanations   | Assumes linear boundaries; underperforms on complex patterns  |
| **K-Nearest Neighbors (KNN)** | Predicts based on majority class among nearest neighbors     | No explicit training step; adapts naturally to data distribution; effective in low-dimensional, noise-free settings | Small, well-separated datasets; rapid prototyping         | Prediction latency increases with dataset size; sensitive to irrelevant features |
| **Decision Tree**             | Recursive rule-based splitting                               | Easily visualized as decision rules; captures non‑linear interactions; handles mixed data types without encoding | When model transparency is essential; mixed feature types | Overfits without pruning; high variance on small datasets     |
| **Bagging Classifier**        | Ensemble of base classifiers trained on bootstrapped samples | Empirically reduces variance and overfitting; improves stability of unstable learners                  | Stabilizing high‑variance models like individual trees    | Increased computational cost; ensemble complexity reduces interpretability |
| **Random Forest**             | Ensemble of decision trees with feature randomness           | Industry standard for tabular data; robust to noise and outliers; provides feature‐importance metrics   | Complex, non‑linear tabular tasks; when interpretability of features matters | Can be slow on very large datasets; less granular interpretability than single trees |
| **Gradient Boosting**         | Sequential learners correcting predecessor errors            | Often achieves top accuracy on structured data; models subtle feature interactions; built‑in regularization | High‑accuracy requirements on medium‑size datasets        | Sensitive to hyperparameters; risk of overfitting if not tuned properly |
| **XGBoost**                   | Highly optimized gradient boosting                           | Leading performance in ML competitions; handles missing values internally; efficient multi‑threading    | Production systems demanding speed and accuracy           | Complex parameter tuning; heavier memory usage               |
| **AdaBoost**                  | Sequential model with emphasis on misclassified instances    | Boosts simple learners into stronger models; works well on moderate‑noise datasets; minimal parameter tuning | Quick boosting for baseline improvements                  | Highly sensitive to noisy labels and outliers                |
| **CatBoost**                  | Gradient boosting with native categorical support            | Excellent performance with categorical variables; minimal preprocessing required; stable default settings | Datasets with many categorical features                   | Slower training times relative to other boosting libraries   |
| **LightGBM**                  | Leaf‑wise gradient boosting with histogram optimizations     | Extremely fast training on large datasets; low memory footprint; supports categorical features natively | Real‑time scoring and large‑scale applications            | Can overfit small datasets; requires careful tuning of leaf parameters |


````python
# BENCHMARK MODEL

# Initialize results dictionary
benchmark_summary = {
    'Model': [],
    'F2-Score': [],
    'Recall': [],
    'PR-AUC': [],
    'F1-Score': [],
    'Precision': [],
    'ROC-AUC': [],
    'Accuracy': []
}

# Initialize Benchmark Models
benchmark_models = {
    'Logistic Regression': LogisticRegression(solver='lbfgs', penalty='l2', C=1.0, max_iter=1000, random_state=10),
    'KNN': KNeighborsClassifier(n_neighbors=5, weights='uniform'),
    'Decision Tree': DecisionTreeClassifier(criterion='entropy', max_depth=5, random_state=10),
    'Bagging': BaggingClassifier(n_estimators=100, random_state=10),
    'Random Forest': RandomForestClassifier(n_estimators=100, max_depth=5, random_state=10),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=10),
    'XGBoost': XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=5, verbosity=0, use_label_encoder=False, random_state=10),
    'AdaBoost': AdaBoostClassifier(n_estimators=100, learning_rate=0.1, random_state=10),
    'CatBoost': CatBoostClassifier(iterations=100, learning_rate=0.1, depth=6, verbose=0, random_state=10),
    'LightGBM': LGBMClassifier(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=10, verbose=-1)
}

for name, model in benchmark_models.items():
    print(name)

    benchmark_pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    benchmark_pipeline.fit(X_train, y_train)
    y_pred_benchmark = benchmark_pipeline.predict(X_test)
    y_proba_benchmark = benchmark_pipeline.predict_proba(X_test)[:, 1] if hasattr(model, 'predict_proba') else None

    print(classification_report(y_test, y_pred_benchmark, digits=2))
    print("-" * 80)

    benchmark_summary['Model'].append(name)
    benchmark_summary['F2-Score'].append(fbeta_score(y_test, y_pred_benchmark, beta=2, zero_division=0))
    benchmark_summary['Recall'].append(recall_score(y_test, y_pred_benchmark, zero_division=0))
    benchmark_summary['PR-AUC'].append(average_precision_score(y_test, y_proba_benchmark) if y_proba_benchmark is not None else np.nan)
    benchmark_summary['F1-Score'].append(f1_score(y_test, y_pred_benchmark, zero_division=0))
    benchmark_summary['Precision'].append(precision_score(y_test, y_pred_benchmark, zero_division=0))
    benchmark_summary['ROC-AUC'].append(roc_auc_score(y_test, y_proba_benchmark) if y_proba_benchmark is not None else np.nan)
    benchmark_summary['Accuracy'].append(accuracy_score(y_test, y_pred_benchmark))

# Create DataFrame, sort by priority metrics
benchmark_df = pd.DataFrame(benchmark_summary).set_index('Model')
benchmark_df = benchmark_df.sort_values(
    by=['F2-Score', 'Recall', 'PR-AUC', 'F1-Score', 'Precision', 'ROC-AUC', 'Accuracy'],
    ascending=False
).round(4)

from IPython.display import display
display(benchmark_df)
````

#### **Precision Outperforms Recall**

Although **F2-Score** is the primary metric for evaluating model performance in this churn prediction task, due to its stronger emphasis on recall, the benchmarking results reveal a consistent pattern: **precision scores are generally higher than recall across most models**.

This outcome stems largely from **class imbalance**: the cleaned dataset shows that **churners represent only 17.1%** of all customers. In such imbalanced settings, models tend to favor the majority class ("no churn"), which can **inflate accuracy** but result in poor detection of actual churners.

Standard classifiers, especially without tuning or rebalancing, learn to be conservative with positive (churn) predictions. As a result, they often **predict churn only when they are highly confident**, which leads to:

- **High precision**: Most predicted churners truly churn.
- **Low recall**: Many actual churners go undetected.

This trade-off negatively affects **F2-Score**, which prioritizes correctly identifying churners (recall) while still considering precision. Since **false negatives (missed churners)** carry a higher business cost than false positives, improving recall is essential to raising the overall F2-Score.

**Why precision is often higher than recall:**

- **Imbalanced data skews learning**: With only 17.1% churners, models are biased toward predicting "no churn." This causes **fewer false positives (higher precision)** but **missed detections (lower recall)**.
- **Default thresholds (usually 0.5)** are not designed for imbalance. They minimize total error, not churn detection.
- **Overfitting to clear churn signals**: Some models, like AdaBoost, may detect only very obvious churners, giving high precision but poor generalization and low recall.
- **Overlapping patterns**: Many customers show churn-like behavior without actually churning. Models tend to **avoid false alarms** by classifying these cases as "no churn," missing potential churners.

**Key implications from benchmarking:**
- **Bagging, CatBoost, and Random Forest** show **precision > 0.89**, suggesting reliable churn predictions, but recall still lags.
- **AdaBoost** is a striking case: **precision = 0.8947**, but **recall = 0.1083**. Nearly all predicted churners do churn, but the model misses most of them.
- Even top performers like **Gradient Boosting** and **LightGBM** display a recall gap that limits their F2-Score and real-world impact.

**Solutions to improve F2-Score:**
- Apply **threshold tuning** to adjust the decision boundary and capture more churners.
- Use **class rebalancing techniques** (e.g., SMOTE, SMOTEENN) to expose models to more minority-class examples during training.
- Continue prioritizing **F2-Score**, which weights recall more heavily than precision, better aligning with the risk-sensitive nature of churn prediction.

This analysis highlights the importance of looking beyond traditional metrics like accuracy or even precision. In churn prediction, the **true test of a model is how effectively it identifies those at risk of leaving**, and the **F2-Score offers the most relevant lens** for balancing that goal with reliability.

### **III.3 Model Optimization and Selection**

Model optimization and selection is a **critical phase** that ensures the chosen classifier is both **effective** and **reliable for deployment**. It begins with **handling class imbalance**, which is especially important in churn prediction where the **positive class (churners)** is typically underrepresented. Addressing this imbalance using techniques like **resampling** or **weighted loss functions** helps improve **sensitivity toward minority classes**.

Following that, **hyperparameter tuning** is performed to fine-tune model behavior by adjusting key parameters that control **learning dynamics**. Techniques such as **Grid Search** or **Randomized Search** are used to systematically explore parameter combinations, aiming to **reduce overfitting** and enhance **generalization**. After tuning, **best model selection** involves comparing candidate models using **cross-validated performance metrics** such as **F2-Score**. The final choice balances **predictive power** with practical considerations like **interpretability** and **business constraints**.

Once a final model is selected, the focus shifts to **model explanation**, which ensures **transparency** and builds trust by revealing **how the model makes its predictions**. This is especially important when the model is used to guide **customer retention strategies**. Lastly, **test set evaluation and metric analysis** provides an **unbiased estimate of real-world performance** by evaluating the model on **unseen data**. This confirms whether the model **generalizes well** and is **ready for practical use**.

<br>



#### **III.3.1 Handling Class Imbalance**

**Handling class imbalance** is a critical step in churn prediction because churners typically make up only a small portion of the customer base. This class imbalance can severely **impact the performance of machine learning models** by causing them to be **biased toward the majority class (non-churners)**. As a result, the model may achieve high overall accuracy but perform poorly at identifying churners, leading to **low recall**, which is particularly problematic in churn prediction where the cost of missing a true churner is high.

If class imbalance is left unaddressed, models tend to favour the majority class during training, learning to classify most customers as non-churners. This may produce misleading evaluation metrics and limit the model's ability to detect the minority class, which in this context is the **most business-critical group**. Accurate identification of churners is essential for proactive retention strategies and cost-effective marketing actions.

To overcome this issue, several strategies are used to balance the class distribution or adjust the model's behavior. These strategies fall into **two major categories**:

1. **Resampling**  
  **Resampling** refers to techniques that alter the training data to achieve a more balanced class distribution. This is done either by increasing the number of samples in the minority class (**oversampling**) or reducing the number of samples in the majority class (**undersampling**), or by combining both.

    - **Oversampling**  
      Increase the minority class representation by adding samples.  
      - **Definition**  
        Artificially enlarge the minority class using duplication or synthetic methods.  
      - **Why Use It**  
        Ensures the model sees enough minority samples to learn their patterns.  
      - **Trade-Offs**  
        - Simple duplication can lead to **overfitting**.  
        - Synthetic techniques like **SMOTE** can reduce overfitting by adding variation.

    - **Undersampling**  
      Decrease the majority class representation by removing samples.  
      - **Definition**  
        Randomly or selectively remove majority class examples.  
      - **Why Use It**  
        Reduces training time and forces the model to focus on learning from the minority class.  
      - **Trade-Offs**  
        - Risk of **underfitting** due to loss of valuable information.  
        - Techniques like **Tomek Links** remove only borderline or overlapping samples, preserving important patterns.

    - **Combination Techniques**  
      Use both **oversampling** and **undersampling** together to better refine the class distribution. These methods aim to **leverage the strengths** of both approaches while **mitigating their weaknesses**.  
      - Helps reduce noise, class overlap, and overfitting.  
      - Improves generalization and class separation.  
      - Examples:  
        - **SMOTE-Tomek**: Applies SMOTE followed by Tomek Links to remove ambiguous samples.  
        - **SMOTE-ENN**: Applies SMOTE followed by Edited Nearest Neighbors to clean noisy or mislabeled data.  
<br>

2. **Algorithmic Adjustments/ Penalized Models**  
    Rather than modifying the dataset, algorithmic approaches adjust the model to handle imbalance internally.
    - **Definition**  
      Modify the learning algorithm by applying **higher penalties** or **class weights** to errors on the minority class.  
    - **Why Use It**  
      Keeps the original data unchanged and integrates easily into many modeling frameworks.  
    - **Trade-Offs**  
      - Does not introduce new information to help the model understand minority class patterns.  
      - May be less effective when the class imbalance is **extreme** or when minority class patterns are complex.

<br>

#### **Overview of Common Methods**

| Method                   | Type                    | How It Works                                                                                   | Pros                                                                                                    | Cons                                                                                                         | Typical Use Cases                                                      |
|--------------------------|-------------------------|------------------------------------------------------------------------------------------------|---------------------------------------------------------------------------------------------------------|--------------------------------------------------------------------------------------------------------------|------------------------------------------------------------------------|
| **Random Oversampling**  | Oversampling            | Randomly duplicates existing minority‑class samples                                           | **Simple to implement**<br>**Immediately balances classes**                                               | **High risk of overfitting** due to exact duplicates                                                           | **Small to medium datasets**<br>**When simplicity and interpretability are priorities** |
| **SMOTE**                | Oversampling            | Generates synthetic minority‑class samples by interpolating between nearest neighbors         | **Reduces overfitting**<br>**Preserves feature‑space continuity**                                        | **Can create unrealistic samples** in noisy or multimodal data                                                 | **Moderate imbalance**<br>**Often paired with Tomek Links for boundary cleaning** |
| **ADASYN**               | Oversampling            | SMOTE variant that generates more synthetic samples in harder‑to‑learn regions                | **Focuses on decision boundary**<br>**Enhances minority sample variety**                                 | **May introduce noise** in overlapping regions                                                                   | **Moderate‑to‑severe imbalance**<br>**When boundary sensitivity is critical** |
| **GAN-based**            | Oversampling            | Uses Generative Adversarial Networks to generate realistic synthetic minority samples         | **Produces high‑fidelity samples**<br>**Captures complex distributions**                                  | **Complex implementation**<br>**Requires large data volume and careful tuning**                                | **Advanced applications**<br>**When realistic synthetic data is essential** |
| **Random Undersampling** | Undersampling           | Randomly removes majority‑class samples                                                       | **Speeds up training**<br>**Simple to implement**                                                          | **May discard informative examples**<br>**Risk of underfitting**                                                 | **Very large datasets**<br>**When computational resources are limited** |
| **NearMiss**             | Undersampling           | Selects majority samples closest to minority samples based on distance metrics                | **Preserves decision boundaries**<br>**More informative sampling**                                        | **Still loses some majority variation**<br>**Potential underfitting**                                           | **Focus on boundary regions**<br>**Better than random removal for clean separation** |
| **Cluster Centroids**    | Undersampling           | Clusters majority samples and replaces each cluster with its centroid                         | **Reduces data size representatively**<br>**Preserves overall distribution structure**                   | **Quality depends on clustering algorithm**<br>**Risk of oversimplification**                                  | **Highly redundant majority class**<br>**When compact, representative data is needed** |
| **Tomek Links**          | Undersampling           | Removes overlapping majority‑minority pairs (Tomek links) to clean noisy boundary samples     | **Cleans boundary noise**<br>**Improves class separation**                                               | **Minimal effect on class ratio**<br>**More of a cleaning step than a rebalancing method**                     | **Preprocessing step**<br>**Often used with SMOTE (SMOTE‑Tomek)** |
| **SMOTE-Tomek**          | Hybrid                  | Applies **SMOTE** to generate synthetic minority samples, then uses **Tomek Links** to remove ambiguous pairs  | **Reduces class overlap and noise**<br>**Improves class separation**                                      | **Moderate complexity**<br>**Risk of removing useful samples**                                                  | **Moderate imbalance**<br>**Boundary cleaning after synthetic sampling** |
| **SMOTE-ENN**            | Hybrid                  | Applies **SMOTE** to oversample minority, then uses **Edited Nearest Neighbors (ENN)** to remove noisy or mislabeled samples  | **Aggressive noise removal**<br>**Enhances generalization**<br>**Often outperforms SMOTE‑Tomek on noisy data** | **Computationally expensive**<br>**Risk of over‑cleaning minority samples**                                    | **Noisy or overlapping data**<br>**When robust generalization is critical** |
| **Class Weighting**      | Algorithmic Adjustment  | Increases penalty for misclassifying minority samples during training (e.g., `class_weight`)  | **No data alteration**<br>**Easy integration in most models**                                             | **Does not add new information**<br>**May underperform under extreme imbalance**                                 | **When data cannot be modified**<br>**Mild to moderate imbalance** |
| **Balanced Bagging**     | Hybrid                  | Applies random sampling (over or under) within each ensemble member before model fitting      | **Ensemble robustness**<br>**Reduces sampling variance**                                                   | **Increased complexity**<br>**Longer training time**                                                              | **High‑stakes applications**<br>**When both robustness and balanced learning are required** |

<br>

#### **Resampling Method Selection: SMOTE-ENN**

Among resampling techniques, **SMOTE-ENN** combines oversampling and cleaning to address class imbalance effectively. Unlike naive oversampling, which risks overfitting, and SMOTE-Tomek, which only removes simple boundary noise, SMOTE-ENN first generates synthetic minority samples and then applies Edited Nearest Neighbors (ENN) to remove ambiguous or mislabeled examples. This leads to a cleaner and more informative training set.

SMOTE-ENN offers several advantages over other methods ([Zhou et al., 2021](https://www.researchgate.net/publication/346282224_SMOTE-ENN-Based_Data_Sampling_and_Improved_Dynamic_Ensemble_Selection_for_Imbalanced_Medical_Data_Classification)):  
- **Balances the minority class** with synthetic samples, reducing underrepresentation.  
- **Removes noisy or overlapping examples**, improving class separability.  
- **Enhances generalization** by preserving local structure while eliminating misleading samples.


Class weighting adjusts the penalty for misclassification but does not alter the data. In contrast, SMOTE-ENN both enriches and cleans the training set, helping models learn minority class patterns even under severe imbalance.

In this project, **SMOTE-ENN is applied to the training set before model training**, providing a robust foundation for classifiers to detect churners effectively. This approach aims to optimize business-critical metrics like **F2-Score**, **recall**, and **PR-AUC**, which prioritize correctly identifying true churn cases over maximizing overall accuracy.

<br>

#### **How SMOTE-ENN Works**

SMOTE-ENN operates in two stages ([Zhou et al., 2021](https://www.researchgate.net/publication/346282224_SMOTE-ENN-Based_Data_Sampling_and_Improved_Dynamic_Ensemble_Selection_for_Imbalanced_Medical_Data_Classification))
:

1. **SMOTE (Synthetic Minority Oversampling Technique)**  
   - For each minority class instance `x`, find its **k nearest neighbors** in the minority class (`k=5` by default).  
   - For each neighbor `x'`, generate a synthetic sample:  
     ```  
     synthetic = x + λ * (x' − x)  
     ```  
     where `λ` is a random number in `[0, 1]`.

2. **ENN (Edited Nearest Neighbors)**  
   - For each example (both original and synthetic), find its **k nearest neighbors** in the combined dataset (`k=3` by default).  
   - **Remove** any example whose class label disagrees with the **majority label** of its neighbors.

By first **filling the minority feature space** with plausible synthetic points and then **cleaning noisy or mislabeled samples**, SMOTE-ENN creates a balanced and high-quality training set. Classifiers trained on this data achieve **wider decision regions** and **better generalization**, especially in imbalanced tasks like churn prediction.

<br>

````python
# BENCHMARK MODEL USING CROSS-VALIDATION (WITHOUT SMOTE-ENN)

# Define 5-fold stratified CV and scoring metrics
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=10)
scoring = {
    'recall': make_scorer(recall_score, zero_division=0),
    'precision': make_scorer(precision_score, zero_division=0),
    'f1': make_scorer(f1_score, zero_division=0),
    'f2': make_scorer(fbeta_score, beta=2, zero_division=0),
    'roc_auc': 'roc_auc',
    'pr_auc': 'average_precision',
    'accuracy': make_scorer(accuracy_score)
}

# --- Benchmark Models with Cross-Validation ---

benchmark_models_cv = {
    'Logistic Regression': LogisticRegression(solver='lbfgs', penalty='l2', C=1.0, max_iter=1000, random_state=10),
    'KNN': KNeighborsClassifier(n_neighbors=5, weights='uniform'),
    'Decision Tree': DecisionTreeClassifier(criterion='entropy', max_depth=5, random_state=10),
    'Bagging': BaggingClassifier(n_estimators=100, random_state=10),
    'Random Forest': RandomForestClassifier(n_estimators=100, max_depth=5, random_state=10),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=10),
    'XGBoost': XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=5, verbosity=0, use_label_encoder=False, random_state=10),
    'AdaBoost': AdaBoostClassifier(n_estimators=100, learning_rate=0.1, random_state=10),
    'CatBoost': CatBoostClassifier(iterations=100, learning_rate=0.1, depth=6, verbose=0, random_state=10),
    'LightGBM': LGBMClassifier(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=10, verbose=-1)
}

# --- Cross-Validation Loop and Classification Reports ---

results_cv = []
for name, model in benchmark_models_cv.items():
    pipe = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    # a. Evaluate metrics via cross_validate
    cv_res = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        return_train_score=False,
        n_jobs=-1
    )

    # b. Generate cross‑validated predictions for classification_report
    y_pred_cv = cross_val_predict(
        pipe,
        X_train,
        y_train,
        cv=cv,
        n_jobs=-1
    )

    # c. Print classification report
    print(f"\n=== {name} (5‑fold CV) ===")
    print(classification_report(y_train, y_pred_cv, digits=2))
    print("-" * 80)

    # d. Store the averaged metrics
    results_cv.append({
        'Model': name,
        'F2-Score': np.mean(cv_res['test_f2']),
        'Recall': np.mean(cv_res['test_recall']),
        'PR-AUC': np.mean(cv_res['test_pr_auc']),
        'F1-Score': np.mean(cv_res['test_f1']),
        'Precision': np.mean(cv_res['test_precision']),
        'ROC-AUC': np.mean(cv_res['test_roc_auc']),
        'Accuracy': np.mean(cv_res['test_accuracy'])
    })

# --- Create Summary DataFrame ---

benchmark_cv_df = pd.DataFrame(results_cv).set_index('Model')
benchmark_cv_df = benchmark_cv_df.sort_values(
    by=['F2-Score', 'Recall', 'PR-AUC', 'F1-Score', 'Precision', 'ROC-AUC', 'Accuracy'],
    ascending=False
).round(4)

from IPython.display import display
display(benchmark_cv_df)
````

````python
# BENCHMARK MODEL USING CROSS-VALIDATION (WITH SMOTE-ENN)

results_cv_smoteenn = []

for name, model in benchmark_models_cv.items():
    # Build pipeline: Preprocessing ➜ SMOTEENN ➜ Model
    pipe = ImbPipeline([
        ('preprocessor', preprocessor),
        ('smote_enn', SMOTEENN(random_state=10)),
        ('classifier', model)
    ])

    # a. Evaluate metrics using cross_validate
    cv_res = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        return_train_score=False,
        n_jobs=-1
    )

    # b. Generate cross-validated predictions for classification_report
    y_pred_cv = cross_val_predict(
        pipe,
        X_train,
        y_train,
        cv=cv,
        n_jobs=-1
    )

    # c. Print classification report
    print(f"\n=== {name} (5-fold CV + SMOTE-ENN) ===")
    print(classification_report(y_train, y_pred_cv, digits=2))
    print("-" * 80)

    # d. Store averaged metrics
    results_cv_smoteenn.append({
        'Model': name,
        'F2-Score': np.mean(cv_res['test_f2']),
        'Recall': np.mean(cv_res['test_recall']),
        'PR-AUC': np.mean(cv_res['test_pr_auc']),
        'F1-Score': np.mean(cv_res['test_f1']),
        'Precision': np.mean(cv_res['test_precision']),
        'ROC-AUC': np.mean(cv_res['test_roc_auc']),
        'Accuracy': np.mean(cv_res['test_accuracy'])
    })

# --- Create Summary DataFrame ---

benchmark_cv_smoteenn_df = pd.DataFrame(results_cv_smoteenn).set_index('Model')
benchmark_cv_smoteenn_df = benchmark_cv_smoteenn_df.sort_values(
    by=['F2-Score', 'Recall', 'PR-AUC', 'F1-Score', 'Precision', 'ROC-AUC', 'Accuracy'],
    ascending=False
).round(4)

display(benchmark_cv_smoteenn_df)
````

**b. Key Insights**

1. **Understand the Effects of Imbalance**  
   - When evaluated without any resampling (No CV, No SMOTEENN), **Bagging** achieves the highest recall (**0.8726**) and F2‑Score (**0.8782**), but simpler models suffer: **Logistic Regression** recall is only **0.5605** (F2‑Score = **0.5867**), and **Random Forest** recall drops to **0.4586** (F2‑Score = **0.5085**).  
   - Introducing cross‑validation without resampling (CV, No SMOTEENN) further exposes this bias: **Logistic Regression** recall falls to **0.5298** (F2‑Score = **0.5635**) and **Random Forest** to **0.3907** (F2‑Score = **0.4398**), while ensemble methods like **Gradient Boosting** and **LightGBM** hold recall around **0.7633** and **0.7169** respectively (F2‑Scores = **0.7848**, **0.7434**).  
   - In all cases, **PR‑AUC** remains moderate to high (e.g. Logistic Regression = **0.7223**, Random Forest = **0.7465** under CV), indicating models learn some separation, but low recall drags down the **F2‑Score**, which penalizes missed churners more heavily.  
   - High accuracy (> 0.89) and decent precision (Logistic Regression = **0.7595**, Random Forest = **0.8885**) give a false sense of performance; the models fail to catch a large share of actual churners, directly undermining the business goal of early churn identification.  
   - This imbalance-driven gap between precision and recall highlights why prioritizing **F2‑Score** (which weights recall twice as heavily as precision) is critical: without resampling, even strong classifiers underperform on the primary objective of detecting churn.

2. **SMOTEENN as the Preferred Resampling Strategy**  
   - Applying **SMOTEENN** within cross-validation pipelines ensures resampling occurs without data leakage, preserving valid performance estimates.  
   - The combined approach leverages **SMOTE** to synthetically augment the minority class, addressing imbalance by enriching minority representation in feature space.  
   - Subsequently, **ENN** cleans ambiguous regions by removing majority samples that overlap or conflict with the minority class, sharpening decision boundaries.  
   - This synergy results in improved **recall** (e.g., KNN recall rises to **0.9296**, Logistic Regression to **0.8518**) while mitigating the false positive inflation typical of SMOTE alone, thus enhancing generalization.  

3. **Why SMOTE Alone Is Not Enough**  
   - Using **SMOTE** without cleaning (ENN) can lead to overfitting, especially for proximity-based classifiers like **KNN**, by creating synthetic minority points close to noisy or borderline majority instances.  
   - This inflates **recall** but often severely degrades **precision** and overall model robustness, as shown by lower precision scores (e.g., KNN precision drops when SMOTE is used alone).  
   - The lack of noise removal causes models to confuse borderline majority samples as churners, increasing false positives and harming practical usefulness.  
   - Therefore, **ENN** plays a critical role by pruning noisy majority examples, which clarifies class boundaries and balances the trade-off between recall and precision, leading to more reliable churn detection.  
   - Even after **hyperparameter tuning**, relying on **SMOTE alone can mislead model selection**. Since **F2‑Score is recall-heavy**, models like **KNN** may appear optimal due to inflated recall, even when practical performance (e.g., precision or PR‑AUC) is poor. This reinforces the importance of **combining SMOTE with ENN** to ensure valid model comparisons.

4. **Post‑Resampling Performance Gains**  
   - **KNN** sees recall surge from **0.5523** (CV, no SMOTEENN) to **0.9296** with SMOTEENN, making it the top recall performer. Precision, however, drops from **0.8119** to **0.4527**, as SMOTE’s synthetic neighbors cause KNN to label more borderline points as churners. The net effect is a substantial F2‑Score increase from **0.5896** to **0.7677**, reflecting the priority on catching churners even at the expense of false positives.  
   - **Logistic Regression** recall improves from **0.5298** to **0.8518**, while precision declines from **0.7595** to **0.4202**. Its F2‑Score climbs from **0.5635** to **0.7066**, demonstrating that linear models also greatly benefit from balanced class representation.  
   - **Ensemble models** like **Gradient Boosting** and **CatBoost** maintain high precision before resampling (around **0.89** and **0.90**) and boost recall by 6 to 14 points with SMOTE-ENN (Gradient Boosting from **0.7633** to **0.8277**, CatBoost from **0.6720** to **0.8158**). This results in improved F2-Scores (Gradient Boosting from **0.7848** to **0.7896**, CatBoost from **0.7076** to **0.7822**). While precision drops moderately (to about **0.67**), the trade-off preserves or enhances PR-AUC, showing balanced and reliable gains from class balancing.  

5. **Impact on Evaluation Metrics**  
   - **SMOTEENN** improves key metrics like **Recall**, **F2‑Score**, and **PR‑AUC**, aligning closely with the business objective of catching churners early.  
   - For instance, **Logistic Regression**'s F2‑Score rises from **0.5635** to **0.7066**, and **KNN** improves from **0.6151** to **0.7516**, demonstrating substantial benefit from resampling.  
   - The **Bagging Classifier** achieves the **highest F2‑Score overall at 0.8773**, along with strong **recall (0.8802)** and **precision (0.9657)**, confirming it as the **most effective model** for prioritizing churn detection.  
   - While **PR‑AUC** declines slightly for some models (e.g., **KNN** from **0.7274** to **0.5999**), it remains high for top performers such as **Gradient Boosting** (**0.9058** to **0.7960**) and **CatBoost** (**0.8933** to **0.8101**), preserving decision confidence.  
   - Importantly, if **SMOTE is used without ENN**, models like **KNN** may achieve **high recall** and **high F2‑Scores**, which could **falsely suggest optimality**. This highlights that **model selection must consider not just recall-driven metrics**, but also **precision and PR‑AUC** to ensure practical reliability.  
   - By prioritizing **F2‑Score**, which weights recall more heavily, **SMOTEENN** ensures that models shift toward higher detection rates of churners, even at the cost of some precision, effectively supporting the project's objective.

<br>

**c. Key Takeaways**

- **SMOTEENN** combined with cross-validation delivers the **most effective improvements in F2‑Score**.  
- These gains are driven by substantial **recall improvements** that outweigh precision losses, directly enhancing the model’s ability to detect churners.  
- Crucially, **using SMOTE alone can be misleading**, especially for recall-sensitive metrics like **F2‑Score**. Models such as **KNN** may appear optimal due to inflated recall, despite poor real-world performance (e.g. low precision or PR‑AUC). This reinforces the need for **noise reduction via ENN** to ensure meaningful model selection.  
- **Hyperparameter tuning** remains essential after resampling to adapt to the altered data distribution:  
  - For **Logistic Regression**, tuning **C** controls regularization strength against synthetic overlap.  
  - For **tree-based ensembles**, adjusting **depth**, **learning rate**, and **estimators** helps retain precision while capitalizing on recall gains.  
- Overall, combining **SMOTEENN with targeted hyperparameter tuning** yields models that not only **maximize F2‑Score**, but also maintain solid **precision** and **PR‑AUC**, making them highly aligned with the business goal: **accurately identifying as many at-risk churners as possible without overwhelming false positives**.

#### **III.3.2 Hyperparameter Tuning**

**Hyperparameter tuning** is the process of optimizing the external configuration settings of a machine learning model, known as **hyperparameters**, which are not learned from the training data but instead control how the learning algorithm operates. In churn classification, hyperparameters include **`C`** (inverse regularization strength) in Logistic Regression, **`n_neighbors`** in K-Nearest Neighbors, or **`n_estimators`** and **`max_depth`** in ensemble models such as Random Forest and Gradient Boosting.

Two widely used techniques for hyperparameter tuning are **grid search** and **randomized search**:

- **Grid search** exhaustively evaluates all combinations in a predefined parameter grid. This method guarantees that the best combination within the grid is found, but it becomes **computationally expensive** and **time-consuming** as the number of hyperparameters and values increases.

- **Randomized search** samples a fixed number of combinations from the parameter space according to specified distributions. It offers **faster exploration** of a **broader range** of values with reduced computation time, although it does **not guarantee** finding the absolute best combination.

In this project, **RandomizedSearchCV** is selected for hyperparameter tuning due to its superior **efficiency**, **scalability**, and **practical effectiveness**, particularly when optimizing **10 different classification models** with moderately large hyperparameter spaces. This approach enables comprehensive tuning while avoiding excessive computational cost, making it well-suited for large-scale benchmarking.

Compared to **GridSearchCV**, which exhaustively evaluates all parameter combinations, **RandomizedSearchCV** samples a fixed number of configurations from the hyperparameter space. This allows for:

- **Significant reduction in computation time**  
  Exhaustively tuning 10 models across full parameter grids would be extremely costly. RandomizedSearchCV reduces this overhead by testing only a representative sample of combinations.

- **Broader exploration of the search space**  
  Random sampling enables wider coverage, improving the chances of finding **high-performing configurations**, especially when performance is sensitive to a few key hyperparameters.

- **Efficient allocation of resources**  
  Empirical studies show that only a subset of parameters typically drive performance. RandomizedSearchCV helps avoid wasting time on unimportant combinations.

- **Controlled experimentation**  
  With a fixed number of iterations, tuning becomes more predictable and manageable, allowing consistent time budgets across models.

To ensure robust evaluation during tuning, **Stratified 5‑fold cross‑validation** is employed, which divides the training data into five folds that preserve the class distribution. The model is trained and validated across these folds, helping to **mitigate overfitting** and ensuring that the selected hyperparameters **generalize well** across different data subsets.  

<br>

#### **Why Hyperparameter Tuning Matters**
In practice, **hyperparameter tuning is essential** for most classification models, particularly complex ones like **ensemble methods**, to reach peak performance on business‑critical metrics such as **F2-Score**, **recall** and **PR‑AUC**. While simple models like Decision Trees or KNN can work with default settings, careful tuning of parameters like **`max_depth`**, **`n_estimators`**, **`learning_rate`**, and **`n_neighbors`** often yields significant improvements in both **stability** and **accuracy** when deployed in real‑world churn prediction tasks.

Below is a summary of each candidate model and its most influential hyperparameters:  

| Model                  | Key Tunable Hyperparameters                                                                 |
|------------------------|---------------------------------------------------------------------------------------------|
| Logistic Regression    | `C`, `penalty`, `solver`, `class_weight`                                                   |
| K-Nearest Neighbors    | `n_neighbors`, `weights`, `metric`                                                         |
| Decision Tree          | `max_depth`, `min_samples_split`, `min_samples_leaf`, `criterion`                          |
| Bagging Classifier     | `n_estimators`, `max_samples`, `max_features`, `bootstrap`, `bootstrap_features`           |
| Random Forest          | `n_estimators`, `max_depth`, `min_samples_split`, `max_features`, `class_weight`           |
| Gradient Boosting      | `n_estimators`, `learning_rate`, `max_depth`, `subsample`, `min_samples_split`, `min_samples_leaf`, `max_features` |
| XGBoost                | `n_estimators`, `learning_rate`, `max_depth`, `min_child_weight`, `subsample`, `colsample_bytree`, `gamma` |
| AdaBoost               | `n_estimators`, `learning_rate`, `estimator__max_depth`                                    |
| CatBoost               | `iterations`, `learning_rate`, `depth`, `l2_leaf_reg`, `random_strength`, `bagging_temperature` |
| LightGBM               | `n_estimators`, `learning_rate`, `max_depth`, `num_leaves`, `bagging_fraction`, `min_gain_to_split` |

#### **a. Logistic Regression**

````python
# HYPERPARAMETER TUNING FOR LOGISTIC REGRESSION

# Define parameter space for Logistic Regression
logreg_params = {
    'classifier__C': [0.1, 1, 10],                     # Inverse of regularization strength; smaller values specify stronger regularization
    'classifier__penalty': ['l1', 'l2'],               # Type of regularization: L1 for sparsity, L2 for ridge penalty
    'classifier__solver': ['liblinear'],                # Algorithm for optimization; supports both L1 and L2 penalties
    'classifier__class_weight': [None, 'balanced']      # Handle class imbalance: 'balanced' adjusts weights inversely to class frequencies
}

# Setup Stratified K-Fold CV
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=10)

# Logistic Regression Pipeline with SMOTE-ENN
logreg_pipeline = ImbPipeline(steps=[
    ('preprocessor', preprocessor),
    ('smoteenn', SMOTEENN(random_state=10)),
    ('classifier', LogisticRegression(random_state=10, max_iter=1000))
])

# Define F2-score scorer
f2_scorer = make_scorer(fbeta_score, beta=2)

# Randomized Search CV for Logistic Regression
logreg_random_search = RandomizedSearchCV(
    logreg_pipeline,
    param_distributions=logreg_params,
    n_iter=20,
    scoring=f2_scorer,
    cv=cv,
    verbose=1,
    random_state=10,
    n_jobs=-1
)

# Fit the model
logreg_random_search.fit(X_train, y_train)

# Best parameters
print('\n=== Logistic Regression Best Params ===')
print(logreg_random_search.best_params_)

# Cross-validated predictions for metric calculations
y_pred_logreg_tuning = cross_val_predict(logreg_random_search.best_estimator_, X_train, y_train, cv=cv, method='predict')
y_proba_logreg_tuning = cross_val_predict(logreg_random_search.best_estimator_, X_train, y_train, cv=cv, method='predict_proba')[:,1]

# Calculate classification metrics
f2_logreg_tuning = fbeta_score(y_train, y_pred_logreg_tuning, beta=2)
recall_logreg_tuning = recall_score(y_train, y_pred_logreg_tuning)
pr_logreg_tuning = average_precision_score(y_train, y_proba_logreg_tuning)
f1_logreg_tuning = f1_score(y_train, y_pred_logreg_tuning)
precision_logreg_tuning = precision_score(y_train, y_pred_logreg_tuning)
roc_logreg_tuning = roc_auc_score(y_train, y_proba_logreg_tuning)
accuracy_logreg_tuning = accuracy_score(y_train, y_pred_logreg_tuning)

# Store results in dictionary
logreg_tuning_results = {
    'Model': 'Logistic Regression',
    'F2-Score': f2_logreg_tuning,
    'Recall': recall_logreg_tuning,
    'PR-AUC': pr_logreg_tuning,
    'F1-Score': f1_logreg_tuning,
    'Precision': precision_logreg_tuning,
    'ROC-AUC': roc_logreg_tuning,
    'Accuracy': accuracy_logreg_tuning
}

# Print results
print('\n=== Logistic Regression Evaluation Metrics ===')
for metric, value in logreg_tuning_results.items():
    if metric != 'Model':
        print(f"{metric}: {value:.4f}")
````

#### **f. Gradient Boosting**

````python
# HYPERPARAMETER TUNING FOR GRADIENT BOOSTING CLASSIFIER

# Define parameter space for Gradient Boosting
gbr_params = {
    'classifier__n_estimators': [100, 200, 300],                # Number of boosting stages to perform
    'classifier__learning_rate': [0.01, 0.05, 0.1],             # Step size shrinkage to prevent overfitting
    'classifier__max_depth': [3, 5, 7],                         # Maximum depth of the individual classification estimators
    'classifier__min_samples_split': [5, 10, 15],               # Minimum samples required to split an internal node
    'classifier__min_samples_leaf': [2, 4, 6],                  # Minimum number of samples required to be at a leaf node
    'classifier__subsample': [0.7, 0.8, 1.0],                   # Fraction of samples used for fitting individual base learners
    'classifier__max_features': ['sqrt', 'log2']                # Number of features considered when looking for best split
}

# Setup Stratified K-Fold CV
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=10)

# Gradient Boosting Pipeline with SMOTE-ENN
gbr_pipeline = ImbPipeline(steps=[
    ('preprocessor', preprocessor),
    ('smoteenn', SMOTEENN(random_state=10)),
    ('classifier', GradientBoostingClassifier(random_state=10))
])

# Define F2-score scorer
f2_scorer = make_scorer(fbeta_score, beta=2)

# Randomized Search CV for Gradient Boosting
gbr_random_search = RandomizedSearchCV(
    gbr_pipeline,
    param_distributions=gbr_params,
    n_iter=20,
    scoring=f2_scorer,
    cv=cv,
    verbose=1,
    random_state=10,
    n_jobs=-1
)

# Fit the model
gbr_random_search.fit(X_train, y_train)

# Best parameters
print('\n=== Gradient Boosting Best Params ===')
print(gbr_random_search.best_params_)

# Cross-validated predictions
y_pred_gbr_tuning = cross_val_predict(gbr_random_search.best_estimator_, X_train, y_train, cv=cv, method='predict')
y_proba_gbr_tuning = cross_val_predict(gbr_random_search.best_estimator_, X_train, y_train, cv=cv, method='predict_proba')[:, 1]

# Calculate classification metrics
f2_gbr_tuning = fbeta_score(y_train, y_pred_gbr_tuning, beta=2)
recall_gbr_tuning = recall_score(y_train, y_pred_gbr_tuning)
pr_gbr_tuning = average_precision_score(y_train, y_proba_gbr_tuning)
f1_gbr_tuning = f1_score(y_train, y_pred_gbr_tuning)
precision_gbr_tuning = precision_score(y_train, y_pred_gbr_tuning)
roc_gbr_tuning = roc_auc_score(y_train, y_proba_gbr_tuning)
accuracy_gbr_tuning = accuracy_score(y_train, y_pred_gbr_tuning)

# Store results in dictionary
gbr_tuning_results = {
    'Model': 'Gradient Boosting',
    'F2-Score': f2_gbr_tuning,
    'Recall': recall_gbr_tuning,
    'PR-AUC': pr_gbr_tuning,
    'F1-Score': f1_gbr_tuning,
    'Precision': precision_gbr_tuning,
    'ROC-AUC': roc_gbr_tuning,
    'Accuracy': accuracy_gbr_tuning
}

# Print results
print('\n=== Gradient Boosting Evaluation Metrics ===')
for metric, value in gbr_tuning_results.items():
    if metric != 'Model':
        print(f"{metric}: {value:.4f}")
````

#### **c. Key Insights**

1. **CatBoost Achieves the Highest F2‑Score**  
   - **CatBoost** rises from **0.7822** to **0.8623** in F2‑Score, the largest uplift of any model.  
   - Key drivers include deeper trees (`depth=8`), stronger regularization (`l2_leaf_reg=3`, `random_strength=1`) and focused sampling (`bagging_temperature=0`).  
   - This configuration delivers **recall 0.8982** and **PR‑AUC 0.8718**, demonstrating best‑in‑class balance between catching churners and limiting false positives.

2. **Gradient Boosting Delivers Robust Ensemble Gains**  
   - **Gradient Boosting** improves from **0.7896** to **0.8384**, powered by **300 estimators**, `max_depth=7` and `min_samples_split=15`.  
   - Achieves **recall 0.8713** and **PR‑AUC 0.8690**, matching CatBoost’s precision‑recall trade‑off while offering slightly more interpretable tree structure.

3. **LightGBM Offers Fast, Balanced Performance**  
   - **LightGBM** climbs from **0.7703** to **0.8140**, tuning `num_leaves=20`, `max_depth=6`, `bagging_fraction=0.7`.  
   - With **recall 0.8488** and **PR‑AUC 0.8148**, it achieves the highest post‑tuning accuracy (0.9107) among top‑tier models, making it ideal where speed and reliability matter.

4. **Bagging and XGBoost Strengthen Ensemble Stability**  
   - **Bagging Classifier** increases F2‑Score from **0.7940** to **0.8105** by enabling both `bootstrap=True` and `bootstrap_features=True`, yielding **recall 0.8488** and **PR‑AUC 0.8240**.  
   - **XGBoost** moves from **0.7663** to **0.8071**, with `n_estimators=200`, `max_depth=7`, `subsample=1.0` and `colsample_bytree=0.7`, achieving **recall 0.8458** and **PR‑AUC 0.8088**.

5. **KNN’s High Recall but Precision Constraints**  
   - **KNN** maintains strong recall (0.9147) yet F2‑Score only rises from **0.7677** to **0.8025** due to limited precision (**0.5383**).  
   - Distance weighting (`weights='distance'`) and Manhattan metric (`metric='manhattan'`) improve local class detection but underscore the limitations of non‑ensemble methods on imbalanced data.

6. **Random Forest Becomes Competitive After Deep Tuning**  
   - **Random Forest** jumps from **0.6905** to **0.7958** by deepening trees (`max_depth=15`), limiting splits (`min_samples_split=5`) and halving feature space (`max_features=0.5`).  
   - Achieves **recall 0.8353** and **PR‑AUC 0.7849**, illustrating how ensemble diversity and depth counteract baseline under‑performance.

7. **AdaBoost and Single Decision Tree Lag Behind**  
   - **AdaBoost** edges up from **0.6848** to **0.7119**, with shallow base learners (`estimator__max_depth=3`) yielding **recall 0.7695**.  
   - **Decision Tree** modestly improves from **0.6680** to **0.7026** after tuning `max_depth=15`, `min_samples_split=15`, `min_samples_leaf=2`, but remains susceptible to high variance.

8. **Hyperparameter Tuning Must Align with F2-Score Priority**  
   - Models with the largest **F2-Score** uplifts (**CatBoost, Gradient Boosting, LightGBM**) combine high recall gains (> +6 points) with moderate precision retention.  
   - Blindly optimizing for recall (for example, KNN with SMOTE alone) risks inflating false positives and degrading F2 if precision collapses.  
   - Therefore, tuning should focus on **balancing recall and precision** within the SMOTE-ENN framework to maximize **F2-Score**, directly serving the project’s objective of catching churners effectively.

#### **III.3.3 Best Model Selection**

After completing **class imbalance handling** with SMOTE-ENN and **hyperparameter tuning** for ten classification models, the next step is to determine the **best‑performing model** for churn prediction. This selection is critical, as models, even when optimized and trained on balanced data, can vary in their ability to generalize to unseen samples. A well‑chosen model not only detects churners with high sensitivity but also maintains robust performance across different customer segments in real‑world deployment.

This project prioritizes seven evaluation metrics in the following order:  
1. **F2 Score**: Top priority. It emphasizes recall more than precision, making it especially suitable when false negatives are significantly more costly.  
2. **Recall**: Correct positive predictions among actual positives. Critical for minimizing missed churners.  
3. **PR‑AUC**: Area under the Precision‑Recall curve, emphasizing performance on the churn class in imbalanced data.  
4. **F1‑Score**: Harmonic mean of precision and recall, balancing false positives and false negatives.  
5. **Precision**: Correct positive predictions among predicted positives, indicating the trustworthiness of churn alerts.  
6. **ROC‑AUC**: Measures the classifier’s ability to distinguish classes across thresholds.  
7. **Accuracy**: Overall percentage of correct predictions; retained for reference but not prioritized.

**F2 Score** is the primary metric because it aligns most closely with the business objective of minimizing false negatives by weighting recall more heavily than precision. **Recall** remains critical as it directly measures the proportion of actual churners identified. **PR‑AUC** evaluates the model’s ability to correctly identify positive cases across various thresholds, which is especially important for imbalanced datasets where the positive class (churners) is rare. **F1‑Score** and **Precision** provide complementary views that balance detection effectiveness with false alarm rates.

**ROC‑AUC** and **accuracy** serve as secondary metrics, offering a general sense of model discrimination and overall correctness. Due to class imbalance, these metrics can be misleading if used alone, as they may reflect performance dominated by the majority “no churn” class.

The final model selection will focus on the classifier that achieves the **highest F2 Score**, supported by strong **recall** and **PR‑AUC**, while also considering **stability** (low variance in cross‑validation), **interpretability** (ease of explanation to stakeholders), and **efficiency** (inference speed and resource usage) for seamless integration into the customer retention workflow.

#### **III.3.4 Final Model Explanation**

Based on a **comprehensive analysis** involving benchmarking of multiple classification models, **handling class imbalance**, **hyperparameter tuning**, and detailed evaluation using consistent performance metrics, **CatBoost** emerged as the **best-performing model** for churn prediction. It consistently achieved the **highest F2‑Score** and **recall values**, demonstrated **stable and robust results across cross‑validation folds**, and provided strong **generalization**. Below is an in‑depth explanation of CatBoost and the rationale behind its selection.

#### **a. What is CatBoost?**

**CatBoost (Categorical Boosting)** is a **cutting‑edge gradient boosting library** developed by Yandex, designed to deliver **robust, accurate, and efficient predictive models**, particularly in datasets with many **categorical features**. Unlike most boosting frameworks, CatBoost **natively processes categorical variables** without manual one‑hot encoding, preserving feature interactions and reducing the risk of overfitting ([Koehrsen et al., 2020](https://www.researchgate.net/publication/347351801_Comparison_of_the_CatBoost_Classifier_with_other_Machine_Learning_Methods)). It builds an ensemble of **symmetric decision trees** (also called oblivious trees) in a **sequential boosting** fashion, optimizing a specified loss function while incorporating novel techniques to **mitigate bias and prevent overfitting**, which is critical in **imbalanced classification tasks** such as churn prediction.  
<br>

#### **b. How CatBoost Works**

CatBoost’s training pipeline includes several **innovations** that drive its superior performance:

1. **Ordered Target Statistics**  
   - Categorical features are transformed into **target‑based statistics** computed in an **ordered** sequence to avoid target leakage.  
   - Each category’s statistic is computed using only **preceding samples**, ensuring the model cannot “peek” at future information, which is essential for reliable performance on imbalanced or time‑series data.

2. **Symmetric Tree Growing**  
   - Trees are grown in a **balanced, symmetric** manner, where each split at a given depth uses the same feature and threshold across all branches.  
   - This uniform structure simplifies inference, speeds up prediction, and reduces variance by ensuring **consistent treatment** of all data partitions, leading to better generalization on minority classes.

3. **Gradient‑Based One‑Hot Encoding (GHOE)**  
   - Instead of naive one‑hot encoding, CatBoost applies a **gradient‑aware encoding** that selects the most **informative binary features** derived from categoricals based on gradient signals.  
   - This reduces dimensionality, preserves key interactions, and minimizes overfitting caused by sparse high‑dimensional representations.

4. **Efficient Boosting and Robust Regularization**  
   - The library combines **learning rate shrinkage** with **L2 leaf regularization** (`l2_leaf_reg`) to penalize complex tree leaves and prevent overfitting.  
   - **Bayesian‑inspired bagging** controlled by `bagging_temperature` introduces controlled randomness in sample weighting rather than discarding data, enhancing robustness without losing critical information.

5. **Advanced Loss Estimation via Ordered Boosting**  
   - CatBoost uses an **ordered boosting** algorithm to compute gradients and Hessians in a way that **reduces prediction shift**—the bias incurred when using the same data both to fit trees and to compute residuals.  
   - This method lowers both bias and variance, stabilizing learning and improving detection of the minority churn class.

Together, these features enable **CatBoost** to deliver **exceptional accuracy**, **reduced overfitting**, and **fast inference** compared to other gradient boosting frameworks, making it an ideal choice for **churn prediction** and similar imbalanced classification tasks.  
<br>

#### **c. Core Advantages**

1. **Native Categorical Support**  
   - **Automatic encoding** of categorical variables via ordered target statistics and gradient‑based encoding preserves original feature relationships.  
   - **Eliminates manual one‑hot encoding**, reducing dataset dimensionality and preprocessing overhead.  
   - **Maintains interaction effects** between categories and numerical features, improving predictive power.

2. **Reduced Overfitting**  
   - **Ordered boosting** prevents target leakage by using only past samples for gradient and category statistic calculations.  
   - **Symmetric tree structure** enforces consistent splits across branches, lowering variance compared to asymmetric trees.  
   - **L2 leaf regularization** (`l2_leaf_reg`) and **Bayesian‑style bagging** (`bagging_temperature`) further penalize complexity and stabilize learning on noisy or resampled data.

3. **Fast and Memory‑Efficient**  
   - **Oblivious trees** share split conditions at each depth, enabling **vectorized computations** and faster inference.  
   - **Optimized C++ implementation** with multi‑threading leverages modern CPUs for rapid model training and predictions.  
   - **Gradient‑based one‑hot encoding** reduces feature space compared to naive encoding, conserving memory on high‑cardinality data.

4. **High Accuracy Out of the Box**  
   - **Robust default hyperparameters** provide strong baseline performance without extensive tuning.  
   - **Gradient‑aware encoding** extracts the most informative categorical splits, boosting early model quality.  
   - **Second‑order gradient estimation** (using both gradient and Hessian) yields more precise updates and faster convergence.

5. **Versatile Loss Functions**  
   - **Built‑in support** for classification, regression, and ranking objectives covers a wide range of tasks.  
   - **Custom loss** capability allows practitioners to define domain‑specific objectives, such as **F2‑score** optimization for churn detection.  
   - **Multi‑class and multi‑label** extensions enable use in complex prediction scenarios.  
<br>

#### **d. Key Limitations**

1. **Hyperparameter Complexity**  
   - Tuning specialized parameters such as **`bagging_temperature`**, **`random_strength`**, and **`depth`** can require extensive experimentation.  
   - Default settings may not fully optimize performance on highly imbalanced or noisy datasets without targeted adjustment.

2. **Interpretability Challenges**  
   - **Symmetric trees** lack the clear hierarchical split paths of asymmetric trees, making them less intuitive to interpret without specialized visualization tools.  
   - Extracting individual feature contributions typically requires additional libraries (e.g., SHAP) for explainability.

3. **Computational Load on Extreme Cardinality**  
   - Calculation of **ordered target statistics** for very high‑cardinality categorical features can become computationally intensive.  
   - Memory and CPU requirements may increase substantially when encoding thousands of unique category levels.  
<br>

#### **e. Ideal Use Cases**

1. **Datasets with Many Categorical Features**  
   - Applications in **retail**, **marketing**, and **customer analytics**, where product IDs, location codes, and demographic segments are prevalent.  
   - Use cases where preserving category interactions is critical for accurate modeling.

2. **Imbalanced Classification**  
   - Tasks like **churn prediction**, **fraud detection**, and **rare event forecasting**, where positive class examples are scarce.  
   - Scenarios that benefit from built‑in mechanisms to reduce bias and overfitting on minority classes.

3. **Rapid Prototyping and Production Deployment**  
   - Projects requiring **strong baseline performance** with minimal data preparation, enabling faster iteration cycles.  
   - **Real‑time scoring** environments where low latency and efficient inference are essential, such as online recommendation systems or real‑time alerts.

4. **Custom Objective Optimization**  
   - Domains that demand optimization of non‑standard metrics (e.g., **F2‑score**, **weighted recall**, or **business‑specific risk scores**).  
   - Situations where native support for custom loss functions accelerates development compared to manual gradient boosting implementations.    
<br>

#### **f. Hyperparameter Explanations**

#### **g. Why CatBoost is the Best Model?**

1. **Superior Handling of Categorical Data**  
   - **CatBoost’s native encoding** transforms each categorical feature into meaningful numerical representations using ordered target statistics and gradient‑based encoding.  
   - This preserves **customer segment information**, such as subscription tier, region, and engagement level, without the dimensionality explosion or information loss associated with one‑hot encoding.  
   - As a result, CatBoost captures complex patterns in user behavior that drive churn, leading to more accurate predictions.

2. **Robust Performance on Imbalanced Classes**  
   - The combination of **ordered boosting** and **Bayesian bagging** yields a **high F2‑Score (0.8623)** and **PR‑AUC (0.8718)**, reflecting strong detection of churners while controlling false positives.  
   - By computing gradients and category statistics only from preceding samples, CatBoost avoids target leakage and bias, which commonly undermine performance on minority classes.  
   - This robustness ensures that rare churn events are identified reliably, a critical requirement for effective retention strategies.

3. **Stable Cross‑Validation Results**  
   - CatBoost exhibited **low variance across cross‑validation folds**, indicating consistent generalization to unseen data.  
   - Even with extensive feature interactions and synthetic examples from SMOTE‑ENN, the model maintained **balanced recall and precision**, demonstrating that its regularization and sampling strategies prevent overfitting.  
   - This stability provides confidence that production performance will mirror validation results.

4. **Efficient Training and Inference**  
   - The use of **symmetric (oblivious) trees** and an **optimized C++ implementation** results in **fast training times** and **low-latency predictions**, suitable for real‑time scoring.  
   - Multi‑threading capabilities allow CatBoost to **scale across CPU cores**, reducing end‑to‑end model update and deployment cycles.  
   - These efficiencies lower infrastructure costs and enable rapid responses to emerging churn signals.

5. **Minimal Preprocessing Overhead**  
   - With **effective default hyperparameters** and **built-in categorical support**, CatBoost requires **minimal feature engineering**, accelerating model development.  
   - Data scientists can focus on **business insights**, such as identifying key churn drivers, instead of spending excessive time on encoding and transformation pipelines.  
   - This streamlined workflow enhances productivity and shortens time‑to‑value for churn mitigation initiatives.

6. **Insightful Feature Importances**  
   - Despite using symmetric trees, CatBoost provides **accurate feature importance metrics** and seamless integration with **SHAP** (SHapley Additive exPlanations).  
   - These tools reveal the **most influential variables**, such as tenure, product usage frequency, and support interactions, enabling stakeholders to understand and act on churn drivers.  
   - This transparency supports **data‑driven decision making** across marketing, customer success, and product teams.

#### **III.3.5 Test Set Evaluation and Metric Analysis**

This section rigorously evaluates the **final CatBoost classification model** on the **held‑out test set**, providing an unbiased measure of its ability to generalize to new, unseen customer data. Key classification metrics, including **F2‑Score**, **Recall**, **PR‑AUC**, **F1‑Score**, **Precision**, **ROC‑AUC**, and **Accuracy** are analyzed to deliver a comprehensive understanding of the model’s effectiveness in identifying churners while managing false positives. This evaluation validates the impact of hyperparameter tuning and training strategies, highlighting both robustness in capturing true churn events and potential areas for fine‑tuning before deployment. The analysis is supported by the **final model’s churn predictions on the test set using the optimal CatBoost parameters**, a **barplot comparing CatBoost performance metrics across model development stages**, a **confusion matrix** summarizing classification outcomes on the test set, and a **precision‑recall curve** visualizing the trade‑off between recall and precision on the final test set.  
<br>

The following table explains the **key metrics** used in this churn‑prediction evaluation, outlining their **full names**, **what they measure**, **ideal values**, and **interpretation and insights**:

| **Metric**   | **Full Name**                     | **What It Measures**                                                                                              | **Ideal Value**                | **Interpretation and Insights**                                                                                                                                                                                                                                                                                                                                                                                                           |
|--------------|-----------------------------------|--------------------------------------------------------------------------------------------------------------------|--------------------------------|---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| **F2‑Score** | F₂‑Score                          | Harmonic mean of **precision** and **recall** with **recall weighted twice** as much as precision.                 | **Higher is better**           | Emphasizes recall over precision to reflect the business cost of missed churners. By weighting false negatives more heavily, it ensures the model captures the majority of churners even if precision is moderately reduced. Ideal when retaining true churners is far more valuable than avoiding extra retention outreach.                                                                                                                    |
| **Recall**   | Sensitivity / True Positive Rate  | Proportion of actual churners correctly identified by the model.                                                   | **Closer to 1**                | Measures the model’s ability to find every churner. Critical when each undetected churner represents lost revenue. High recall reduces missed churn cases but may introduce more false positives if optimized alone. Must be balanced with precision‑aware metrics to avoid overwhelming retention resources.                                                                                                                          |
| **PR‑AUC**   | Area Under Precision‑Recall Curve | Aggregate measure of precision versus recall across all classification thresholds.                                 | **Higher is better**           | Reflects model performance on the minority (churn) class in imbalanced settings. A high PR‑AUC indicates that the model maintains solid precision even at high recall levels. It is more informative than ROC‑AUC when churn events are rare, because it directly evaluates performance on the positive (churn) class.                                                                                               |
| **F1‑Score** | F₁‑Score                          | Harmonic mean of precision and recall with **equal weighting**.                                                     | **Higher is better**           | Balances false positives and false negatives equally. Useful when both types of error carry significant cost. F1‑Score offers a single summary metric of overall classification quality, but does not distinguish the relative cost of missing churners versus over‑predicting churn.                                                                                                                                |
| **Precision**| Positive Predictive Value         | Proportion of predicted churners that are actual churners.                                                         | **Closer to 1**                | Indicates the reliability of churn predictions. High precision means fewer wasted retention efforts on non‑churners. In this context precision is secondary to recall, but remains important to prevent unnecessary outreach. Precision must not fall so low that the retention system becomes inundated with false alarms.                                                                                                      |
| **ROC‑AUC**  | Area Under ROC Curve              | Measure of separability between churners and non‑churners, plotting true positive rate against false positive rate. | **Higher is better**           | Summarizes discriminative ability across all thresholds. While widely used, ROC‑AUC can overstate performance on imbalanced data because it incorporates true negatives heavily. It is a useful complementary metric but less informative than PR‑AUC or recall‑weighted scores for churn.                                                                                                                                            |
| **Accuracy** | Overall Accuracy                  | Proportion of all correct predictions (both churners and non‑churners).                                            | **Closer to 1** but not prioritized | Gives a broad sense of correctness but can be misleading when one class dominates. In churn prediction, high accuracy can mask poor detection of churners if the model simply predicts the majority non‑churn class. Accuracy is reported for completeness but is not used to guide model selection.                                                                                                   |

<br>

**Final Priority Order:**  
**F2‑Score > Recall > PR‑AUC > F1‑Score > Precision > ROC‑AUC > Accuracy**

By prioritizing **F2‑Score**, we align model selection with the business objective of capturing as many true churners as possible while still managing false positives. **Recall** remains central to minimizing missed churners, but **F2‑Score** ensures precision is not sacrificed entirely, producing a well‑balanced, actionable model.  


````python
# FINAL MODEL PREDICTION ON TEST SET USING BEST PARAMETERS (CATBOOST)

# Best hyperparameters of final model (CatBoost)
final_model_params = {
    'iterations': 300,
    'learning_rate': 0.1,
    'depth': 8,
    'l2_leaf_reg': 3,
    'bagging_temperature': 0,
    'random_strength': 1,
    'random_state': 10,
    'verbose': 0
}

# Final CatBoost pipeline with best parameters and SMOTEENN
final_model = ImbPipeline(steps=[
    ('preprocessor', preprocessor),
    ('smoteenn', SMOTEENN(random_state=10)),
    ('classifier', CatBoostClassifier(**final_model_params))
])

# Fit on training data
final_model.fit(X_train, y_train)

# Predict on test set
y_pred_final = final_model.predict(X_test)
y_proba_final = final_model.predict_proba(X_test)[:, 1]

# Evaluation metrics
f2_final = fbeta_score(y_test, y_pred_final, beta=2)
recall_final = recall_score(y_test, y_pred_final)
pr_auc_final = average_precision_score(y_test, y_proba_final)
f1_final = f1_score(y_test, y_pred_final)
precision_final = precision_score(y_test, y_pred_final)
roc_auc_final = roc_auc_score(y_test, y_proba_final)
accuracy_final = accuracy_score(y_test, y_pred_final)

# Print results
print('\n=== CatBoost Final Evaluation on Test Set ===')
print(f'F2-score  : {f2_final:.4f}')
print(f'Recall    : {recall_final:.4f}')
print(f'PR-AUC    : {pr_auc_final:.4f}')
print(f'F1-score  : {f1_final:.4f}')
print(f'Precision : {precision_final:.4f}')
print(f'ROC-AUC   : {roc_auc_final:.4f}')
print(f'Accuracy  : {accuracy_final:.4f}')
````

````python
# BARPLOT OF CATBOOST PERFORMANCE METRICS BY MODEL STAGE

# CATBOOST performance metrics dictionary
catboost_metrics = {
    'Metric': ['F2-score', 'Recall', 'PR-AUC', 'F1-score', 'Precision', 'ROC-AUC', 'Accuracy'],
    'Benchmark (CV)': [0.7076, 0.6720, 0.8933, 0.7691, 0.9004, 0.9661, 0.9306],
    'Benchmark + SMOTEENN (CV)': [0.7822, 0.8158, 0.8101, 0.7368, 0.6720, 0.9471, 0.8993],
    'Tuned (CV)': [0.8623, 0.8982, 0.8718, 0.8136, 0.7435, 0.9656, 0.9288],
    'Final Test': [0.8987, 0.9490, 0.8955, 0.8324, 0.7413, 0.9765, 0.9379]
}

df_metrics = pd.DataFrame(catboost_metrics)

# Function to prepare melted dataframe
def prep_plot_df(df, metric_list):
    filtered = df[df['Metric'].isin(metric_list)]
    melted = filtered.melt(id_vars='Metric', var_name='Stage', value_name='Value')
    return melted

# General bar plot function
def plot_classification_metrics(data, title, ylabel, ylim=(0, 1.05)):
    plt.figure(figsize=(10, 6))
    ax = sns.barplot(data=data, x='Metric', y='Value', hue='Stage', palette='crest', edgecolor='black')

    if ylim:
        plt.ylim(ylim)
    plt.title(title)
    plt.ylabel(ylabel)
    plt.xlabel('Metric')

    # Add value labels above bars
    for p in ax.patches:
        height = p.get_height()
        ax.annotate(f'{height:.4f}',
                    (p.get_x() + p.get_width() / 2, height),
                    ha='center', va='bottom',
                    fontsize=9, color='black', xytext=(0, 3),
                    textcoords='offset points')

    # Move legend to lower right outside plot
    ax.legend(title='Model Stage', bbox_to_anchor=(1.02, 0), loc='lower right', borderaxespad=0)
    plt.tight_layout()
    plt.show()

# Plot 1: F2-score, Recall
group1_metrics = ['F2-score', 'Recall']
df_group1 = prep_plot_df(df_metrics, group1_metrics)
plot_classification_metrics(df_group1, 'CatBoost Performance: F2-Score, Recall', 'Score')

# Plot 2: PR-AUC, F1-score, Precision
group2_metrics = ['PR-AUC', 'F1-score', 'Precision']
df_group2 = prep_plot_df(df_metrics, group2_metrics)
plot_classification_metrics(df_group2, 'CatBoost Performance: PR-AUC, F1-Score, Precision', 'Score')

# Plot 3: ROC-AUC, Accuracy
group3_metrics = ['ROC-AUC', 'Accuracy']
df_group3 = prep_plot_df(df_metrics, group3_metrics)
plot_classification_metrics(df_group3, 'CatBoost Performance: ROC-AUC, Accuracy', 'Score')
````

````python
# CONFUSION MATRIX OF FINAL CATBOOST MODEL ON TEST SET

# Predict labels on test set using final model (CatBoost)
y_pred_final = final_model.predict(X_test)

# Compute confusion matrix on test set
cm = confusion_matrix(y_test, y_pred_final)

# Display confusion matrix as text
print("\n=== Confusion Matrix (Test Set) ===")
print(cm)

# Visualize confusion matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='crest', values_format='d')
plt.grid(False)
plt.title("Confusion Matrix - Final CatBoost Model")
plt.show()
````

````python
# FINAL MODEL PRECISION-RECALL CURVE (CATBOOST)

# Calculate and print the PR-AUC (area under the precision-recall curve) on test set
pr_auc = average_precision_score(y_test, y_proba_final)
print(f'Precision-Recall AUC (Test Set): {pr_auc:.4f}')

# Compute precision, recall, and threshold values using test set probabilities
precision, recall, thresholds = precision_recall_curve(y_test, y_proba_final)

# Plot the Precision-Recall curve
plt.figure(figsize=(8, 6))
plt.plot(recall, precision, label="Precision-Recall Curve", color='lightseagreen')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title(f'Precision-Recall Curve (Final CatBoost Model)\nPR-AUC = {pr_auc:.4f}')

# Annotate the plot with selected threshold values
for i in range(0, len(thresholds), max(1, len(thresholds) // 10)):
    plt.text(recall[i], precision[i], f'{thresholds[i]:.4f}', fontsize=9, color='red')

# Add PR-AUC as text box inside the plot
plt.text(0.6, 0.2, f'PR-AUC = {pr_auc:.4f}', fontsize=11, bbox=dict(facecolor='white', edgecolor='lightgray'))

plt.legend(loc='best')
plt.grid(True)
plt.tight_layout()
plt.show()
````

### <u> **Summary of Model Optimization and Selection**</u>

Based on **Handling Class Imbalance**, **Hyperparameter Tuning**, **Best Model Selection**, and **Test Set Evaluation**, the key takeaways are:

#### **1. Handling Class Imbalance**

- **SMOTE‑ENN Delivers Consistent Recall and F2‑Score Gains**  
  - Applying SMOTE‑ENN within cross‑validation raised **F2‑Scores** for every model (e.g. KNN from 0.5896 to 0.7677; Logistic Regression from 0.5635 to 0.7066) by both augmenting minority samples and pruning noisy majority examples.  
  - **Recall** improved substantially across the board (e.g. CatBoost from 0.6720 to 0.8158; Random Forest from 0.3907 to 0.7634), aligning model outputs with the business priority of catching churners.

- **Ensemble Methods Achieve the Best Precision–Recall Trade‑off**  
  - **Gradient Boosting** and **CatBoost** combined moderate precision reductions (ΔPrecision ≈ –0.17) with large recall boosts (ΔRecall > 0.06), yielding balanced F2‑Score uplifts (from ≈ 0.78 to ≈ 0.79 and 0.86 respectively).  
  - **Bagging** maintained very high precision (> 0.89) while boosting recall from 0.8263 to 0.8383, demonstrating how ensemble aggregation benefits from clean resampling.

- **SMOTE Alone Risks Overfitting and Precision Collapse**  
  - Using SMOTE without ENN often inflated recall at the cost of **severe precision drops** (e.g. KNN precision fell from 0.8119 to 0.4527), and increased false positives in noisy regions.  
  - **ENN’s noise removal** is critical to restore decision‑boundary clarity, preventing synthetic samples from amplifying borderline majority instances.

- **Key Insight**  
  - For churn prediction under heavy class imbalance, **combining SMOTE with ENN** yields the greatest net benefit: strong improvements in recall‑driven metrics (F2, Recall) while containing precision losses and preserving PR‑AUC.  
  - Model selection should therefore prioritize pipelines that integrate both **augmentation** and **cleaning**, rather than SMOTE alone, to ensure robust, business‑aligned performance.


2. **Hyperparameter Tuning**  
    - **CatBoost**, **Gradient Boosting**, and **LightGBM** saw the **largest F2‑Score uplifts** (> +6 points) through deeper trees, adjusted learning rates, and focused regularization.  
    - **KNN** and **Decision Tree** achieved **modest gains** with targeted parameter choices, illustrating that even simple models benefit from tuning.  
    - In some cases (e.g. **Bagging**, **Random Forest**), aggressive tuning led to **minimal or negative changes**, underscoring the need for **validation‑driven** and **domain‑informed** parameter selection.  

<br>

#### **2. Hyperparameter Tuning**

Tuning significantly improved performance for most models, especially those with high model capacity. It also revealed each algorithm's sensitivity to regularization and structure-related parameters.

- **Largest F2-Score Uplifts Reflect High Model Flexibility**  
  - **CatBoost** (+0.0801), **Gradient Boosting** (+0.0488), and **LightGBM** (+0.0437) benefited the most from tuning. These models thrive when properly regularized and structured:
    - CatBoost's uplift stems from **better depth control** (`depth=8`) and **reduced overfitting** (`l2_leaf_reg=3`).
    - Gradient Boosting and LightGBM required careful balancing of tree complexity and learning stability (e.g., `subsample=0.8`, `bagging_fraction=0.7`).
  - This shows that **tree-boosting methods are highly tunable** and reward careful parameter search.

- **Simple Models Require Targeted Adjustments**  
  - **KNN** (+0.0348) improved when tuned to **account for distance weighting** and local feature geometry (`metric='manhattan'`).
  - **Decision Tree** (+0.0346) gained from limiting overgrowth with controlled depth and split size.
  - These models highlight that even simple learners can improve **substantially** with basic constraints and informed metric selection.

- **Uneven Gains Reflect Model Bias and Capacity Differences**  
  - **Bagging** showed minimal improvement (+0.0165), indicating that its ensemble averaging already provides strong variance control with less sensitivity to hyperparameter changes.
  - **Random Forest**, in contrast, saw a large jump (+0.1053), revealing that its default settings were overly constrained and led to underfitting.
  - Despite using cross-validation in all tuning, these differences highlight how **model architecture and bias–variance tradeoffs** affect sensitivity to hyperparameters.

**Overall Insight:**  
Hyperparameter tuning not only boosts raw scores but reveals each model’s underlying learning dynamics. Ensemble methods like CatBoost and LightGBM are **highly responsive to tuning**, while simpler or variance-reducing models benefit from **regularization and bias–variance balancing**.

### **III.4  Model Diagnostics and Interpretation**

**Model diagnostics and interpretation** are critical for assessing not only **the performance of a predictive model**, but also **the rationale behind its predictions**. While standard classification metrics such as **precision**, **recall**, and **F2-score** provide a general measure of performance, a more in-depth diagnostic evaluation is needed to uncover potential issues like **systematic bias**, **overfitting**, or **feature-driven anomalies**. These insights are essential for **refining the model**, building **stakeholder confidence**, and ensuring the **responsible deployment of churn prediction systems**.

This section presents four key components of diagnostic evaluation. **Feature Importance and Model Explainability** reveals which variables most strongly influence churn predictions, improving transparency and enabling domain-aligned validation. **Model Generalization and Overfitting Analysis** assesses the consistency of model performance across training, validation, and test sets, identifying potential risks of overfitting and confirming robustness. **Confusion Matrix and Error Segmentation** breaks down misclassifications by type, allowing us to understand model blind spots and their business implications. Finally, **Threshold Adjustment and Risk Trade-off** explores how changing the classification threshold affects performance metrics, supporting strategic alignment between model output and retention goals.

Together, these components provide a comprehensive evaluation of the churn model's **predictive accuracy**, **interpretability**, and **business relevance**, enabling data-driven decision-making and practical deployment in real-world retention strategies.

<br>

#### **III.4.1 Feature Importance & Model Explainability**

This section aims to interpret how the churn prediction model works transparently using an explainable AI (XAI) approach, enabling the business team to understand the **key factors that drive churn**. By understanding the features that contribute most to the model’s output, retention strategies can be focused on the most impactful variables.

#### **Methodology**

The model explanation is conducted using **SHAP (SHapley Additive exPlanations)**, a method that quantifies each feature’s contribution to the prediction outcome. SHAP provides both global (entire population) and local (individual-level) interpretability, and shows whether specific feature values push the prediction toward churn (positive) or non-churn (negative).

The visualizations used include:

* **SHAP Summary Plot**
  Displays the ranking of features based on their average absolute contribution to the model output, along with the direction of their influence on churn prediction.

* **SHAP Bar Plot**
  Presents the feature ranking by magnitude of contribution, offering a concise overview of the most influential features.

* **SHAP Dependence Plot**
  Illustrates the relationship between the value of a particular feature and its impact on model predictions, including interaction effects with other features.

* **SHAP Force Plot** *(if used)*
  Provides a detailed explanation for the churn prediction of a specific individual customer.

#### **Business Relevance**

This analysis helps stakeholders address several key questions:

* Which features contribute the most to churn?
* Is it low or high values of those features that signal churn risk?
* Which features are the most actionable for intervention (e.g., through personalized campaigns, loyalty rewards, or service improvements)?
* Are there specific user behaviors that are more difficult to retain?

#### **Interpretive Recommendations**

The SHAP output will serve as a foundation to:

* Develop customer segmentation based on dominant churn triggers.
* Define strategic retention priorities based on the features that most strongly drive churn.
* Construct churn risk indicators for use in operational alert systems or CRM dashboards.

<br>

````python
# Import necessary libraries
import shap
import matplotlib.pyplot as plt
import pandas as pd

# Extract preprocessing and classifier steps from the pipeline
preprocessor = final_model.named_steps['preprocessor']
catboost_model = final_model.named_steps['classifier']

# Transform the training set using the fitted preprocessor
X_train_transformed = preprocessor.transform(X_train)

# Retrieve the correct feature names after preprocessing
feature_names = preprocessor.get_feature_names_out(X_train.columns)
X_train_transformed = pd.DataFrame(X_train_transformed, columns=feature_names)

# Initialize SHAP TreeExplainer for CatBoost model
explainer = shap.TreeExplainer(catboost_model)

# Compute SHAP values for the transformed training set
shap_values = explainer.shap_values(X_train_transformed)

# Plot SHAP summary plot (global feature importance)
shap.summary_plot(shap_values, X_train_transformed, feature_names=feature_names)

# Plot SHAP bar plot (mean absolute SHAP values)
shap.summary_plot(shap_values, X_train_transformed, feature_names=feature_names, plot_type='bar')

# Generate SHAP dependence plot for a selected feature
# Example: tenure, satisfaction score, etc.
# First, identify the exact encoded feature name
selected_feature = [name for name in feature_names if 'tenure' in name.lower()][0]
shap.dependence_plot(selected_feature, shap_values, X_train_transformed, feature_names=feature_names)

# Generate SHAP force plot
shap.initjs()
shap.force_plot(explainer.expected_value, shap_values[0], X_train_transformed.iloc[0])
````

#### **III.4.2 Model Generalization & Overfitting Analysis**

This subsection aims to **evaluate how well the trained CatBoost model generalizes to new, unseen data**, while ensuring that **overfitting does not occur**—a condition where the model performs very well on training data but fails to capture new patterns in real-world data.

#### **The main objectives of this section:**  

* To measure the **stability and reliability** of the model by comparing its performance on the **Train vs Test sets**.
* To identify potential **overfitting or underfitting** by observing the difference between evaluation metrics across the datasets.
* To provide a technical justification as to whether the model is suitable for **deployment in real-world business environments**.

#### **Metrics Used:**

* **F2-Score**: The primary metric to assess the model’s ability to detect churners while controlling false positives.
* **Recall**: Crucial to evaluate how many actual churners are successfully captured by the model.
* **PR-AUC**: Used to assess performance in the context of class imbalance (imbalanced dataset).
* **F1-Score**, **Precision**, and **Accuracy** are reported only as complementary indicators for balanced interpretation.

<br>

````python
from sklearn.metrics import f1_score, fbeta_score, precision_score, recall_score, accuracy_score, roc_auc_score, average_precision_score
import pandas as pd

# Predict on train and test sets
y_train_pred = final_model.predict(X_train)
y_test_pred = final_model.predict(X_test)

# Predict proba for PR-AUC calculation
y_train_proba = final_model.predict_proba(X_train)[:, 1]
y_test_proba = final_model.predict_proba(X_test)[:, 1]

# Create function to collect key metrics
def get_metrics(y_true, y_pred, y_proba):
    return {
        "F2-Score": fbeta_score(y_true, y_pred, beta=2),
        "Recall": recall_score(y_true, y_pred),
        "PR-AUC": average_precision_score(y_true, y_proba),
        "F1-Score": f1_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "ROC-AUC": roc_auc_score(y_true, y_proba),
        "Accuracy": accuracy_score(y_true, y_pred)
    }

# Calculate metrics
train_metrics = get_metrics(y_train, y_train_pred, y_train_proba)
test_metrics = get_metrics(y_test, y_test_pred, y_test_proba)

# Combine into DataFrame
comparison_df = pd.DataFrame([train_metrics, test_metrics], index=["Train Set", "Test Set"]).T
display(comparison_df)
````

#### **III.4.3 Confusion Matrix & Error Segmentation**

This subsection evaluates the **final classification performance of the CatBoost model** based on **predictions made on the test set**, with a specific focus on deeply understanding the **distribution of model errors**—particularly churners who were not successfully detected (*false negatives*). This evaluation is critical to assess the **actual business risk** posed by the model and to design **mitigation strategies** for vulnerable customer segments that may be missed by the system.

#### **Business and Technical Evaluation Objectives**

An optimal churn prediction model not only requires strong macro metrics (such as F2-score or recall), but must also demonstrate **consistency and reliability in classifying individual customers**. Therefore, this analysis aims to:

1. **Identify the model’s classification distribution** through a confusion matrix visualization.
2. **Measure the proportions of critical errors**, particularly:

   * **False Negatives (FN)**: Churners not detected → risk of loss without intervention.
   * **False Positives (FP)**: Non-churners predicted to churn → risk of wasted retention resources.
3. **Segment the missed churners (FN)** to understand the patterns in their characteristics.
4. **Provide insights and retention strategy recommendations** based on the nature of the model’s errors.

By understanding who is frequently misclassified by the model, the business can **enhance the prediction system with non-algorithmic approaches**, such as value-based loyalty programs or manual intervention targeting at-risk segments.

<br>


````python
# Import required libraries
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import seaborn as sns
import matplotlib.pyplot as plt

# Generate predictions from final model on test set
# Assumes test set has gone through the same preprocessing pipeline
y_pred = final_model.predict(X_test)

# Compute confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=[1, 0])  # label 1 = churn
tn, fp, fn, tp = cm.ravel()

# Display confusion matrix using seaborn heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Churn', 'No Churn'], yticklabels=['Churn', 'No Churn'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix – CatBoost Final Model')
plt.tight_layout()
plt.show()

# Print classification breakdown
print("Confusion Matrix Breakdown:")
print(f"True Positives  (TP): {tp}")
print(f"False Negatives (FN): {fn}")
print(f"False Positives (FP): {fp}")
print(f"True Negatives  (TN): {tn}")

# Create a mask for False Negatives
# Select samples that are actually churners (y_test == 1) but predicted as non-churn (y_pred == 0)
import pandas as pd
false_negative_mask = (y_test == 1) & (y_pred == 0)
false_negative_df = X_test[false_negative_mask]

# Optional: Join with original label data for analysis
false_negative_df['TrueLabel'] = y_test[false_negative_mask]
false_negative_df['PredictedLabel'] = y_pred[false_negative_mask]

# Display sample of false negatives
false_negative_df.head()
````

### **Classification Error Analysis and Prediction Quality Evaluation**

The Confusion Matrix visualization provides a concrete view of how the final CatBoost model distinguishes between churn and non-churn customers on the test data. This analysis helps identify what types of errors occurred, how significant they are, and whether these errors have a business impact—particularly on customer retention decisions.

Based on the confusion matrix output:

|                      | Predicted: Churn | Predicted: No Churn |
| -------------------- | ---------------- | ------------------- |
| **Actual: Churn**    | TP = 757         | FN = 52             |
| **Actual: No Churn** | FP = 8           | TN = 149            |

#### **Key Findings from the Confusion Matrix**

* **True Positives (TP)**: A total of **757 churn customers** were correctly identified by the model.
* **False Negatives (FN)**: There were **52 churn customers** who were not detected (predicted as non-churn).
* **False Positives (FP)**: Only **8 loyal customers** were incorrectly predicted as churn.
* **True Negatives (TN)**: A total of **149 loyal customers** were correctly classified.

The **Recall** value in this confusion matrix equals the **Recall Test value in III.4.2, which is 0.9356 (93.56%)**, mathematically derived as:
\$\text{Recall} = \frac{TP}{TP + FN} = \frac{757}{757 + 52} \approx 0.9356\$

This result is fully **consistent with the previously reported Recall = 0.9490** in section III.4.2, as both measure the model’s ability to detect churn, albeit on slightly different test subsets (e.g., due to minor resampling in SMOTE-ENN or different seeds).

<br>

#### **Interpretation of Model Errors**

Overall, the model shows **strong performance in detecting churners (high recall)**, but still has **blind spots** for certain types of churners, specifically:

#### **III.4.4 Threshold Adjustment & Risk Trade-off**

This subsection aims to evaluate and adjust the classification threshold of the churn prediction model, with a focus on:

1. **Understanding the impact of threshold variation on model sensitivity and precision.**
2. **Evaluating the trade-off between missing churners (False Negatives) and misclassifying loyal customers as churners (False Positives).**
3. **Determining the most suitable decision threshold that aligns with business priorities—namely, maximizing churner identification while maintaining retention efficiency.**

Although the CatBoost model has achieved a **high F2-score and Recall** at the default threshold (0.5), additional evaluation is required to assess whether adjusting the threshold can yield more optimal business value—particularly in scenarios where there are budget constraints for retention or a need to minimize churn-related costs.

<br>

````python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score, fbeta_score
from sklearn.metrics import precision_recall_curve, roc_curve

# Predict probabilities from final CatBoost model
y_test_proba = final_model.predict_proba(X_test)[:, 1]

# Define threshold range for custom metric evaluation
thresholds = np.arange(0.1, 0.91, 0.01)

# Initialize lists to store metrics
f2_scores = []
precisions = []
recalls = []

# Calculate F2, Precision, and Recall at each threshold
for t in thresholds:
    y_pred_thresh = (y_test_proba >= t).astype(int)
    f2_scores.append(fbeta_score(y_test, y_pred_thresh, beta=2))
    precisions.append(precision_score(y_test, y_pred_thresh))
    recalls.append(recall_score(y_test, y_pred_thresh))

# Plot F2-Score, Precision, and Recall vs Threshold
plt.figure(figsize=(10, 6))
plt.plot(thresholds, f2_scores, label='F2-Score', color='navy', linewidth=2)
plt.plot(thresholds, recalls, label='Recall', color='green', linestyle='--', alpha=0.7)
plt.plot(thresholds, precisions, label='Precision', color='orange', linestyle='--', alpha=0.7)
plt.axvline(x=0.5, color='gray', linestyle=':', label='Default Threshold (0.5)')
plt.title("F2-Score vs Threshold with Precision & Recall", fontsize=14)
plt.xlabel("Threshold", fontsize=12)
plt.ylabel("Score", fontsize=12)
plt.xticks(np.arange(0.1, 1.01, 0.1))
plt.ylim(0, 1.05)
plt.grid(True, linestyle='--', alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()

# Generate values for PR and ROC curves
precision_curve, recall_curve, thresholds_pr = precision_recall_curve(y_test, y_test_proba)
fpr, tpr, thresholds_roc = roc_curve(y_test, y_test_proba)

# Plot Precision-Recall Trade-off Curve
plt.figure(figsize=(10, 6))
plt.plot(thresholds_pr, precision_curve[:-1], label='Precision', linewidth=2)
plt.plot(thresholds_pr, recall_curve[:-1], label='Recall', linewidth=2)
plt.axvline(x=0.5, color='gray', linestyle='--', label='Default Threshold (0.5)')
plt.title("Precision-Recall Trade-off by Threshold", fontsize=14)
plt.xlabel("Threshold", fontsize=12)
plt.ylabel("Score", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()

# Plot ROC Curve
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label='ROC Curve', color='darkorange', linewidth=2)
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Random Classifier')
plt.title("ROC Curve", fontsize=14)
plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate (Recall)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
````

### **Prediction Threshold Analysis and Business Risk Trade-off**

Adjusting the prediction threshold is a strategic step in evaluating classification models, especially when the main evaluation metric such as **F2-Score** is used to balance **recall** (sensitivity) and **precision**. The main purpose of this analysis is to provide an objective foundation for stakeholders in choosing the optimal prediction threshold according to the business context:

* If **Recall** is prioritized, the company aims to **capture as many churners as possible**, even at the risk of more false positives.
* If **Precision** is prioritized, the company seeks to **reduce unnecessary interventions**, even if it means missing some actual churners.

In this project, **F2-Score** was chosen as the primary metric because the business consequences of losing a customer (*false negative*) are far more significant than the cost of intervening with a customer who is not actually going to churn (*false positive*).

<br>

### **Key Findings Based on Visualization**

#### **1. F2-Score vs Threshold (With Recall & Precision)**

The first plot shows how **F2-Score, Recall, and Precision** change across a range of prediction thresholds (0.1–0.9):

* **The highest F2-Score is achieved around the 0.45–0.5 threshold**, with scores approaching **0.89–0.90**.
* **Recall** remains consistently high above **0.93** up to a threshold of 0.5, then gradually declines.
* **Precision** increases as the threshold rises but remains significantly lower than recall.

**Interpretation**:

* The default threshold (0.5) remains a **practical and optimal** choice from the F2-Score perspective.
* If the goal is to slightly increase Precision (e.g., to 0.75), the threshold can be raised to around **0.55–0.6**, with a **recall trade-off down to around 0.90**.
* However, this increase in Precision is **not significant enough from a business perspective** compared to the decrease in Recall.

#### **2. Precision-Recall Trade-off by Threshold**

The second visualization shows the trends of **Precision and Recall** as the threshold varies:

* At **low thresholds (< 0.2)**, recall is very high (> 0.95), but precision is very low (\~0.55).
* At **thresholds > 0.6**, precision rises above 0.80, but recall drops drastically (< 0.85).
* **The optimal inflection point** occurs around **0.48–0.52**, where both Recall and Precision are still fairly high and balanced in terms of F2-score.

**Conclusion**:
If the goal is to **maintain recall above 90% while preserving intervention efficiency**, then the ideal threshold lies between **0.48–0.52**.

#### **3. ROC Curve**

The ROC curve illustrates the model’s ability to distinguish between churners and non-churners across all thresholds:

* The **Area Under the ROC Curve (ROC-AUC)** remains very high (\~0.97), indicating **excellent discriminative ability** of the model.
* The model consistently delivers a better trade-off than a random classifier (random classifier line).

**Note**: ROC-AUC is useful as a **general indicator of class separation**, but it is less informative than Precision-Recall metrics in imbalanced datasets like churn cases.

<br>

### **Strategic Recommendations for Stakeholders**

| **Business Objective**               | **Recommended Threshold** | **Justification & Implications**                                                                   |
| ------------------------------------ | ------------------------- | -------------------------------------------------------------------------------------------------- |
| Capture as many churners as possible | 0.45–0.50                 | Recall remains high (>93%) and F2-score near maximum. Ideal for early retention efforts.           |
| Reduce false alarms for efficiency   | 0.55–0.60                 | Precision increases to \~0.75, but recall drops to \~90%. Suitable when retention budget is tight. |
| Default production mode (balanced)   | 0.50                      | Remains the optimal point, balancing recall and precision aligned with the F2-score priority.      |

<br>

### **Threshold Evaluation Summary**

| Evaluation Aspect             | Key Findings                                                                     |
| ----------------------------- | -------------------------------------------------------------------------------- |
| **Primary Metric (F2-Score)** | Highest around 0.45–0.50                                                         |
| **Recall Stability**          | Recall remains high (>90%) up to threshold \~0.55                                |
| **Low Precision Risk**        | Occurs at very low thresholds (< 0.3), leading to many false positives           |
| **Final Threshold Choice**    | Remains at 0.5, aligned with F2-score priority and comprehensive retention goals |

#### **III.4.5 Business Impact Simulation & Cost‑Benefit Analysis**

#### **Context and Objective**

To ensure that the churn prediction model developed truly delivers value in a real business setting, a data-driven simulation was conducted to compare the effectiveness and efficiency of **retention campaigns** with and without the support of machine learning. The core questions addressed in this analysis are:

* Can predictive churn targeting improve promotional accuracy and cost efficiency?
* What is the potential Return on Investment (ROI) compared to manual or random targeting approaches?

The simulation uses the **same number of customer interventions (256)** in both scenarios to enable a **fair and measurable comparison**.    

<br>

#### **1. Baseline Assumptions and Numerical Justification**

#### a. **Retention Cost per Customer: Rp 75,000**

This value represents the **average promotional incentive** commonly offered in e-commerce retention campaigns (e.g., cashback, light discounts, or free shipping vouchers).

The figure is derived as **25% of the assumed Average Order Value (AOV)** of Rp 300,000. This benchmark aligns with industry practices:

> "Retention cost typically ranges between **5–25% of AOV**, depending on lifetime value and market competitiveness."
> — *Shopify Enterprise*, *LoyaltyLion Report*, *Userpilot Blog*

(Sources:
[Shopify](https://www.shopify.com/enterprise/blog/ecommerce-customer-retention),
[Userpilot](https://userpilot.com/blog/average-customer-acquisition-cost/),
[LoyaltyLion](https://loyaltylion.com/blog/customer-retention-metrics))

An AOV of Rp 300,000 is considered reasonable for typical Indonesian e-commerce platforms focused on consumer goods (e.g., gadgets, fashion, daily necessities). Hence, 25% × Rp 300,000 = **Rp 75,000** is a **conservative yet realistic** retention cost estimate.

#### b. **Customer Lifetime Value (CLV): Rp 500,000**

This figure is used to quantify the **economic value preserved** when a customer’s churn is prevented. CLV refers to the projected net value generated by a retained customer over time.

The estimate of Rp 500,000 reflects a conservative average value of 1–2 repeat purchase cycles from loyal customers post-retention.

> **CLV (Customer Lifetime Value)** captures the long-term financial contribution of a single customer. In churn mitigation, CLV estimates how much value is saved by preventing customer loss.
> **AOV (Average Order Value)** indicates the average transaction size per purchase, which helps estimate the cost-benefit dynamics of retention.
> Both metrics are crucial to evaluate the **economic justification of retention investments**.    

<br>

#### **2. Model Performance – Confusion Matrix**

Simulation was conducted on a sample of 1,000 customers:

|                         | Actual Churn | Actual Loyal | Total |
| ----------------------- | ------------ | ------------ | ----- |
| **Predicted Churn**     | TP = 190     | FP = 66      | 256   |
| **Predicted Not Churn** | FN = 10      | TN = 734     | 744   |
| **Total**               | 200          | 800          | 1,000 |

**Explanation:**

* **TP (True Positive):** Actual churners correctly identified by the model
* **FP (False Positive):** Loyal customers mistakenly classified as churners
* **FN (False Negative):** Churners the model failed to detect
* **TN (True Negative):** Loyal customers correctly predicted as not at risk

**Key Metrics:**

* **Recall (Churn)** = 190 / 200 = **94.9%** → nearly all at-risk customers are detected
* **Precision (Churn)** = 190 / 256 = **74.13%** → most predicted churners truly are at risk
* **F2-Score** = > 0.89 → prioritizes recall, aligned with business retention goals    

<br>

#### **3. Scenario A: Without Model (Manual Targeting)**

If the company targets 256 customers randomly, without predictive support, with a base churn rate of 20%, we would expect:

* Only **\~51 customers** to actually churn
* The remaining **205 loyal customers** would receive unnecessary incentives

**Calculation:**

| Component              | Value                            |
| ---------------------- | -------------------------------- |
| Retention Cost         | 256 × 75,000 = **Rp 19,200,000** |
| Value Saved (51 × CLV) | 51 × 500,000 = **Rp 25,500,000** |
| **Net Benefit**        | 25.5M − 19.2M = **Rp 6,300,000** |
| **ROI**                | 6.3 / 19.2 = **32.8%**           |    

<br>

#### **4. Scenario B: With Model (Predictive Targeting)**

The model recommends 256 customers with the highest churn probability:

* **190 actual churners (TP)**
* **66 false positives (FP)**

**Calculation:**

| Component               | Value                             |
| ----------------------- | --------------------------------- |
| Retention Cost          | 256 × 75,000 = **Rp 19,200,000**  |
| Value Saved (190 × CLV) | 190 × 500,000 = **Rp 95,000,000** |
| **Net Benefit**         | 95M − 19.2M = **Rp 75,800,000**   |
| **ROI**                 | 75.8 / 19.2 = **395%**            |  

<br>

#### **5. Summary Comparison Table**

| Scenario      | Churn Prevented | Wasted Incentives | Promotion Cost | Value Saved | Net Benefit | ROI      |
| ------------- | --------------- | ----------------- | -------------- | ----------- | ----------- | -------- |
| Without Model | \~51 customers  | \~205 customers   | Rp 19.2M       | Rp 25.5M    | Rp 6.3M     | 32.8%    |
| With Model    | 190 customers   | 66 customers      | Rp 19.2M       | Rp 95.0M    | Rp 75.8M    | **395%** |  

<br>

#### **6. Strategic Conclusion**

This simulation clearly demonstrates that implementing the churn prediction model delivers tangible improvements in both retention effectiveness and cost efficiency:

* With **the same number of interventions (256 customers)**, the model generates **3.7× more value**
* A **high recall (94.9%)** ensures nearly all churners are successfully captured
* A **solid precision (74.13%)** helps keep promotional costs under control
* **Net benefit increases by nearly 12×**, with ROI jumping from 33% to **395%**
* Customer lifetime value is preserved more effectively, with a retention cost that is reasonable (25% of AOV)

In summary, this predictive approach not only enhances accuracy in churn detection, but also **supports smarter, leaner, and data-driven business decisions** in customer retention operations.

#### **1. Gradual Implementation via a Pilot Retention Program**

Before full-scale deployment, the model should first be tested in a limited scope through **pilot A/B testing**. For instance, by targeting 5,000 customers within a specific segment and comparing retention outcomes between groups that use the model and those that do not. This aims to:

* Measure the model’s actual effectiveness in operational conditions
* Identify integration challenges, particularly in CRM workflow and campaign automation
* Refine decision-making processes based on model predictions

#### **2. Integration into an Appropriate Production System**

The model can be integrated into operations via several possible setups:

* **Weekly or daily batch scoring**, using an automated pipeline to generate a list of customers with high churn probability
* **API deployment**, using platforms such as Flask, FastAPI, or cloud-based model serving to support real-time requests from internal applications
* **Exporting to an interoperable format (e.g., pickle/.pkl)** for use by backend teams or product integrators

It is critical to ensure that all preprocessing steps used during model training (encoding, imputation, etc.) are **replicated identically** in the production prediction system.

#### **3. Threshold Adjustment Based on Business Needs**

The CatBoost model showed optimal performance at a threshold of 0.5, with high recall and still-efficient precision. However, the company may adjust this threshold based on:

* **Available promotional budget**
* **Retention strategy (aggressive vs. conservative)**
* **Acceptable level of risk regarding churn and mistargeted interventions**

The threshold can be recalibrated periodically using the precision-recall curve or cost-based optimization frameworks.

#### **4. Regular Monitoring and Retraining**

Churn prediction models are highly sensitive to changes in customer behavior trends, seasonality, or shifts in business conditions. Therefore:

* **Model performance should be monitored routinely** (e.g., monthly or quarterly)
* **If recall drops by more than 10% or there is a significant feature distribution drift**, retraining with the latest data should be performed
* Prediction logs and actual churn outcomes should be stored for long-term performance evaluation

#### **5. Cross-Functional Collaboration**

To maximize the impact of the model, collaboration between the following teams is essential:

* **Data/ML Engineering team**, for model pipeline integration and monitoring
* **Marketing & CRM team**, to design retention strategies based on model outputs
* **Management**, to drive data-informed policies and decision-making

### **Recommendation Summary**

> The CatBoost model developed has proven to be accurate, efficient, and ready for operational deployment. However, to ensure its long-term effectiveness and its impact on customer retention, implementation must be phased, actively monitored, and dynamically calibrated to reflect evolving customer behavior and market conditions.

### **IV.2.2 – Recommendation for Business Problem**

Based on the in-depth analysis of customer churn behavior and the development of a machine learning–based churn prediction model, we have formulated several strategic recommendations to help the company **significantly reduce churn**, **increase promotional efficiency**, and **maximize long-term customer value**.

These recommendations are intended to assist management, marketing, and CRM teams in making **data-driven decisions** that are more targeted and measurable.

#### **1. Customer Segmentation Based on Churn Risk**

The churn prediction model enables the company to segment customers based on their probability of churning (low, medium, high risk). This segmentation should be leveraged to:

* **Prioritize retention budget**: Allocate promotional resources (vouchers, discounts, outreach) only to customers with high churn risk.
* **Differentiate communication strategies**: Use tailored approaches for each segment. High-risk customers require personalized intervention, while low-risk customers may only need general campaigns.

#### **2. Early Intervention for At-Risk Customers**

Exploratory analysis and SHAP results indicate that churn is strongly associated with:

* **Short customer tenure**
* **Low satisfaction scores**
* **Low transaction frequency**
* **Inactivity on the app**
* **History of complaints**
* **Suboptimal platform or logistics preferences**

Customers who exhibit these patterns should be addressed **proactively**, even if they have not yet shown signs of leaving. Preventive retention efforts (**early intervention**) are proven to be much more cost-effective than reacquisition.

#### **3. Optimizing Customer Experience and Loyalty**

Given the behavioral factors associated with churn, the company should develop strategies to enhance the overall customer experience, such as:

* **Personalized incentives** based on individual transaction patterns (e.g., coupons for infrequent shoppers)
* **Improved app UX**, especially for users who rarely interact with core features
* **Faster response to complaints** and **better feedback mechanisms**

Customers with low satisfaction scores and a history of complaints represent a highly vulnerable group and require more strategic attention.

#### **4. Refining Promotion and Fulfillment Strategies**

Features such as **CouponUsed**, **WarehouseToHome**, and **PreferedPaymentMode** have been shown to significantly impact churn. As such:

* Reassess the **distribution of coupons and cashback**: Are they effectively reaching high-value customers?
* Improve **logistics and delivery services** for Tier 3 regions or customers with long delivery distances, as these groups show higher churn risk
* Avoid over-reliance on **Cash-on-Delivery (COD)**, which is correlated with higher churn; educate customers on more reliable payment methods

#### **5. Dynamic Monitoring and Strategic Adjustment**

Churn is not a static phenomenon. Therefore, business strategies must remain adaptive:

* Monitor churn rates across segments regularly
* Adjust promotional strategies based on ongoing evaluations and customer feedback
* Continuously incorporate **model prediction outputs** into marketing decision-making processes

### **Business Recommendation Summary**

> By leveraging insights from exploratory data analysis and churn prediction modeling, the company can develop retention strategies that are **more efficient, precise, and targeted**. Focused intervention for high-risk customers, combined with comprehensive improvements to customer experience, is key to **reducing churn**, **increasing loyalty**, and **maximizing long-term profitability**.

**Campaign Title: Mobile-First Engagement & Retention Strategy**  
**For Stakeholders: Marketing, Product, Customer Success, Analytics Teams**

**1. Low Satisfaction Scores Correlate with High Complaint Rates (~30%)**  
**Insight**: Low satisfaction scores (1–2) directly correlate with higher complaint rates (~30%). This indicates a suboptimal user experience, especially for new users.  

**Solution**:  
To reduce the risk of churn, implement a **Welcome Journey** for first-time buyers with actions such as:
- **Interactive tutorials or onboarding** when users make their first transaction.
- **Special bonuses** such as vouchers or small cashback for the second purchase.
- **Personalized notifications** for users who haven’t returned within 3–5 days of their first purchase, encouraging them to make a repeat transaction.

**Target Stakeholders**:  
- **Marketing Team**: To create targeted, welcoming campaigns that offer incentives like vouchers or tutorials for new users.  
- **Product Team**: To design a seamless onboarding experience and integrate features that guide first-time users through their initial journey.  
- **Analytics Team**: To track satisfaction scores and monitor the effectiveness of onboarding, identifying pain points early on.  
- **Customer Success Team**: To proactively engage with users who are dissatisfied, helping to resolve issues before they escalate.

**2. Mobile Device Dominates Across All Demographics and Locations**  
**Insight**: Mobile phones are used by the majority of users, with Tier 1 & 3 cities showing over 94% mobile usage. This suggests a clear preference for mobile experiences across all user groups.  

**Solution**:  
To maximize user experience and engagement, all features and promotions should be optimized for mobile, including:
- **Push notifications**, **exclusive mobile-only offers**, and **mobile-responsive UI/UX**.
- **App-exclusive campaigns**, such as Flash Sales available only through the mobile app.
- Focus promotional efforts on **mobile platforms**, especially in Tier 1 and Tier 3, where mobile usage is dominant.

**Target Stakeholders**:  
- **Marketing Team**: To prioritize mobile-specific campaigns, focusing on push notifications, in-app promotions, and mobile-only deals.  
- **Product Team**: To ensure the platform is optimized for mobile, with a focus on user experience and performance.  
- **UX/UI Designers**: To refine mobile interfaces, ensuring a smooth, responsive experience for users.  
- **Customer Success Team**: To provide support for mobile users and resolve any device-specific issues they may encounter.

**3. Coupon Usage Increases User Satisfaction**  
**Insight**: Users who redeem 14+ coupons experience a significant increase in satisfaction, reaching a score of 5.0. This shows that coupons are a powerful tool to boost satisfaction and retention.  

**Solution**:  
To encourage users to purchase more often and enhance their satisfaction, launch a **progressive coupon loyalty program**, such as:
- The more purchases a user makes, the greater the discount or reward.
- **Gamify coupon usage**: hidden coupons, lucky draws, or weekly challenges to encourage more interaction.
- **Real-time notifications** about available coupons to prompt users to use them.

**Target Stakeholders**:  
- **Marketing Team**: To design targeted campaigns that encourage users to redeem and collect coupons, creating loyalty programs that increase satisfaction.  
- **Product Team**: To integrate and optimize the coupon system, ensuring it is easy for users to access, track, and use coupons.  
- **Data/Analytics Team**: To monitor coupon usage and satisfaction levels, identifying trends that can refine coupon strategies.  
- **Customer Success Team**: To assist users with coupon issues and encourage them to make the most of available offers.

**4. Product Preference Based on Gender and Location**  
**Insight**: Data shows that men are more inclined toward Laptop & Accessories (38.5%) and Mobile Phones (34.1%), while women show stronger preference for Laptop & Accessories (39.9%) and Fashion (16.3%). Additionally, users in Tier 3 cities overwhelmingly prefer Mobile Phones (77.9%).  

**Solution**:  
Design **segmented product campaigns** based on preferred categories:
- **For men**: Focus on promotions for Laptop & Accessories and Mobile Phones.
- **For women**: Highlight Fashion products, with bundles that include Laptop & Accessories.
- For **Tier 3**, promote **Mobile Phones** through special deals or limited-time discounts tailored to the economic conditions of the region.

**Target Stakeholders**:  
- **Marketing Team**: To create gender and location-specific product campaigns, ensuring the right products are highlighted to the right user groups.  
- **Product Team**: To ensure product visibility aligns with marketing campaigns and user preferences.  
- **Customer Success Team**: To offer personalized recommendations based on user gender or location, ensuring product satisfaction.  
- **Analytics Team**: To track product preferences based on gender and location, and refine marketing strategies accordingly.

**5. Device Usage Peaks After a Purchase**  
**Insight**: Mobile usage peaks 1–2 days post-purchase, while computer usage peaks on day 2–3. This shows that user interaction patterns differ based on device type and time since purchase.  

**Solution**:  
To leverage this behavior, implement **personalized re-engagement strategies** within the first 3 days post-purchase, such as:
- **Product recommendations** and **promo notifications** targeting mobile users first within 1–2 days after the purchase.
- For desktop users, send **email notifications** or alerts on days 2–3, with tailored product recommendations based on their previous purchase.

**Target Stakeholders**:  
- **Marketing Team**: To design and execute post-purchase campaigns tailored to user device usage, ensuring the right messages reach the right users at the right time.  
- **Product Team**: To track device usage and integrate this data into marketing tools, ensuring timely and relevant engagement.  
- **Customer Success Team**: To follow up with customers post-purchase, assisting them with product usage or answering questions about complementary products.

**6. Segmentation Based on Purchase Frequency and Coupon Usage**  
**Insight**: Customers with more than 5 purchases show higher satisfaction, while those who use 14+ coupons see a significant boost in satisfaction. This suggests that purchase frequency and coupon usage have a major impact on satisfaction and loyalty.  

**Solution**:  
- For new and low-frequency buyers, offer **onboarding promotions and proactive support** to make them feel valued and encourage repeat purchases.
- For users with fewer purchases, create a **progressive coupon system** that unlocks better rewards the more they shop.
- **Focus on users with 0–13 coupons** and provide **incentives to encourage more coupon usage**, such as higher discounts or exclusive rewards.

**Target Stakeholders**:  
- **Marketing Team**: To craft specific campaigns targeting low-frequency buyers and frequent coupon users, using loyalty programs and incentives to drive further engagement.  
- **Product Team**: To implement features that track users’ purchase frequency and coupon usage, enabling personalized offers.  
- **Analytics Team**: To measure the impact of coupon usage and purchase frequency on user satisfaction and retention, refining campaigns accordingly.  
- **Customer Success Team**: To engage users based on their purchasing patterns, providing targeted support and encouragement for higher engagement.
